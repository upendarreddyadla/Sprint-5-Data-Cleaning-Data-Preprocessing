# Notebook 4 – Duplicate Data Handling

### What are Duplicate Records?


Duplicate Records are rows in a dataset that contain the same information more than once

In simple words, when the same record appears multiple times in a dataset, those records are called duplicate records

Duplicate records can occur because of:

- Repeated data entry.
- Multiple file uploads.
- Data import errors.
- System synchronization issues.
- Repeated customer registrations.

```python
Example:

Consider a customer dataset:

Customer_ID	Name	Age	City

  101	    Ravi	25	Hyderabad
  102	    Kumar	30	Chennai
  103	    Priya	28	Bangalore
  101	    Ravi	25	Hyderabad
  104	    Suresh	32	Mumbai
  102	    Kumar	30	Chennai

Here:

- Customer 101 appears twice.
- Customer 102 appears twice.
- These repeated rows are duplicate records.
```

##### Mathematical Concept

Suppose a dataset contains 1,000 records.

- Total Records = 1,000
- Duplicate Records = 100
- Unique Records = 900

##### Unique Records Formula

$$
\text{Unique Records} =
\text{Total Records} - \text{Duplicate Records}
$$

$$
=1000-100=900
$$

##### Duplicate Percentage

$$
\text{Duplicate Percentage} =
\frac{\text{Duplicate Records}}{\text{Total Records}}
\times 100
$$

$$
=\frac{100}{1000}\times100=10\%
$$

Real-World Business Example:

Suppose an e-commerce company maintains customer information.

A customer registers twice using the same details.

As a result:

- The company counts the same customer twice.
- Customer statistics become incorrect.
- Marketing campaigns may send duplicate emails.
- Business analysis may produce misleading results.
  
Removing inappropriate duplicate records helps maintain data quality.

```python
import pandas as pd

# Create sample dataset
data = {
    "Customer_ID": [101, 102, 103, 101, 104, 102],
    "Name": ["Ravi", "Kumar", "Priya", "Ravi", "Suresh", "Kumar"],
    "Age": [25, 30, 28, 25, 32, 30],
    "City": ["Hyderabad", "Chennai", "Bangalore",
             "Hyderabad", "Mumbai", "Chennai"]
}

df = pd.DataFrame(data)

# Display dataset
print("Original Dataset:")
print(df)

# Identify duplicate records
print("\nDuplicate Records:")
print(df[df.duplicated()])

# Count duplicate records
print("\nNumber of Duplicate Records:")
print(df.duplicated().sum())

# Remove duplicate records
df_cleaned = df.drop_duplicates()

print("\nDataset After Removing Duplicates:")
print(df_cleaned)

# Count remaining records
print("\nTotal Unique Records:")
print(len(df_cleaned))
```

### Exact Duplicates


Exact Duplicates are records in a dataset where all column values are exactly the same as another record

In simple words, when two or more rows contain identical information across every column, they are called exact duplicates

```python
Example:

Consider a customer dataset:

Customer_ID	Name	Age	City
    101	    Ravi	25	Hyderabad
    102	    Kumar	30	Chennai
    103	    Priya	28	Bangalore
    101	    Ravi	25	Hyderabad
    104	    Suresh	32	Mumbai

Here:

- The first and fourth rows contain exactly the same values.
- Customer_ID, Name, Age, and City are identical.
- Therefore, the fourth row is an exact duplicate.
```


##### Mathematical Concept

Suppose a dataset contains \(N\) records.

If two records have identical values across all columns:

$$
R_i = R_j
$$

Where:

- \(R_i\) = First record.
- \(R_j\) = Second record.
- \(=\) means all column values are identical.

##### Example

$$
R_1 = (101, \text{Ravi}, 25, \text{Hyderabad})
$$

$$
R_2 = (101, \text{Ravi}, 25, \text{Hyderabad})
$$

Therefore:

$$
R_1 = R_2
$$

Both records are exact duplicates.


Real-World Business Example:

Suppose an e-commerce company stores customer registration information.

Due to a system error, the same customer information is inserted twice.

As a result:

- The customer count increases incorrectly
- Duplicate marketing emails may be sent
- Sales and customer reports may become inaccurate
- Data quality decreases
  
Removing unnecessary exact duplicates helps maintain accurate customer information

Importance in Data Preprocessing

- Improves data quality.
- Prevents repeated records.
- Avoids incorrect customer counts.
- Reduces unnecessary data.
- Helps prepare reliable datasets for machine learning.

```python
import pandas as pd

# Create customer dataset
data = {
    "Customer_ID": [101, 102, 103, 101, 104],
    "Name": ["Ravi", "Kumar", "Priya", "Ravi", "Suresh"],
    "Age": [25, 30, 28, 25, 32],
    "City": ["Hyderabad", "Chennai", "Bangalore",
             "Hyderabad", "Mumbai"]
}

df = pd.DataFrame(data)

# Display original dataset
print("Original Dataset:")
display(df)

# Identify exact duplicate records
duplicates = df[df.duplicated()]

print("Exact Duplicate Records:")
display(duplicates)

# Count exact duplicates
print("Number of Exact Duplicates:",
      df.duplicated().sum())

# Remove exact duplicates
df_cleaned = df.drop_duplicates()

print("Dataset After Removing Exact Duplicates:")
display(df_cleaned)
```

### Partial Duplicates


Partial Duplicates are records in a dataset where only some column values are identical, while other column values are different

In simple words, when two or more rows have the same values in selected columns but different values in other columns, they are called partial duplicates

```python
Example:

Consider a customer dataset:

Customer_ID	Name	Age	City
    101	    Ravi	25	Hyderabad
    102	    Kumar	30	Chennai
    103	    Priya	28	Bangalore
    104 	Ravi	25	Mumbai
    105	    Suresh	32	Mumbai

Here:

- Ravi appears twice.
- Both records have the same Name and Age.
- However, Customer_ID and City are different.
- Therefore, these records are partial duplicates based on Name and Age.
```


##### Mathematical Concept

Suppose a dataset contains multiple records with different columns.

Let:

$$
R_i = (X_1, X_2, X_3, X_4)
$$

$$
R_j = (X_1, X_2, X_3', X_4')
$$

If some selected column values are identical but other values are different, the records are partial duplicates.

##### Example

$$
R_1 = (101, \text{Ravi}, 25, \text{Hyderabad})
$$

$$
R_2 = (104, \text{Ravi}, 25, \text{Mumbai})
$$

Here:

$$
\text{Name}_1 = \text{Name}_2
$$

$$
\text{Age}_1 = \text{Age}_2
$$

But:

$$
\text{CustomerID}_1 \neq \text{CustomerID}_2
$$

$$
\text{City}_1 \neq \text{City}_2
$$

Therefore, these records are partial duplicates based on Name and Age.


Real-World Business Example:

Suppose an e-commerce company maintains customer registration information.

The same customer registers multiple times using different customer IDs or different addresses.

As a result:

- The company may count one customer as multiple customers.
- Customer analysis may become inaccurate.
- Marketing campaigns may target the same customer multiple times.
- Customer purchase history may become fragmented.

Identifying partial duplicates helps businesses maintain accurate customer information.

Importance in Data Preprocessing

- Identifies repeated customer information.
- Improves data consistency.
- Prevents incorrect customer counts.
- Helps combine customer information.
- Improves data quality before machine learning.

```python
import pandas as pd

# Create customer dataset
data = {
    "Customer_ID": [101, 102, 103, 104, 105],
    "Name": ["Ravi", "Kumar", "Priya", "Ravi", "Suresh"],
    "Age": [25, 30, 28, 25, 32],
    "City": ["Hyderabad", "Chennai", "Bangalore",
             "Mumbai", "Mumbai"]
}

df = pd.DataFrame(data)

# Display original dataset
print("Original Dataset:")
display(df)

# Identify partial duplicates using Name and Age
duplicates = df[df.duplicated(subset=["Name", "Age"])]

print("Partial Duplicate Records:")
display(duplicates)

# Count partial duplicates
print("Number of Partial Duplicates:",
      df.duplicated(subset=["Name", "Age"]).sum())

# Remove partial duplicates based on Name and Age
df_cleaned = df.drop_duplicates(subset=["Name", "Age"])

print("Dataset After Removing Partial Duplicates:")
display(df_cleaned)
```

### Identifying Duplicates


Identifying Duplicates is the process of finding repeated records in a dataset

In simple words, it means checking whether the same record or selected column values appear multiple times in a dataset

Identifying duplicates helps detect:

- Exact duplicate records.
- Partial duplicate records.
- Repeated customer information.
- Unnecessary repeated data.

```python
Example:

Consider a customer dataset:

Customer_ID	Name	Age	City
     101	Ravi	25	Hyderabad
     102	Kumar	30	Chennai
     103	Priya	28	Bangalore
     101	Ravi	25	Hyderabad
     104	Suresh	32	Mumbai
     105	Ravi	25	Mumbai

Here:

- Customer 101 appears twice with identical information.
- Ravi appears multiple times.
- The first case is an exact duplicate.
- The Ravi records can be identified as partial duplicates based on Name and Age.
```


##### Mathematical Concept

Suppose a dataset contains \(N\) records.

We can identify duplicate records by comparing each record with previous records.

##### Duplicate Condition

For exact duplicates:

$$
R_i = R_j
$$

For partial duplicates based on selected columns:

$$
X_{i,k} = X_{j,k}
$$

Where:

- \(R_i, R_j\) = Two records.
- \(X_{i,k}, X_{j,k}\) = Values in a selected column.
- \(k\) = Selected column.

##### Duplicate Percentage

$$
\text{Duplicate Percentage} =
\frac{\text{Duplicate Records}}{\text{Total Records}}
\times 100
$$


Why Is Identifying Duplicates Important

1. Improves data quality.
2. Prevents incorrect customer counts.
3. Avoids repeated information.
4. Improves statistical analysis.
5. Helps prepare reliable data for machine learning.

```python
import pandas as pd

# Create customer dataset
data = {
    "Customer_ID": [101, 102, 103, 101, 104, 105],
    "Name": ["Ravi", "Kumar", "Priya", "Ravi", "Suresh", "Ravi"],
    "Age": [25, 30, 28, 25, 32, 25],
    "City": ["Hyderabad", "Chennai", "Bangalore",
             "Hyderabad", "Mumbai", "Mumbai"]
}

df = pd.DataFrame(data)

# Display original dataset
print("Original Dataset:")
display(df)

# Identify exact duplicates
exact_duplicates = df[df.duplicated()]

print("Exact Duplicate Records:")
display(exact_duplicates)

# Identify partial duplicates based on Name and Age
partial_duplicates = df[
    df.duplicated(subset=["Name", "Age"], keep=False)
]

print("Partial Duplicate Records:")
display(partial_duplicates)

# Count exact duplicates
print("Exact Duplicate Count:", df.duplicated().sum())

# Count partial duplicates
print(
    "Partial Duplicate Count:",
    df.duplicated(subset=["Name", "Age"]).sum()
)
```


##### Real-World Business Example

### duplicated()


The duplicated() function in Pandas is used to identify duplicate records in a dataset.

It checks whether a row has already appeared in the dataset and returns a Boolean value:

- True – The record is a duplicate.
- False – The record is not a duplicate.

```python
Syntax:

DataFrame.duplicated(subset=None, keep='first')
```

```python
Example:

Consider a customer dataset:

Customer_ID	Name	Age	City
     101	Ravi	25	Hyderabad
     102	Kumar	30	Chennai
     103	Priya	28	Bangalore
     101	Ravi	25	Hyderabad
     104	Suresh	32	Mumbai

Here, the fourth row is an exact duplicate of the first row.
```

```python
import pandas as pd

# Create customer dataset
data = {
    "Customer_ID": [101, 102, 103, 101, 104],
    "Name": ["Ravi", "Kumar", "Priya", "Ravi", "Suresh"],
    "Age": [25, 30, 28, 25, 32],
    "City": ["Hyderabad", "Chennai", "Bangalore",
             "Hyderabad", "Mumbai"]
}

df = pd.DataFrame(data)

# Display original dataset
print("Original Dataset:")
display(df)

# Identify duplicate records
df["Is_Duplicate"] = df.duplicated()

print("Dataset with Duplicate Identification:")
display(df)

# Display only duplicate records
print("Duplicate Records:")
display(df[df["Is_Duplicate"] == True])

# Count duplicate records
print("Total Duplicate Records:", df.duplicated().sum())
```

```python
# Identify duplicates based on Name and Age          #Identifying Partial Duplicates
partial_duplicates = df.duplicated(
    subset=["Name", "Age"]
)

print("Partial Duplicate Identification:")
display(df[partial_duplicates])
```

```python
# Keep first occurrence                   #Using Different keep Parameters
print("Keep First:")
print(df.duplicated(keep="first"))

# Keep last occurrence
print("\nKeep Last:")
print(df.duplicated(keep="last"))

# Mark all duplicate occurrences
print("\nKeep False:")
print(df.duplicated(keep=False))
```

### drop_duplicates()

The drop_duplicates() function in Pandas is used to remove duplicate records from a dataset.

It identifies repeated rows and removes them while keeping one occurrence of each unique record.

In simple words, it helps eliminate unnecessary repeated information from a dataset.

```python
Syntax:

DataFrame.drop_duplicates(
    subset=None,
    keep='first',
    inplace=False
)


Example:

Consider a customer dataset:

Customer_ID	Name	Age	City
     101	Ravi	25	Hyderabad
     102	Kumar	30	Chennai
     103	Priya	28	Bangalore
     101	Ravi	25	Hyderabad
     104	Suresh	32	Mumbai
     102	Kumar	30	Chennai

Here:

- Customer 101 appears twice.
- Customer 102 appears twice.
- These are exact duplicate records.
- We can remove the repeated records using drop_duplicates().
```

```python
import pandas as pd

# Create customer dataset
data = {
    "Customer_ID": [101, 102, 103, 101, 104, 102],
    "Name": ["Ravi", "Kumar", "Priya", "Ravi", "Suresh", "Kumar"],
    "Age": [25, 30, 28, 25, 32, 30],
    "City": ["Hyderabad", "Chennai", "Bangalore",
             "Hyderabad", "Mumbai", "Chennai"]
}

df = pd.DataFrame(data)

# Display original dataset
print("Original Dataset:")
display(df)

# Count duplicate records
print("Duplicate Records:", df.duplicated().sum())

# Remove duplicate records
df_cleaned = df.drop_duplicates()

# Display cleaned dataset
print("\nDataset After Removing Duplicates:")
display(df_cleaned)

# Count remaining records
print("Total Records After Cleaning:", len(df_cleaned))
```

### Duplicate Detection Based on Selected Columns


Duplicate Detection Based on Selected Columns is the process of identifying duplicate records by comparing only specific columns in a dataset instead of checking all columns

In simple words, we select important columns and check whether their values are repeated across multiple records

This is useful for identifying partial duplicates

```python
Example:

Consider a customer dataset:

Customer_ID	Name	Age	City

     101	Ravi	25	Hyderabad
     102	Kumar	30	Chennai
     103	Priya	28	Bangalore
     104	Ravi	25	Mumbai
     105	Suresh	32	Mumbai

Here:

- Ravi appears twice.
- Both Ravi records have the same Name and Age.
- Their Customer_ID and City are different.
- Therefore, they are duplicates based on the selected columns Name and Age
```


##### Mathematical Concept

Suppose a dataset contains \(N\) records and \(K\) selected columns.

Let:

$$
S = \{X_1, X_2, \ldots, X_K\}
$$

Where:

- \(S\) = Set of selected columns.
- \(X_1, X_2, \ldots, X_K\) = Selected features.
- \(K\) = Number of selected columns.

Two records are considered duplicates based on selected columns if:

$$
R_i[S] = R_j[S]
$$



##### Example

Consider two customer records:

$$
R_1 = (101, \text{Ravi}, 25, \text{Hyderabad})
$$

$$
R_2 = (104, \text{Ravi}, 25, \text{Mumbai})
$$

Select the columns:

$$
S = \{\text{Name}, \text{Age}\}
$$

Then:

$$
R_1[S] = (\text{Ravi}, 25)
$$

$$
R_2[S] = (\text{Ravi}, 25)
$$

Therefore:

$$
R_1[S] = R_2[S]
$$

Both records are duplicates based on Name and Age.


```python
import pandas as pd

# Create customer dataset
data = {
    "Customer_ID": [101, 102, 103, 104, 105],
    "Name": ["Ravi", "Kumar", "Priya", "Ravi", "Suresh"],
    "Age": [25, 30, 28, 25, 32],
    "City": ["Hyderabad", "Chennai", "Bangalore",
             "Mumbai", "Mumbai"]
}

df = pd.DataFrame(data)

# Display original dataset
print("Original Dataset:")
display(df)

# Identify duplicates based on selected columns
duplicates = df[df.duplicated(
    subset=["Name", "Age"],
    keep=False
)]

print("Duplicate Records Based on Name and Age:")
display(duplicates)

# Count duplicate occurrences after the first
duplicate_count = df.duplicated(
    subset=["Name", "Age"]
).sum()

print("Duplicate Count:", duplicate_count)

# Remove duplicates based on selected columns
df_cleaned = df.drop_duplicates(
    subset=["Name", "Age"],
    keep="first"
)

print("Dataset After Removing Duplicates:")
display(df_cleaned)
```

### Handling Duplicate Records

Handling Duplicate Records is the process of identifying, analyzing, and managing repeated records in a dataset to improve data quality and accuracy

In simple words, it means finding repeated information, checking whether it is genuinely duplicated, and deciding whether to remove, retain, or combine those records

Duplicate records can be handled using:

- Identifying duplicate records.
- Removing exact duplicates.
- Detecting partial duplicates.
- Checking duplicates based on selected columns.
- Keeping the most recent or most reliable record.
- Combining information from duplicate records when appropriate.

```python
Example:

Consider a customer dataset:

Customer_ID	Name	Age	City

     101	Ravi	25	Hyderabad
     102	Kumar	30	Chennai
     103	Priya	28	Bangalore
     101	Ravi	25	Hyderabad
     104	Suresh	32	Mumbai
     102	Kumar	30	Chennai

Here:

- Customer 101 appears twice.
- Customer 102 appears twice.
- Both are exact duplicate records.
- We can identify and remove unnecessary repeated rows.
```


##### Mathematical Concept

Suppose a dataset contains \(N\) records.

Let:

- \(D\) = Number of duplicate records.
- \(U\) = Number of unique records.

The number of unique records is:

$$
U = N-D
$$

##### Duplicate Percentage

$$
\text{Duplicate Percentage} =
\frac{D}{N}\times100
$$

##### Example

Suppose an e-commerce company has 1,000 customer records, including 100 duplicate records.

$$
N=1000,\quad D=100
$$

Unique records:

$$
U=1000-100=900
$$

Duplicate percentage:

$$
\frac{100}{1000}\times100=10\%
$$

Therefore, 10% of the records are duplicates.


Steps for Handling Duplicate Records

1. Identify Duplicates
   
   Use duplicated() to find repeated records.
2. Analyze Duplicates
   
   Check whether records are exact duplicates or partial duplicates.
3. Select Duplicate Handling Method
   
   Decide whether to remove, retain, or combine records.
4. Remove Unnecessary Duplicates
   
   Use drop_duplicates() to remove confirmed duplicate records.
5. Verify Cleaned Dataset
   
   Check the remaining duplicate count and dataset dimensions.

```python
import pandas as pd

# Create customer dataset
data = {
    "Customer_ID": [101, 102, 103, 101, 104, 102],
    "Name": ["Ravi", "Kumar", "Priya", "Ravi", "Suresh", "Kumar"],
    "Age": [25, 30, 28, 25, 32, 30],
    "City": ["Hyderabad", "Chennai", "Bangalore",
             "Hyderabad", "Mumbai", "Chennai"]
}

df = pd.DataFrame(data)

# Step 1: Display original dataset
print("Original Dataset:")
display(df)

# Step 2: Identify duplicate records
duplicates = df[df.duplicated()]

print("Duplicate Records:")
display(duplicates)

# Step 3: Count duplicate records
duplicate_count = df.duplicated().sum()

print("Total Duplicate Records:", duplicate_count)

# Step 4: Calculate duplicate percentage
total_records = len(df)

duplicate_percentage = (duplicate_count / total_records) * 100

print("Duplicate Percentage:", duplicate_percentage, "%")

# Step 5: Remove duplicate records
df_cleaned = df.drop_duplicates()

print("\nCleaned Dataset:")
display(df_cleaned)

# Step 6: Verify remaining duplicates
remaining_duplicates = df_cleaned.duplicated().sum()

print("Remaining Duplicate Records:", remaining_duplicates)

# Step 7: Display final dataset dimensions
print("Original Shape:", df.shape)
print("Cleaned Shape:", df_cleaned.shape)
```

```python
Real-World Business Example:
    
Suppose an e-commerce company maintains 10,000 customer records.

After duplicate detection:
- Total records = 10,000
- Duplicate records = 500
- Unique records = 9,500
The company removes confirmed duplicate records and maintains 9,500 unique records.
```

### Business Rules for Duplicate Removal

Business Rules for Duplicate Removal are guidelines used to decide which duplicate records should be removed, retained, updated, or combined based on business requirements

In simple words, businesses should not remove every repeated record automatically. They must first understand whether the records represent the same customer, transaction, product, or business activity

Business rules help prevent the accidental deletion of important information

 Why Are Business Rules Important?

- Prevent accidental data loss.
- Maintain accurate customer information.
- Avoid incorrect business reports.
- Preserve important transaction records.
- Improve data quality.
- Ensure reliable machine learning datasets.


##### Mathematical Concept

Suppose a dataset contains \(N\) records.

Let:

- \(D\) = Number of duplicate records.
- \(R\) = Number of records removed.
- \(U\) = Number of remaining unique records.

The number of remaining records is:

$$
U=N-R
$$

### Example

Suppose an e-commerce company has 1,000 customer records.

- Total Records = 1,000
- Duplicate Records Identified = 100
- Records Removed = 100

$$
U=1000-100
$$

$$
U=900
$$

Therefore, 900 records remain after duplicate removal.


```python
Real-World Business Example:
    
Suppose an e-commerce company has 10,000 customer records.

After duplicate detection:
    
- Total records = 10,000
- Duplicate records identified = 500
- Records removed = 450
- Records requiring manual investigation = 50
                                         
The company removes 450 confirmed duplicate records and investigates the remaining 50 records because they contain different customer information

This helps maintain accurate customer information without losing valuable business data
```

```python
Blindly deleting duplicates means removing records without checking whether they represent the same real-world entity or 
whether the repeated information is important

In real-world datasets, two records may look similar but represent different customers, transactions, or business activities.
    
1. Loss of Important Information

Two records may contain similar information but different details. Deleting one record may result in losing useful information.

Example:

Customer ID	Name	Age	City
101	Ravi	25	Hyderabad
104	Ravi	25	Mumbai

If we delete one record based only on Name and Age, we might accidentally remove a different customer

2. Financial Data Can Be Lost

In banking or e-commerce datasets, multiple transactions with the same customer ID and amount may be legitimate.

Example:

Transaction ID	Customer ID	Amount
T101	C101	500
T102	C101	500

Both transactions may be valid purchases. Deleting one could lead to incorrect financial reports.
    
3. Incorrect Business Decisions

Removing valid records can affect:

- Total sales calculations.
- Customer counts.
- Revenue reports.
- Inventory analysis.
- Business performance metrics

4. Incorrect Machine Learning Results

Deleting valid records may change the dataset distribution and affect model training.

For example, removing legitimate repeated transactions may cause a fraud detection model to learn incorrect patterns

5. Loss of Historical Information

Repeated records may represent changes over time, such as customer addresses, employee roles, or product prices.

Deleting them without checking timestamps may remove important historical information.
```
