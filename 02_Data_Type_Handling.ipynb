# Notebook 2 – Data Type Handling

Data Type Handling is the process of identifying, checking, and converting the data types of columns in a dataset into their correct formats.

In Python Pandas, every column has a data type (dtype) that determines what kind of values it stores and what operations can be performed on those values.

### Identifying Data Types


Identifying Data Types is the process of examining each column in a dataset to understand what kind of information it contains and which data type is currently assigned to it

In simple words, we identify whether a column contains numerical values, text, categories, Boolean values, or dates

In Data Preprocessing, identifying data types is one of the first steps because it helps us detect incorrect data formats before performing data analysis or machine learning

Why Do We Identify Data Types?

Identifying data types is important for the following reasons:

1. Accurate calculations: Helps perform mathematical operations on numerical columns
2. Data quality: Identifies columns containing incorrect data formats
3. Data analysis: Helps select appropriate statistical methods
4. Data visualization: Helps choose suitable graphs
5. Machine learning: Helps prepare numerical and categorical features correctly
6. Data conversion: Helps identify which columns need type conversion

```python
Types of Data Types:
    
DataType	            Description	                                Example
Integer (int64)  	   Whole numbers	                           10, 20, 30
Float (float64)	       Decimal numbers	                           10.5, 20.75
Object (object)	       General text or mixed values	               Ravi, Hyderabad
String (string)	       Text values                              	Python, Data
Category (category) 	Limited groups or labels	                IT, HR, Sales
Boolean (bool)	           True or False	                         True, False
Datetime (datetime64)	Date and time values	                     2026-11-05
```

```python
import pandas as pd

# Create Employee Dataset
df = pd.DataFrame({
    "EmployeeID": [101, 102, 103, 104, 105],
    "Age": ["25", "30", "28", "35", "27"],
    "Salary": [25000.5, 35000.5, 30000.5, 40000.5, 30000.0],
    "Department": ["IT", "HR", "Sales", "IT", "HR"],
    "JoiningDate": [
        "2024-01-15",
        "2023-05-20",
        "2022-08-10",
        "2024-03-12",
        "2023-11-25"
    ],
    "IsActive": [True, False, True, True, False]
})

# Display Dataset
print("EMPLOYEE DATASET")
print(df)

# Display Dataset Information
print("\nDATASET INFORMATION")
df.info()

# Identify Current Data Types
print("\nCURRENT DATA TYPES")
print(df.dtypes)

# Identify Numerical Columns
numerical = df.select_dtypes(include="number").columns
print("\nNUMERICAL COLUMNS:")
print(list(numerical))

# Identify Categorical Columns
categorical = df.select_dtypes(
    include=["object", "category", "string"]
).columns
print("\nCATEGORICAL COLUMNS:")
print(list(categorical))

# Identify Boolean Columns
boolean = df.select_dtypes(include="bool").columns
print("\nBOOLEAN COLUMNS:")
print(list(boolean))

# Identify Datetime Columns
datetime = df.select_dtypes(include="datetime").columns
print("\nDATETIME COLUMNS:")
print(list(datetime))

# Identify Incorrect Data Types
print("\nINCORRECT DATA TYPE CHECK")

if df["Age"].dtype == "object":
    print("Age column is stored as text instead of numerical data.")

if df["JoiningDate"].dtype == "object":
    print("JoiningDate column is stored as text instead of datetime.")

# Convert Incorrect Data Types
df["Age"] = pd.to_numeric(df["Age"])
df["JoiningDate"] = pd.to_datetime(df["JoiningDate"])

# Display Corrected Data Types
print("\nCORRECTED DATA TYPES")
print(df.dtypes)

# Final Dataset
print("\nFINAL DATASET")
print(df)
```

### Numerical Data

Numerical Data is a type of data that represents values in the form of numbers. It is used to perform mathematical calculations, statistical analysis, comparisons, and machine learning operations

In simple words, Numerical Data represents information that can be counted or measured using numbers

Examples:

- Age = 25
- Salary = 35000
- Height = 5.8
- Weight = 63.5
- Experience = 3 years
- Product Price = ₹1500

```python
Types of Numerical Data

Numerical data is mainly divided into two types.
    
2.1 Discrete Data
    
Discrete Data represents countable values. It usually consists of whole numbers.

Examples:

- Number of employees
- Number of students
- Number of products
- Number of customers
- Number of orders

Number of Employees   	Number of Products
        10	                    25
        20	                    50
        30	                    75

Example:

The number of employees can be 10, 20, or 30, but it cannot normally be 10.5 employees

2.2 Continuous Data

Continuous Data represents measurable values. It can contain decimal values.

Examples:

- Height
- Weight
- Temperature
- Distance
- Time
- Salary (when measured with fractional units)

Person	Height	Weight
 Ravi	5.6	     60.5
 Kiran	5.8	     63.2
 Arjun	5.9	     65.7

Continuous values can take decimal values within a measurement range.

Why is Numerical Data Important?

1. Mathematical calculations: Calculate sum, average, minimum and maximum
2. Statistical analysis: Calculate mean, median, variance and standard deviation
3. Data visualization: Create histograms, box plots and scatter plots
4. Correlation analysis: Identify relationships between numerical features
5. Machine learning: Many algorithms use numerical features for model training
6. Business decisions: Analyze salary, sales, revenue, expenses and profits
```

```python
import pandas as pd

# Create Employee Dataset
df = pd.DataFrame({
    "EmployeeID": [101, 102, 103, 104, 105],
    "Age": [25, 30, 28, 35, 27],
    "Experience": [2, 5, 3, 8, 4],
    "Salary": [25000, 35000, 30000, 45000, 32000],
    "Height": [5.6, 5.8, 5.7, 5.9, 5.5],
    "Weight": [60.5, 65.2, 62.5, 70.5, 58.5],
    "Department": ["IT", "HR", "Sales", "IT", "HR"]
})

# Display Dataset
print("EMPLOYEE DATASET")
print(df)

# Identify Numerical Columns
numerical = df.select_dtypes(include="number")

print("\nNUMERICAL DATA")
print(numerical)

# Identify Discrete Features
discrete = ["Age", "Experience"]

print("\nDISCRETE DATA:")
print(df[discrete])

# Identify Continuous Features
continuous = ["Salary", "Height", "Weight"]

print("\nCONTINUOUS DATA:")
print(df[continuous])

# Statistical Analysis
print("\nSTATISTICAL SUMMARY")
print(numerical.describe())

# Calculate Mean
print("\nMEAN:")
print(numerical.mean())

# Calculate Median
print("\nMEDIAN:")
print(numerical.median())

# Calculate Standard Deviation
print("\nSTANDARD DEVIATION:")
print(numerical.std())

# Calculate Correlation
print("\nCORRELATION:")
print(numerical.drop(columns="EmployeeID").corr())
```

### Categorical Data

Categorical Data is a type of data that represents information in the form of groups, labels, names, or categories

In simple words, Categorical Data is used to classify data into different groups rather than represent measurable numerical quantities

Examples:

- Gender → Male, Female
- Department → IT, HR, Sales
- City → Hyderabad, Chennai, Bangalore
- Customer Type → Regular, Premium
- Payment Method → Cash, UPI, Credit Card

```python
Types of Categorical Data

Categorical data is mainly divided into two types

2.1 Nominal Data

Nominal Data represents categories that do not have any natural order or ranking.

Examples:

- Gender → Male, Female
- Department → IT, HR, Sales
- City → Hyderabad, Chennai, Mumbai
- Payment Method → Cash, UPI, Card
- Blood Group → A, B, AB, O

Example:

Employee	Department
   Ravi	        IT
   Kiran	    HR
   Arjun	   Sales
   Priya	    IT

There is no ranking between IT, HR and Sales

2.2 Ordinal Data

Ordinal Data represents categories that have a meaningful order or ranking.

Examples:

- Education → School, Bachelor's, Master's, PhD
- Customer Satisfaction → Low, Medium, High
- Product Quality → Poor, Average, Good, Excellent
- Priority → Low, Medium, High

Example:

Customer	Satisfaction
  Ravi	      High
 Kiran	     Medium
 Arjun	      Low
 Priya	    Excellent

Here, categories have a meaningful order.

Difference between Nominal and Ordinal Data

Nominal	                          Ordinal

No natural order	     Has a natural order
Used for classification	  Used for ranking
IT, HR, Sales	         Low, Medium, High
No ranking	             Ranking exists

Why is Categorical Data Important?

1. Customer Segmentation: Group customers based on customer type, location or preferences
2. Business Analysis: Analyze sales across departments, regions and product categories
3. Data Visualization: Create bar charts and count plots
4. Machine Learning: Convert categories into numerical representations using encoding techniques
5. Pattern Identification: Identify which categories have higher or lower frequencies
6. Decision Making: Understand customer behavior and business performance
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Create Employee Dataset
df = pd.DataFrame({
    "EmployeeID": [101, 102, 103, 104, 105, 106],
    "Department": ["IT", "HR", "Sales", "IT", "HR", "Sales"],
    "Gender": ["Male", "Female", "Male", "Female", "Male", "Female"],
    "City": ["Hyderabad", "Chennai", "Bangalore",
             "Hyderabad", "Chennai", "Mumbai"],
    "CustomerType": ["Premium", "Regular", "Premium",
                     "Regular", "Premium", "Regular"],
    "Satisfaction": ["High", "Medium", "Low",
                     "High", "Excellent", "Medium"]
})

# Display Dataset
print("EMPLOYEE DATASET")
print(df)

# Identify Categorical Columns
categorical = df.select_dtypes(
    include=["object", "category", "string"]
).columns

print("\nCATEGORICAL COLUMNS:")
print(list(categorical))

# Identify Nominal Features
nominal = ["Department", "Gender", "City", "CustomerType"]

print("\nNOMINAL DATA:")
print(df[nominal])

# Identify Ordinal Features
ordinal = ["Satisfaction"]

print("\nORDINAL DATA:")
print(df[ordinal])

# Analyze Unique Categories
print("\nUNIQUE CATEGORIES:")
for col in categorical:
    print(col, ":", df[col].unique())

# Count Unique Categories
print("\nNUMBER OF UNIQUE CATEGORIES:")
print(df[categorical].nunique())

# Category Frequency
print("\nCATEGORY FREQUENCY:")
for col in categorical:
    print("\n", col)
    print(df[col].value_counts())

# Percentage Distribution
print("\nCATEGORY PERCENTAGE:")
for col in categorical:
    print("\n", col)
    print(df[col].value_counts(normalize=True) * 100)

# Check Missing Values
print("\nMISSING VALUES:")
print(df[categorical].isnull().sum())

# Visualize Department Distribution
plt.figure(figsize=(7, 4))
sns.countplot(data=df, x="Department")
plt.title("Employee Department Distribution")
plt.xlabel("Department")
plt.ylabel("Employee Count")
plt.show()
```

### Boolean Data

Boolean Data is a data type that represents only two possible values:
- True
- False
- 
In simple words, Boolean Data is used to represent conditions, logical states, or yes/no information.

It helps us identify whether a particular condition is satisfied or not.

Examples:

Feature   	 Boolean Value

IsActive	   True
IsMarried	   False
IsPurchased	   True
IsEligible	   False
HasLoan	       True
IsFraud  	   False

Example dataset:

CustomerID	Age	IsActive	IsPurchased
  101	    25	True	      True
  102	    30	False	      False
  103	    28	True	      True
  104	    35	True	      False

Here, IsActive and IsPurchased are Boolean columns

Why is Boolean Data Important?

1. Condition Checking: Identifies whether a particular condition is True or False.
2. Data Filtering: Filters records based on specific conditions.
3. Business Decisions: Identifies active customers, eligible employees and completed transactions.
4. Data Analysis: Calculates True and False frequencies.
5. Machine Learning: Represents binary features and binary target variables.
6. Data Quality: Identifies whether records satisfy specific validation conditions.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Create Customer Dataset
df = pd.DataFrame({
    "CustomerID": [101, 102, 103, 104, 105, 106],
    "Age": [25, 30, 28, 35, 27, 32],
    "IsActive": [True, False, True, True, False, True],
    "IsPurchased": [True, False, True, False, True, True],
    "IsEligible": [True, True, False, True, False, True]
})

# Display Dataset
print("CUSTOMER DATASET")
print(df)

# Identify Boolean Columns
boolean_columns = df.select_dtypes(include="bool").columns

print("\nBOOLEAN COLUMNS:")
print(list(boolean_columns))

# Analyze True and False Counts
print("\nBOOLEAN VALUE COUNTS:")

for col in boolean_columns:
    print("\n", col)
    print(df[col].value_counts())

# Calculate True Percentage
print("\nTRUE PERCENTAGE:")

for col in boolean_columns:
    percentage = df[col].mean() * 100
    print(col, ":", round(percentage, 2), "%")

# Calculate False Percentage
print("\nFALSE PERCENTAGE:")

for col in boolean_columns:
    percentage = (1 - df[col].mean()) * 100
    print(col, ":", round(percentage, 2), "%")

# Identify Missing Values
print("\nMISSING VALUES:")
print(df[boolean_columns].isnull().sum())

# Filter Active Customers
active_customers = df[df["IsActive"] == True]

print("\nACTIVE CUSTOMERS:")
print(active_customers)

# Filter Customers Who Purchased
purchased_customers = df[df["IsPurchased"] == True]

print("\nPURCHASED CUSTOMERS:")
print(purchased_customers)

# Visualize Boolean Distribution
for col in boolean_columns:
    plt.figure(figsize=(5, 3))
    sns.countplot(data=df, x=col)
    plt.title(f"{col} Distribution")
    plt.xlabel(col)
    plt.ylabel("Customer Count")
    plt.show()
```

### Date/Time Data

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Create E-commerce Dataset
df = pd.DataFrame({
    "OrderID": [101, 102, 103, 104, 105, 106],

    "OrderDate": [
        "2024-01-15",
        "2024-02-20",
        "2024-03-10",
        "2024-04-05",
        "Invalid Date",
        "2024-06-18"
    ],

    "DeliveryDate": [
        "2024-01-18",
        "2024-02-25",
        "2024-03-14",
        "2024-04-10",
        "2024-05-15",
        "2024-06-22"
    ],

    "Sales": [2500, 3500, 4000, 3000, 4500, 5000],

    "CustomerType": [
        "Regular", "Premium", "Regular",
        "Premium", "Regular", "Premium"
    ]
})

# Display Original Dataset
print("ORIGINAL DATASET")
print(df)

# Identify Original Data Types
print("\nORIGINAL DATA TYPES")
print(df.dtypes)

# Convert String into Datetime
df["OrderDate"] = pd.to_datetime(
    df["OrderDate"],
    errors="coerce"
)

df["DeliveryDate"] = pd.to_datetime(
    df["DeliveryDate"],
    errors="coerce"
)

# Identify Datetime Columns
datetime_columns = df.select_dtypes(
    include="datetime"
).columns

print("\nDATETIME COLUMNS")
print(list(datetime_columns))

# Extract Date Components
df["Year"] = df["OrderDate"].dt.year
df["Month"] = df["OrderDate"].dt.month
df["Day"] = df["OrderDate"].dt.day
df["DayName"] = df["OrderDate"].dt.day_name()

# Calculate Delivery Duration
df["DeliveryDays"] = (
    df["DeliveryDate"] - df["OrderDate"]
).dt.days

# Display Missing Date Values
print("\nMISSING DATE VALUES")
print(df[["OrderDate", "DeliveryDate"]].isnull().sum())

# Display Statistical Summary
print("\nSTATISTICAL SUMMARY")
print(df[["Sales", "DeliveryDays"]].describe())

# Display Monthly Sales
print("\nMONTHLY SALES")
print(df.groupby("Month")["Sales"].sum())

# Visualize Monthly Sales
monthly_sales = df.groupby("Month")["Sales"].sum()

plt.figure(figsize=(8, 4))
sns.lineplot(x=monthly_sales.index, y=monthly_sales.values, marker="o")
plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Total Sales")
plt.xticks(monthly_sales.index)
plt.show()

# Display Final Dataset
print("\nFINAL DATASET")
print(df)

# Display Final Data Types
print("\nFINAL DATA TYPES")
print(df.dtypes)
```

String data is a type of data that contains text, characters, words, sentences, or combinations of letters, numbers, and special symbols

In Pandas, string values are commonly stored using:

- string

Why is String Data Important?

String data is important because real-world datasets contain text information.

For example:

String data helps us understand customer information, identify patterns, and extract useful features for Machine Learning

```python
Examples:

CustomerName	City	  Email	           ProductCode
  Ravi	     Hyderabad	 ravi@gmail.com	    PROD101
  Priya	     Chennai	 priya@gmail.com	PROD102
  Kumar	     Bangalore	 kumar@gmail.com	PROD103

Here, all four columns contain string data
```

```python
import pandas as pd

# Step 1: Create sample dataset
df = pd.DataFrame({
    "CustomerName": [
        " ravi ",
        "PRIYA",
        "kumar ",
        "Anjali"
    ],

    "City": [
        "hyderabad",
        "CHENNAI",
        " Bangalore ",
        "hyderabad"
    ],

    "Email": [
        "RAVI@GMAIL.COM",
        "priya@gmail.com",
        "kumar@gmail.com",
        "anjali@gmail.com"
    ],

    "Product": [
        "Laptop",
        "Mobile",
        "Laptop",
        "Tablet"
    ]
})

print("Original Dataset:")
print(df)


# Step 2: Identify string columns
string_columns = df.select_dtypes(
    include=["object", "string"]
).columns

print("\nString Columns:")
print(string_columns)


# Step 3: Remove extra spaces
for col in string_columns:
    df[col] = df[col].str.strip()


# Step 4: Standardize customer names
df["CustomerName"] = df["CustomerName"].str.title()


# Step 5: Standardize city names
df["City"] = df["City"].str.title()


# Step 6: Convert email addresses to lowercase
df["Email"] = df["Email"].str.lower()


# Step 7: Search for customers from Hyderabad
hyderabad_customers = df[
    df["City"].str.contains("Hyderabad", case=False, na=False)
]

print("\nHyderabad Customers:")
print(hyderabad_customers)


# Step 8: Calculate customer name length
df["NameLength"] = df["CustomerName"].str.len()


# Step 9: Extract email domain
df["EmailDomain"] = df["Email"].str.split("@").str[-1]


# Step 10: Display cleaned dataset
print("\nCleaned Dataset:")
print(df)


# Step 11: Analyze city frequency
print("\nCity Frequency:")
print(df["City"].value_counts())


# Step 12: Analyze product frequency
print("\nProduct Frequency:")
print(df["Product"].value_counts())
```

### Converting Data Types


Data type conversion is the process of changing a column or variable from one data type into another data type

In Data Preprocessing, we convert incorrect or unsuitable data types into appropriate types so that we can perform calculations, comparisons, statistical analysis, and Machine Learning operations correctly

```python
Example:

Suppose a dataset contains an Age column:
    
Age
    
"25"
"30"
"28"

Here, Age is stored as a string (object), but age should be numerical (int64).

After conversion:
    
Age
    
25
30
28

Now we can calculate the average age

Why is Data Type Conversion Important?

1. Enables mathematical calculations.
2. Improves data consistency.
3. Helps perform statistical analysis.
4. Supports correct filtering and comparisons.
5. Makes data suitable for Machine Learning.
6. Helps reduce memory usage when appropriate.
7. Prevents errors during data analysis.
3. Types of Data Type Conversion

     Conversion	                      Example	                    Pandas method

String to Integer	                "25" → 25	                     astype(int)
String to Float	                    "25.5" → 25.5	                 astype(float)
Integer to Float	                 25 → 25.0	                     astype(float)
Float to Integer	                 25.8 → 25	                     astype(int)
String to Datetime	                 "2026-11-05" → Timestamp	     pd.to_datetime()
String to Numeric	                  "100" → 100	                 pd.to_numeric()
String to Category	                 "Male" → category	             astype("category")
Numeric to String	                  25 → "25"	                     astype(str)

Important: Converting a float to an integer using astype(int) removes the decimal part; it does not round the value
```

```python
import pandas as pd

# Step 1: Create sample dataset
df = pd.DataFrame({
    "CustomerID": ["101", "102", "103", "104"],

    "Age": ["25", "30", "28", "35"],

    "Salary": ["25000.50", "30000", "Invalid", "45000.75"],

    "JoiningDate": [
        "2024-01-15",
        "2024-03-20",
        "Invalid",
        "2024-06-10"
    ],

    "Gender": ["Male", "Female", "Male", "Female"]
})

# Step 2: Display original dataset
print("Original Dataset:")
print(df)

# Step 3: Check original data types
print("\nOriginal Data Types:")
print(df.dtypes)

# Step 4: Convert CustomerID to integer
df["CustomerID"] = df["CustomerID"].astype(int)

# Step 5: Convert Age to integer
df["Age"] = pd.to_numeric(
    df["Age"],
    errors="coerce"
).astype("Int64")

# Step 6: Convert Salary to float
df["Salary"] = pd.to_numeric(
    df["Salary"],
    errors="coerce"
)

# Step 7: Convert JoiningDate to datetime
df["JoiningDate"] = pd.to_datetime(
    df["JoiningDate"],
    errors="coerce"
)

# Step 8: Convert Gender to category
df["Gender"] = df["Gender"].astype("category")

# Step 9: Display converted dataset
print("\nConverted Dataset:")
print(df)

# Step 10: Check converted data types
print("\nConverted Data Types:")
print(df.dtypes)

# Step 11: Check missing values created during conversion
print("\nMissing Values:")
print(df.isnull().sum())

# Step 12: Perform numerical calculation
print("\nAverage Age:")
print(df["Age"].mean())

print("\nAverage Salary:")
print(df["Salary"].mean())
```

### astype()

astype() is a Pandas method used to convert a Series or DataFrame column from one data type into another data type

It is commonly used in Data Preprocessing to correct incorrect data types and maintain consistency in a dataset

```python
Example:

Suppose Age is stored as a string:
    
Age
"25"
"30"
"28"

We want to convert it into an integer.
    
df["Age"] = df["Age"].astype(int)

After conversion:
Age
25
30
28


Common Data Type Conversions:
    
Conversion	            Python code

String to Integer	  .astype(int)
String to Float	      .astype(float)
Integer to Float	  .astype(float)
Float to Integer	  .astype(int)
Integer to String	  .astype(str)
String to Category	  .astype("category")
Integer to Boolean 	  .astype(bool)
```

```python
import pandas as pd

# Step 1: Create sample dataset
df = pd.DataFrame({
    "CustomerID": ["101", "102", "103", "104"],
    "Age": ["25", "30", "28", "35"],
    "Salary": [25000, 30000, 35000, 40000],
    "Marks": [85.7, 90.5, 78.9, 88.6],
    "Gender": ["Male", "Female", "Male", "Female"]
})

# Step 2: Display original dataset
print("Original Dataset:")
print(df)

# Step 3: Check original data types
print("\nOriginal Data Types:")
print(df.dtypes)

# Step 4: Convert CustomerID to integer
df["CustomerID"] = df["CustomerID"].astype(int)

# Step 5: Convert Age to integer
df["Age"] = df["Age"].astype(int)

# Step 6: Convert Salary to float
df["Salary"] = df["Salary"].astype(float)

# Step 7: Convert Marks to integer
df["Marks"] = df["Marks"].astype(int)

# Step 8: Convert Gender to category
df["Gender"] = df["Gender"].astype("category")

# Step 9: Display converted dataset
print("\nConverted Dataset:")
print(df)

# Step 10: Check converted data types
print("\nConverted Data Types:")
print(df.dtypes)

# Step 11: Perform numerical calculations
print("\nAverage Age:", df["Age"].mean())

print("Average Salary:", df["Salary"].mean())

print("Average Marks:", df["Marks"].mean())
```

### to_numeric()

pd.to_numeric() is a Pandas function used to convert values into a numerical data type, such as int64 or float64

It is especially useful when a dataset contains numbers stored as strings or contains invalid values that cannot be converted directly

```python
Syntax:

pd.to_numeric(
    arg,
    errors="raise"
)

rg — the column or values to convert

errors — determines how invalid values are handled
```

```python
Example:

Original value	Converted value
"100"	           100
"250"	           250
"300"	           300
"Invalid"	       NaN
```

```python
import pandas as pd

# Step 1: Create sample dataset
df = pd.DataFrame({
    "CustomerID": [101, 102, 103, 104, 105],
    "Age": ["25", "30", "Twenty", "35", "28"],
    "MonthlyIncome": [
        "25000",
        "30000",
        "Invalid",
        "45000",
        "35000"
    ],
    "PurchaseAmount": [
        "1500.50",
        "2000.75",
        "1200",
        "Invalid",
        "3000"
    ]
})

print("Original Dataset:")
print(df)

# Step 2: Check original data types
print("\nOriginal Data Types:")
print(df.dtypes)

# Step 3: Convert Age to numeric
df["Age"] = pd.to_numeric(
    df["Age"],
    errors="coerce"
)

# Step 4: Convert MonthlyIncome to numeric
df["MonthlyIncome"] = pd.to_numeric(
    df["MonthlyIncome"],
    errors="coerce"
)

# Step 5: Convert PurchaseAmount to numeric
df["PurchaseAmount"] = pd.to_numeric(
    df["PurchaseAmount"],
    errors="coerce"
)

# Step 6: Display converted dataset
print("\nConverted Dataset:")
print(df)

# Step 7: Check converted data types
print("\nConverted Data Types:")
print(df.dtypes)

# Step 8: Identify missing values
print("\nMissing Values:")
print(df.isnull().sum())

# Step 9: Calculate average age
print("\nAverage Age:")
print(df["Age"].mean())

# Step 10: Calculate average monthly income
print("\nAverage Monthly Income:")
print(df["MonthlyIncome"].mean())

# Step 11: Calculate average purchase amount
print("\nAverage Purchase Amount:")
print(df["PurchaseAmount"].mean())
```

### to_datetime()


pd.to_datetime() is a Pandas function used to convert values into a datetime data type.

It is useful when dates and times are stored as strings or objects in a dataset.

For example, a dataset may contain:

JoiningDate (original)
"2024-01-15"
"2024-03-20"
"2024-06-10"

These values may be stored as strings. We can convert them into datetime format to perform date-based operations


```python
Syntax:

pd.to_datetime(    arg,    format=None,    errors="raise")

Parameter	     Description

arg	             Column or values to convert
format	         Expected date format, if known
errors	         Controls how invalid dates are handled
```

```python
import pandas as pd

# Step 1: Create sample dataset
df = pd.DataFrame({
    "OrderID": [101, 102, 103, 104, 105],

    "OrderDate": [
        "2024-01-10",
        "2024-02-15",
        "Invalid",
        "2024-04-20",
        "2024-05-25"
    ],

    "DeliveryDate": [
        "2024-01-15",
        "2024-02-20",
        "2024-03-15",
        "2024-04-25",
        "2024-06-01"
    ],

    "Sales": [1500, 2500, 1800, 3000, 2200]
})

print("Original Dataset:")
print(df)

# Step 2: Check original data types
print("\nOriginal Data Types:")
print(df.dtypes)

# Step 3: Convert OrderDate to datetime
df["OrderDate"] = pd.to_datetime(
    df["OrderDate"],
    errors="coerce"
)

# Step 4: Convert DeliveryDate to datetime
df["DeliveryDate"] = pd.to_datetime(
    df["DeliveryDate"],
    errors="coerce"
)

# Step 5: Extract year and month
df["OrderYear"] = df["OrderDate"].dt.year
df["OrderMonth"] = df["OrderDate"].dt.month

# Step 6: Extract weekday
df["OrderWeekday"] = df["OrderDate"].dt.day_name()

# Step 7: Calculate delivery duration
df["DeliveryDays"] = (
    df["DeliveryDate"] - df["OrderDate"]
).dt.days

# Step 8: Display converted dataset
print("\nConverted Dataset:")
print(df)

# Step 9: Identify missing dates
print("\nMissing Date Values:")
print(df[["OrderDate", "DeliveryDate"]].isnull().sum())

# Step 10: Calculate average delivery duration
print("\nAverage Delivery Days:")
print(df["DeliveryDays"].mean())

# Step 11: Analyze monthly sales
print("\nMonthly Sales:")
print(df.groupby("OrderMonth")["Sales"].sum())
```

### Handling Invalid Conversions

Invalid conversion occurs when a value cannot be converted into the required data type

For example, suppose a dataset contains an Age column:

Age

"25"

"30"

"Twenty"

"Unknown"

We want to convert Age into numerical data.

However, "Twenty" and "Unknown" cannot be directly converted into integers

Why Do Invalid Conversions Occur?

Common causes include:

- Text values inside numerical columns
  
- Incorrect date formats


- Extra spaces or special characters


- Currency symbols and commas


- Invalid or impossible dates


- Inconsistent data entry


Column	         Invalid value	       Expected type

Age       	     "Twenty"	            Integer

Salary	         "Unknown"	            Numeric

JoiningDate	    "Not a date"         	Datetime

Marks	        "Ninety"	             Numeric

Price	        "₹2,500"            	 Numeric

```python
import pandas as pd

# Step 1: Create sample dataset
df = pd.DataFrame({
    "CustomerID": [101, 102, 103, 104, 105],

    "Age": ["25", "30", "Twenty", "40", "Unknown"],

    "Salary": [
        "25000",
        "30000",
        "Invalid",
        "45000",
        "50000"
    ],

    "JoiningDate": [
        "2024-01-15",
        "2024-02-20",
        "Invalid",
        "2024-04-30",
        "2024-02-30"
    ]
})

print("Original Dataset:")
print(df)

# Step 2: Convert Age into numeric
df["Age"] = pd.to_numeric(
    df["Age"],
    errors="coerce"
)

# Step 3: Convert Salary into numeric
df["Salary"] = pd.to_numeric(
    df["Salary"],
    errors="coerce"
)

# Step 4: Convert JoiningDate into datetime
df["JoiningDate"] = pd.to_datetime(
    df["JoiningDate"],
    errors="coerce"
)

print("\nConverted Dataset:")
print(df)

# Step 5: Identify missing values
print("\nMissing Values:")
print(df.isnull().sum())

# Step 6: Display rows containing invalid conversions
invalid_rows = df[
    df[["Age", "Salary", "JoiningDate"]].isnull().any(axis=1)
]

print("\nRows with Invalid Conversions:")
print(invalid_rows)

# Step 7: Calculate average age
print("\nAverage Age:")
print(df["Age"].mean())

# Step 8: Calculate average salary
print("\nAverage Salary:")
print(df["Salary"].mean())
```

### Detecting Incorrect Data Types


An incorrect data type occurs when a dataset column is stored using a data type that does not match the actual meaning of its values

```python
For example, an Age column may contain numerical values but be stored as strings

Column	    Stored data type	Correct data type
Age	           object	           Integer
Salary	      object	           Float
JoiningDate	  object	          Datetime
IsActive	 object	              Boolean
Gender	     object	              Category or string

Why Should We Detect Incorrect Data Types?

Detecting incorrect data types helps us:

- Perform mathematical calculations correctly
- Identify values stored as strings instead of numbers
- Recognize dates stored as text
- Detect inconsistent data formats
- Prepare features for statistical analysis and ML
- Avoid conversion errors during preprocessing
```

```python
import pandas as pd

# Step 1: Create sample dataset
df = pd.DataFrame({
    "OrderID": ["ORD101", "ORD102", "ORD103", "ORD104", "ORD105"],

    "Product": ["Laptop", "Mobile", "Tablet", "Laptop", "Mobile"],

    "Quantity": ["2", "3", "Invalid", "1", "5"],

    "Price": ["45000.50", "15000", "25000", "Invalid", "18000"],

    "OrderDate": [
        "2026-01-10",
        "2026-02-15",
        "Invalid",
        "2026-04-20",
        "2026-05-25"
    ],

    "IsDelivered": ["True", "False", "True", "False", "True"]
})

print("Original Dataset:")
print(df)


# Step 2: Display original data types
print("\nOriginal Data Types:")
print(df.dtypes)


# Step 3: Identify object columns
object_columns = df.select_dtypes(
    include="object"
).columns

print("\nObject Columns:")
print(object_columns)


# Step 4: Inspect values in object columns
print("\nInspecting Object Column Values:")

for column in object_columns:
    print(f"\n{column}:")
    print(df[column].unique())


# Step 5: Detect columns that may contain numeric values
print("\nChecking Numeric Conversion:")

for column in ["Quantity", "Price"]:
    converted = pd.to_numeric(
        df[column],
        errors="coerce"
    )

    invalid_values = df.loc[
        converted.isna() & df[column].notna(),
        column
    ]

    print(f"\n{column} invalid values:")
    print(invalid_values)


# Step 6: Convert Quantity to numeric
df["Quantity"] = pd.to_numeric(
    df["Quantity"],
    errors="coerce"
).astype("Int64")


# Step 7: Convert Price to numeric
df["Price"] = pd.to_numeric(
    df["Price"],
    errors="coerce"
)


# Step 8: Convert OrderDate to datetime
df["OrderDate"] = pd.to_datetime(
    df["OrderDate"],
    errors="coerce"
)


# Step 9: Convert IsDelivered to Boolean
df["IsDelivered"] = df["IsDelivered"].map({
    "True": True,
    "False": False
}).astype("boolean")


# Step 10: Display corrected dataset
print("\nCorrected Dataset:")
print(df)


# Step 11: Check corrected data types
print("\nCorrected Data Types:")
print(df.dtypes)


# Step 12: Identify missing values created during conversion
print("\nMissing Values:")
print(df.isnull().sum())


# Step 13: Calculate total order value
df["TotalAmount"] = df["Quantity"] * df["Price"]

print("\nTotal Order Amount:")
print(df[["OrderID", "Quantity", "Price", "TotalAmount"]])
```
