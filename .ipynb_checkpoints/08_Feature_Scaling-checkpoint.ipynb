# Notebook 8 – Feature Scaling

### Why Feature Scaling?


Feature Scaling is a data preprocessing technique used to bring numerical features into a similar range or scale

Why Do We Need Feature Scaling?

There are several important reasons to perform feature scaling

1. To Bring Features into a Similar Range
   
Different features may have different numerical ranges.

Example:

- Age = 25
- Salary = 50,000
- Experience = 3
  
Without scaling, these values have significantly different magnitudes.

Feature scaling brings them into a comparable range

2. To Prevent Feature Dominance
   
Some machine learning algorithms are sensitive to numerical magnitudes.

For example, in distance-based algorithms, Salary may contribute much more to the calculated distance than Age.

Scaling helps prevent this unwanted dominance.

3. To Improve Model Performance
   
Feature scaling can improve the performance of algorithms that depend on distances, gradients, or numerical optimization.

Examples:

- K-Nearest Neighbors (KNN)
- K-Means Clustering
- Support Vector Machine (SVM)
- Logistic Regression
- Linear Regression with regularization
- Neural Networks
  
4. To Improve Model Training Speed
   
Some optimization algorithms converge more efficiently when numerical features have comparable scales.

This can help gradient-based models train faster.

5. To Improve Distance Calculations
   
Distance-based algorithms calculate distances between data points.

For example, Euclidean distance:

\[
d(X,Y)=\sqrt{\sum_{i=1}^{n}(X_i-Y_i)^2}
\]

If one feature has a much larger numerical range, it can dominate the distance calculation.

Feature scaling helps balance the contribution of different features.

Real-World Business Example:
 
Suppose an e-commerce company wants to predict whether a customer will purchase a product.

The dataset contains:

Customer	Age	AnnualSalary	Website Visits
A	         25	30,000	           5
B	         35	60,000	           10
C	         45	90,000	           15
D	         30	40,000	           8

The company wants to use the KNN algorithm.

Problem without Feature Scaling:

- Age values are relatively small.
- Website Visits values are relatively small.
- Annual Salary values are much larger.
  
Because KNN uses distance calculations, Annual Salary may dominate the distance measurement.

Solution:

Apply Feature Scaling to bring Age, Annual Salary, and Website Visits into comparable ranges.

This helps KNN consider all three features more appropriately.


##### Mathematical Concept – Feature Scaling

Suppose a dataset contains numerical features:

$$
X_1, X_2, X_3, \ldots, X_n
$$

Each feature may have a different numerical range.

Feature Scaling transforms the original feature values into scaled values.

The general representation is:

$$
X_{\text{scaled}} = f(X)
$$

Where:

- $X$ = Original feature value.
- $f$ = Feature scaling transformation.
- $X_{\text{scaled}}$ = Transformed feature value.

##### 1. Min-Max Scaling

Min-Max Scaling transforms numerical values into a range of 0 to 1.

Mathematical Formula:

$$
X_{\text{scaled}} =
\frac{X - X_{\min}}{X_{\max} - X_{\min}}
$$

##### 2. Standardization (Z-score Scaling)

Standardization transforms numerical features using their mean and standard deviation.

Mathematical Formula:

$$
Z = \frac{X - \mu}{\sigma}
$$

Where:

- $\mu$ = Mean.
- $\sigma$ = Standard deviation.
- $Z$ = Standardized value.

Standardization transforms features to approximately zero mean and unit variance.


```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler, StandardScaler

# Create a dataset
data = {
    "Age": [20, 25, 30, 35, 40],
    "Salary": [20000, 30000, 40000, 50000, 60000],
    "Experience": [1, 2, 4, 6, 8]
}

df = pd.DataFrame(data)

print("Original Dataset:")
print(df)

# Apply Min-Max Scaling
minmax_scaler = MinMaxScaler()
minmax_scaled = minmax_scaler.fit_transform(df)

minmax_df = pd.DataFrame(
    minmax_scaled,
    columns=df.columns
)

print("\nMin-Max Scaled Dataset:")
print(minmax_df)

# Apply Standardization
standard_scaler = StandardScaler()
standard_scaled = standard_scaler.fit_transform(df)

standard_df = pd.DataFrame(
    standard_scaled,
    columns=df.columns
)

print("\nStandardized Dataset:")
print(standard_df)
```

```python
When Is Feature Scaling Important?

Algorithm	           Scaling Needed?
KNN	                      Yes
K-Means	                   Yes
SVM	                       Yes
Logistic Regression	     Usually yes
Linear Regression	     Usually not, unless regularization or numerical considerations apply
Neural Networks	         Usually yes
Decision Tree	         No
Random Forest	         No
XGBoost	                 Generally no

Important Points to Remember:
    
1. Feature scaling is mainly used for numerical features.
2. It prevents large numerical ranges from dominating distance-based calculations.
3. It can improve optimization and training efficiency.
4. Min-Max Scaling transforms values into a specified range, commonly 0 to 1.
5. Standardization transforms values using mean and standard deviation.
6. Tree-based algorithms generally do not require feature scaling.
7. Fit the scaler only on training data and use the same fitted scaler to transform validation and test data. This prevents data leakage

Important: Feature Scaling does not remove outliers or missing values. These should be handled appropriately as separate preprocessing steps.
```

### Scale-Sensitive Algorithms


Scale-Sensitive Algorithms are machine learning algorithms whose performance or calculations can be affected by the numerical scale or range of input features

When numerical features have different ranges, features with larger values may dominate distance calculations, optimization, or model training

Therefore, Feature Scaling is generally important when using these algorithms

Examples of Scale-Sensitive Algorithms:

1. K-Nearest Neighbors (KNN)
2. K-Means Clustering
3. Support Vector Machine (SVM)
4. Logistic Regression
5. Linear Regression with Regularization (Ridge and Lasso)
6. Principal Component Analysis (PCA)
7. Neural Networks


##### Mathematical Concept 

Many scale-sensitive algorithms use distance calculations or optimization.

For example, Euclidean Distance is calculated as:

$$
d(X,Y)=\sqrt{\sum_{i=1}^{n}(X_i-Y_i)^2}
$$

Where:

- $X$ = First data point.
- $Y$ = Second data point.
- $n$ = Number of features.
- $d(X,Y)$ = Euclidean distance.

 Example

Suppose two customers have the following values:

- Customer A = (25, 20000)
- Customer B = (30, 50000)

Euclidean Distance:

$$
d(X,Y)=\sqrt{(30-25)^2+(50000-20000)^2}
$$

$$
=\sqrt{25+900000000}
$$

$$
=30000.0004
$$


```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import StandardScaler
from scipy.spatial.distance import euclidean

# Create customer data
data = {
    "Age": [25, 30],
    "Salary": [20000, 50000],
    "Experience": [2, 5]
}

df = pd.DataFrame(data, index=["Customer A", "Customer B"])

print("Original Dataset:")
print(df)

# Calculate Euclidean Distance without scaling
distance_original = euclidean(
    df.iloc[0],
    df.iloc[1]
)

print("\nDistance Without Scaling:")
print(distance_original)

# Apply Standardization
scaler = StandardScaler()
scaled_data = scaler.fit_transform(df)

scaled_df = pd.DataFrame(
    scaled_data,
    columns=df.columns,
    index=df.index
)

print("\nScaled Dataset:")
print(scaled_df)

# Calculate Euclidean Distance after scaling
distance_scaled = euclidean(
    scaled_df.iloc[0],
    scaled_df.iloc[1]
)

print("\nDistance With Scaling:")
print(distance_scaled)
```

### Scale-Insensitive Algorithms

Scale-Insensitive Algorithms are machine learning algorithms whose model calculations and predictions are generally not affected by the numerical scale or range of input features

These algorithms do not usually require Feature Scaling because they make decisions based on feature values, thresholds, or splits rather than distance calculations.

For example, Decision Trees split data based on conditions such as:
\[
Age \leq 30
\]

The algorithm focuses on the splitting condition rather than the numerical magnitude of the feature.

Examples of Scale-Insensitive Algorithms

1. Decision Tree
2. Random Forest
3. Extra Trees
4. XGBoost
5. LightGBM
6. CatBoost


##### Mathematical Concept – Decision Tree Splitting

Decision Tree algorithms select splitting conditions based on feature values.

Suppose a dataset contains:

| Customer | Age | Salary | Purchased |
|:---|---:|---:|:---|
| A | 20 | 20,000 | No |
| B | 25 | 30,000 | No |
| C | 35 | 60,000 | Yes |
| D | 40 | 80,000 | Yes |

A Decision Tree may create a splitting condition:

$$
Age \leq 30
$$

The dataset is divided into two groups:

**Group 1:**

$$
Age \leq 30
$$

Customers A and B belong to this group.

**Group 2:**

$$
Age > 30
$$

Customers C and D belong to this group.

The algorithm uses these conditions to make predictions.

Feature Scale Conversion Example

Suppose Salary is converted from rupees into thousands of rupees.

$$
Salary_{\text{original}} = 20000
$$

$$
Salary_{\text{scaled}} = \frac{20000}{1000}
$$

$$
Salary_{\text{scaled}} = 20
$$

The numerical values change, but their ordering remains unchanged.

Therefore, the Decision Tree can adjust its splitting threshold and maintain an equivalent decision structure.


```python
import pandas as pd
from sklearn.tree import DecisionTreeClassifier

# Create dataset
data = {
    "Age": [20, 25, 35, 40, 45, 50],
    "Salary": [20000, 30000, 60000, 80000, 90000, 100000],
    "Purchased": [0, 0, 1, 1, 1, 1]
}

df = pd.DataFrame(data)

# Separate features and target
X = df[["Age", "Salary"]]
y = df["Purchased"]

# Create original Decision Tree
model1 = DecisionTreeClassifier(random_state=42)
model1.fit(X, y)

# Convert Salary into thousands
X_converted = X.copy()
X_converted["Salary"] = X_converted["Salary"] / 1000

# Create another Decision Tree
model2 = DecisionTreeClassifier(random_state=42)
model2.fit(X_converted, y)

# New customer data
new_customer = pd.DataFrame({
    "Age": [32],
    "Salary": [55000]
})

# Predict using original features
prediction1 = model1.predict(new_customer)

# Convert new customer salary
new_customer_converted = new_customer.copy()
new_customer_converted["Salary"] /= 1000

# Predict using converted features
prediction2 = model2.predict(new_customer_converted)

# Display results
print("Prediction Without Scale Conversion:", prediction1[0])
print("Prediction With Scale Conversion:", prediction2[0])

print("\nOriginal Salary:")
print(X["Salary"].tolist())

print("\nConverted Salary:")
print(X_converted["Salary"].tolist())
```

- Scale-insensitive algorithms generally do not require feature scaling.
- Decision Trees use feature-based splitting conditions.
- Random Forest uses multiple Decision Trees.
- Tree-based algorithms are generally unaffected by changes in feature units when the ordering of values is preserved.
- Feature Scaling is usually unnecessary for tree-based models.
- Always handle missing values and categorical features appropriately.
  
Real-World Business Example:

An e-commerce company wants to predict whether a customer will purchase a product using Age, Salary, and Website Visits.

Using Random Forest, the company can build a prediction model without necessarily applying Min-Max Scaling or Standardization.

### Normalization


Normalization is a data preprocessing technique used to transform numerical feature values into a specific range, commonly between 0 and 1.

It helps bring numerical features with different ranges into a comparable scale.

Normalization is commonly performed using Min-Max Scaling.
2. Why Do We Need Normalization?
1. To bring numerical features into a common range.
2. To prevent large numerical values from dominating distance calculations.
3. To improve the performance of distance-based algorithms.
4. To make numerical features comparable.
5. To support machine learning algorithms such as KNN, K-Means and Neural Networks.


##### Mathematical Concept – Min-Max Normalization

Suppose a dataset contains a numerical feature:

$$
X = \{X_1, X_2, X_3, \ldots, X_n\}
$$

Min-Max Normalization transforms each value using the following formula:

$$
X_{\text{normalized}} =
\frac{X-X_{\min}}{X_{\max}-X_{\min}}
$$

Where:

- $X$ = Original feature value.
- $X_{\min}$ = Minimum value of the feature.
- $X_{\max}$ = Maximum value of the feature.
- $X_{\text{normalized}}$ = Normalized feature value.

The resulting values generally fall between:

$$
0 \leq X_{\text{normalized}} \leq 1
$$


```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler

# Create dataset
data = {
    "Age": [20, 25, 30, 35, 40],
    "Salary": [20000, 30000, 40000, 50000, 60000],
    "Experience": [1, 2, 3, 4, 5]
}

df = pd.DataFrame(data)

print("Original Dataset:")
print(df)

# Create MinMaxScaler
scaler = MinMaxScaler()

# Apply Normalization
normalized_data = scaler.fit_transform(df)

# Convert into DataFrame
normalized_df = pd.DataFrame(
    normalized_data,
    columns=df.columns
)

print("\nNormalized Dataset:")
print(normalized_df)

# Display minimum and maximum values
print("\nMinimum Values:")
print(normalized_df.min())

print("\nMaximum Values:")
print(normalized_df.max())
```

```python
Real-World Business Example:
    
Suppose an e-commerce company has customer information containing:

- Age
- Annual Salary
- Number of Website Visits
- Total Purchase Amount

These features have different numerical ranges.

The company applies Min-Max Normalization before using the KNN algorithm to identify customers with similar purchasing behavior.

This makes the distance calculation less dominated by features with larger numerical values.
```

### Standardization

Standardization is a data preprocessing technique used to transform numerical features so that they have approximately:

- Mean = 0
- Standard Deviation = 1
- 
It is also called Z-score Scaling.

Unlike Min-Max Normalization, Standardization does not restrict values to a fixed range between 0 and 1.

Why Do We Need Standardization?

1. To bring numerical features to a comparable scale.
2. To prevent large numerical values from dominating certain algorithms.
3. To improve optimization during model training.
4. To improve the performance of scale-sensitive algorithms.
5. To make numerical features easier to compare based on their distance from the mean.

```python
import pandas as pd
from sklearn.preprocessing import StandardScaler

# Create dataset
data = {
    "Age": [20, 25, 30, 35, 40],
    "Salary": [20000, 30000, 40000, 50000, 60000],
    "Experience": [1, 2, 3, 4, 5]
}

df = pd.DataFrame(data)

print("Original Dataset:")
print(df)

# Create StandardScaler
scaler = StandardScaler()

# Apply Standardization
standardized_data = scaler.fit_transform(df)

# Convert into DataFrame
standardized_df = pd.DataFrame(
    standardized_data,
    columns=df.columns
)

print("\nStandardized Dataset:")
print(standardized_df)

# Check mean and standard deviation
print("\nMean After Standardization:")
print(standardized_df.mean())

print("\nStandard Deviation After Standardization:")
print(standardized_df.std(ddof=0))
```

Real-World Business Example:

Suppose a banking company wants to predict whether a customer will receive a loan.

The dataset contains:

- Customer Age
- Annual Salary
- Credit Score
- Existing Loan Amount
  
These features have different numerical ranges.

The company applies Standardization before training a Logistic Regression model.

This helps the model handle features on comparable scales and improves numerical optimization.

### Min-Max Scaling


Min-Max Scaling is a feature scaling technique used to transform numerical data into a fixed range, usually between 0 and 1.

It converts values with different ranges into a common scale while preserving their relative order.

Example:

Suppose a dataset contains:
- Age: 20 to 60
- Salary: ₹20,000 to ₹80,000
Since these features have different ranges, Min-Max Scaling converts both into values between 0 and 1.


##### Mathematical Concept – Min-Max Scaling

Suppose a dataset contains a numerical feature:

$$
X = \{X_1, X_2, X_3, \ldots, X_n\}
$$

Min-Max Scaling transforms each value into a fixed range, usually between 0 and 1.

##### Mathematical Formula

$$
X_{\text{scaled}} =
\frac{X-X_{\min}}{X_{\max}-X_{\min}}
$$

Where:

- $X$ = Original feature value.
- $X_{\min}$ = Minimum value of the feature.
- $X_{\max}$ = Maximum value of the feature.
- $X_{\text{scaled}}$ = Scaled value.

 Example

Suppose the Age feature contains:

$$
X = \{20, 30, 40, 50, 60\}
$$

Therefore:

$$
X_{\min}=20,\quad X_{\max}=60
$$

For the value $X=40$:

$$
X_{\text{scaled}} =
\frac{40-20}{60-20}
$$

$$
=\frac{20}{40}
$$

$$
=0.5
$$

##### Final Result

The original value 40 is transformed into 0.5.

##### Scaling Range

$$
0 \leq X_{\text{scaled}} \leq 1
$$

- Minimum value becomes 0.
- Maximum value becomes 1.
- Other values are transformed proportionally.

##### Importance in Machine Learning

Min-Max Scaling is useful for algorithms such as:

- K-Nearest Neighbors (KNN)
- K-Means Clustering
- Support Vector Machine (SVM)
- Neural Networks


```python

import pandas as pd
from sklearn.preprocessing import MinMaxScaler

# Step 1: Create a sample dataset
df = pd.DataFrame({
    "Age": [20, 30, 40, 50, 60],
    "Salary": [20000, 30000, 40000, 50000, 60000],
    "Experience": [1, 3, 5, 7, 9]
})

print("Original Dataset:")
print(df)

# Step 2: Create Min-Max Scaler
scaler = MinMaxScaler()

# Step 3: Apply Min-Max Scaling
scaled_data = scaler.fit_transform(df)

# Step 4: Convert scaled data into DataFrame
scaled_df = pd.DataFrame(
    scaled_data,
    columns=df.columns
)

print("\nDataset After Min-Max Scaling:")
print(scaled_df)

# Step 5: Check minimum and maximum values
print("\nMinimum Values:")
print(scaled_df.min())

print("\nMaximum Values:")
print(scaled_df.max())
```

### StandardScaler

StandardScaler is a feature scaling technique provided by Scikit-learn that transforms numerical features so that they have a mean of approximately 0 and a standard deviation of 1.

It is also called Z-score Standardization.

Standardization helps machine learning algorithms work effectively when numerical features have different measurement scales.


##### Mathematical Formula – Standardization

Standardization transforms a numerical feature using its mean and standard deviation.

$$
Z = \frac{X-\mu}{\sigma}
$$

Where:

- $X$ = Original feature value.
- $\mu$ = Mean of the feature.
- $\sigma$ = Standard deviation of the feature.
- $Z$ = Standardized value.

Example

Suppose:

$$
X = 60,\quad \mu = 40,\quad \sigma = 10
$$

Substituting the values:

$$
Z = \frac{60-40}{10}
$$

$$
Z = \frac{20}{10}
$$

$$
Z = 2
$$

##### Interpretation

The standardized value $Z=2$ means that the original value is 2 standard deviations above the mean.


```python

import pandas as pd
from sklearn.preprocessing import StandardScaler

# Step 1: Create a sample dataset
df = pd.DataFrame({
    "Age": [20, 30, 40, 50, 60],
    "Salary": [20000, 30000, 40000, 50000, 60000],
    "Experience": [1, 3, 5, 7, 9]
})

print("Original Dataset:")
print(df)

# Step 2: Create StandardScaler
scaler = StandardScaler()

# Step 3: Apply Standardization
scaled_data = scaler.fit_transform(df)

# Step 4: Convert into DataFrame
scaled_df = pd.DataFrame(
    scaled_data,
    columns=df.columns
)

print("\nDataset After Standardization:")
print(scaled_df)

# Step 5: Check mean and standard deviation
print("\nMean:")
print(scaled_df.mean().round(2))

print("\nStandard Deviation:")
print(scaled_df.std(ddof=0).round(2))
```

### MinMaxScaler

MinMaxScaler is a feature scaling technique used in Machine Learning to transform numerical features into a fixed range, usually between 0 and 1.

It scales the minimum value of a feature to 0 and the maximum value to 1. All other values are adjusted proportionally between these two values.

Purpose:

- Brings numerical features into the same range.
- Prevents large-scale features from dominating smaller-scale features.
- Improves the performance of distance-based ML algorithms.
- Helps algorithms such as KNN, K-Means and Neural Networks.

##### Mathematical Concept

Suppose a dataset contains a numerical feature:

$$
X = \{X_1, X_2, X_3, \ldots, X_n\}
$$

MinMaxScaler transforms each value using the following formula:

$$
X_{\text{scaled}} =
\frac{X - X_{\min}}{X_{\max} - X_{\min}}
$$

Where:

- $X$ = Original feature value.
- $X_{\min}$ = Minimum value of the feature.
- $X_{\max}$ = Maximum value of the feature.
- $X_{\text{scaled}}$ = Scaled value.

Real-World Business Example:

An e-commerce company uses customer data to predict whether customers will purchase a product.

Its dataset contains:

- Customer Age: 18–60
- Annual Income: ₹20,000–₹10,00,000
- Website Visits: 1–100
  
These features have different numerical ranges.

Using MinMaxScaler transforms all three features into the range 0 to 1, helping distance-based algorithms compare them more fairly.

Advantages:
   
1. Converts features into a common numerical range.
2. Easy to understand and implement.
3. Useful for distance-based algorithms.
4. Preserves the relative distribution of values.
   
 Disadvantages
 
1. Sensitive to outliers.
2. New data outside the original range may produce values below 0 or above 1.
3. Does not make the data normally distributed.

```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler

# Create sample dataset
data = {
    'Salary': [20000, 30000, 40000, 50000, 60000],
    'Age': [20, 25, 30, 35, 40]
}

df = pd.DataFrame(data)

print("Original Dataset:")
print(df)

# Create MinMaxScaler object
scaler = MinMaxScaler()

# Apply MinMaxScaler
scaled_data = scaler.fit_transform(df)

# Convert scaled data into DataFrame
scaled_df = pd.DataFrame(
    scaled_data,
    columns=df.columns
)

print("\nScaled Dataset:")
print(scaled_df)
```

### RobustScaler

RobustScaler is a feature scaling technique provided by Scikit-learn that scales numerical features using the median and interquartile range (IQR) instead of the mean and standard deviation.

It is particularly useful when a dataset contains outliers because the median and IQR are less affected by extreme values.

RobustScaler transforms numerical values so that:
- Median becomes approximately 0.
- Interquartile Range (IQR) becomes approximately 1.


##### Mathematical Concept

Suppose a dataset contains a numerical feature:

$$
X = \{X_1, X_2, X_3, \ldots, X_n\}
$$

RobustScaler transforms each value using the following formula:

$$
X_{\text{scaled}} =
\frac{X - X_{\text{median}}}{IQR}
$$

Where:

- $X$ = Original feature value.
- $X_{\text{median}}$ = Median of the feature.
- $IQR$ = Interquartile Range.
- $Q_1$ = First Quartile (25th percentile).
- $Q_3$ = Third Quartile (75th percentile).
- $X_{\text{scaled}}$ = Scaled feature value.

The Interquartile Range is calculated as:

$$
IQR = Q_3 - Q_1
$$


Advantages of RobustScaler:

1. Handles datasets containing outliers effectively.
2. Uses median instead of mean.
3. Uses IQR instead of standard deviation.
4. Reduces the influence of extreme values during scaling.
5. Useful for numerical features with skewed distributions.
   
Disadvantages of RobustScaler

1. Does not remove outliers.
2. Does not guarantee values between 0 and 1.
3. Does not make data normally distributed.
4. Not always necessary for datasets without significant outliers.

```python
import pandas as pd
from sklearn.preprocessing import RobustScaler

# Create sample dataset
data = {
    'Employee_ID': [101, 102, 103, 104, 105, 106, 107, 108],
    'Age': [22, 25, 28, 30, 35, 40, 45, 70],
    'Salary': [20000, 25000, 30000, 35000, 40000, 45000, 50000, 200000],
    'Experience': [1, 2, 3, 5, 7, 10, 12, 30]
}

df = pd.DataFrame(data)

print("Original Dataset:")
print(df)

# Select numerical features
features = ['Age', 'Salary', 'Experience']

# Create RobustScaler object
scaler = RobustScaler()

# Apply RobustScaler
df_scaled = df.copy()

df_scaled[features] = scaler.fit_transform(df[features])

print("\nRobust Scaled Dataset:")
print(df_scaled)

# Display median and IQR
print("\nMedian of Original Features:")
print(df[features].median())

print("\nIQR of Original Features:")
print(df[features].quantile(0.75) - df[features].quantile(0.25))

# Compare original and scaled salary
comparison = pd.DataFrame({
    'Original_Salary': df['Salary'],
    'Scaled_Salary': df_scaled['Salary']
})

print("\nSalary Comparison:")
print(comparison)
```

### MaxAbsScaler

MaxAbsScaler is a feature scaling technique provided by Scikit-learn that scales numerical features by dividing each value by the maximum absolute value of that feature.

It transforms values into a range between -1 and 1 without changing their signs.

It is particularly useful for sparse datasets because it preserves zero values


##### Mathematical Concept

Suppose a dataset contains a numerical feature:

$$
X = \{X_1, X_2, X_3, \ldots, X_n\}
$$

MaxAbsScaler transforms each value using the following formula:

$$
X_{\text{scaled}} =
\frac{X}{|X|_{\max}}
$$

Where:

- $X$ = Original feature value.
- $|X|_{\max}$ = Maximum absolute value of the feature.
- $X_{\text{scaled}}$ = Scaled feature value.


```python
import pandas as pd
from sklearn.preprocessing import MaxAbsScaler

# Create sample dataset
data = {
    'Employee_ID': [101, 102, 103, 104, 105, 106, 107, 108],
    'Age': [22, 25, 28, 30, 35, 40, 45, 70],
    'Salary': [20000, 25000, 30000, 35000,
               40000, 45000, 50000, 200000],
    'Experience': [1, 2, 3, 5, 7, 10, 12, 30]
}

df = pd.DataFrame(data)

print("Original Dataset:")
print(df)

# Select numerical features
features = ['Age', 'Salary', 'Experience']

# Create MaxAbsScaler object
scaler = MaxAbsScaler()

# Apply MaxAbsScaler
df_scaled = df.copy()

df_scaled[features] = scaler.fit_transform(df[features])

print("\nMaxAbs Scaled Dataset:")
print(df_scaled)

# Display maximum absolute values
print("\nMaximum Absolute Values:")
print(df[features].abs().max())

# Compare original and scaled salary
comparison = pd.DataFrame({
    'Original_Salary': df['Salary'],
    'Scaled_Salary': df_scaled['Salary']
})

print("\nSalary Comparison:")
print(comparison)
```

Advantages:

1. Scales numerical features between -1 and 1.
2. Preserves zero values.
3. Does not change positive and negative signs.
4. Useful for sparse datasets.
5. Simple and computationally efficient.
   
Disadvantages:
   
1. Sensitive to extreme values.
2. Does not reduce the influence of outliers.
3. Does not center data around zero.
4. May compress most values when extreme values exist.

```python
Comparison of Scaling Techniques:
    
Technique	      Based on	                             Result
MinMaxScaler	Minimum and Maximum	                   Usually 0 to 1
StandardScaler	Mean and Standard Deviation	Mean 0, Standard deviation 1
RobustScaler	Median and IQR	Median 0,             IQR 1
MaxAbsScaler	Maximum Absolute Value           	-1 to 1
```
