# Notebook 6 – Outlier Treatment

### What is an Outlier?

An Outlier is a data point that is significantly different from most other observations in a dataset

In simple words, an outlier is a value that is unusually high or unusually low compared with the remaining values

Outliers may occur because of:

- Data entry mistakes.
- Measurement errors.
- Incorrect data collection.
- Unusual customer behavior.
- Rare real-world events.
- Natural variations in the data.


##### Mathematical Concept

Suppose a dataset contains \(N\) observations:

$$
X = \{x_1,x_2,x_3,\ldots,x_N\}
$$

An observation is considered an **outlier** when it is unusually far from the overall distribution of the data.

One common method to detect outliers is the **IQR (Interquartile Range) Method**.

##### Interquartile Range (IQR)

$$
IQR = Q_3 - Q_1
$$

Where:

- \(Q_1\) = First quartile (25th percentile).
- \(Q_3\) = Third quartile (75th percentile).
- \(IQR\) = Interquartile Range.

##### Calculate the Lower and Upper Boundaries

$$
Lower = Q_1 - 1.5(IQR)
$$

$$
Upper = Q_3 + 1.5(IQR)
$$

##### Outlier Identification Condition

An observation is flagged as a potential outlier when:

$$
X < Lower
$$

or

$$
X > Upper
$$

**Note:** These boundaries identify potential outliers. They do not automatically determine whether a value is incorrect.


```python
import pandas as pd

# Sample dataset
df = pd.DataFrame({
    "customer_id": [
        "C101", "C102", "C103", "C104",
        "C105", "C106", "C107", "C108"
    ],
    "purchase_amount": [
        500, 600, 700, 800,
        650, 10000, 550, 750
    ]
})

# Calculate quartiles
Q1 = df["purchase_amount"].quantile(0.25)
Q3 = df["purchase_amount"].quantile(0.75)

# Calculate IQR
IQR = Q3 - Q1

# Calculate boundaries
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

# Identify potential outliers
df["is_outlier"] = (
    (df["purchase_amount"] < lower) |
    (df["purchase_amount"] > upper)
)

# Display results
print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)

print("\nPotential Outliers:")
print(df[df["is_outlier"]])
```

### Outlier vs Error

An Outlier is an observation that is unusually high or low compared with most other observations in a dataset

An Error is an incorrect value caused by problems such as data entry mistakes, measurement errors, or system failures

Important: Every error may create an outlier, but not every outlier is an error

```python
Real-World Business Example:
    
Suppose an e-commerce company stores customer purchase amounts

Customer	Purchase     Amount	Observation
    C101	₹500	      Normal
    C102	₹600	      Normal
    C103	₹700	      Normal
    C104	₹800	      Normal
    C105	₹10,000	      Valid Outlier
    C106	₹1,00,000	  Possible Error

Case 1: Valid Outlier

Customer C105 purchased expensive products worth ₹10,000.
- The transaction is genuine.
- The amount is unusually high.
- It is a valid extreme observation.
    
Decision: Retain the value.

Case 2: Data Entry Error

Customer C106 actually purchased products worth ₹10,000, but an employee accidentally entered ₹1,00,000.
    
- The amount is incorrect.
- It resulted from a data entry mistake.
- It creates an unusual observation.
    
Decision: Correct the value using the verified transaction record.
```

```python
How to Identify Outlier vs Error?

Unusual Observation Detected
Investigate the Original Record


Genuine
Valid Outlier

Retain the observation if it is relevant to the analysis.


Incorrect
Data Error

Correct it or mark it as missing if it cannot be verified.
```

```python
Difference Between Outlier and Error

Outlier	                                   Error

An unusually high or low value.	        An incorrect or invalid value.
May be genuine.	                        Represents a data quality problem.
Can occur naturally.	                Usually occurs due to mistakes or failures.
May contain valuable information.	    May mislead analysis.
Should be investigated.	                Should be corrected or handled appropriately.
```

```python
import pandas as pd

# Create sample dataset
df = pd.DataFrame({
    "customer_id": ["C101", "C102", "C103", "C104",
                    "C105", "C106", "C107", "C108"],
    "purchase_amount": [500, 600, 700, 800,
                        650, 10000, 550, 100000]
})

# Calculate Q1 and Q3
Q1 = df["purchase_amount"].quantile(0.25)
Q3 = df["purchase_amount"].quantile(0.75)

# Calculate IQR
IQR = Q3 - Q1

# Calculate boundaries
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

# Detect potential outliers
df["is_outlier"] = (
    (df["purchase_amount"] < lower) |
    (df["purchase_amount"] > upper)
)

# Classify observations after investigation
df["observation_type"] = "Normal"

# C106 is confirmed as a data entry error
df.loc[
    df["customer_id"] == "C106",
    "observation_type"
] = "Error"

# C105 is confirmed as a genuine extreme purchase
df.loc[
    df["customer_id"] == "C105",
    "observation_type"
] = "Valid Outlier"

# Display results
print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Boundary:", lower)
print("Upper Boundary:", upper)

print("\nOutlier vs Error Analysis:")
print(df)
```

### IQR Method (Interquartile Range Method)

The IQR Method is a statistical technique used to identify potential outliers in a dataset.

It identifies unusually low or high values by measuring the spread of the middle 50% of the observations.



##### Mathematical Formula

**Step 1: Calculate First Quartile (Q1)**

$$
Q_1 = 25^{th} \text{ Percentile}
$$

**Step 2: Calculate Third Quartile (Q3)**

$$
Q_3 = 75^{th} \text{ Percentile}
$$

**Step 3: Calculate Interquartile Range**

$$
IQR = Q_3 - Q_1
$$

**Step 4: Calculate Lower Boundary**

$$
Lower = Q_1 - 1.5 \times IQR
$$

**Step 5: Calculate Upper Boundary**

$$
Upper = Q_3 + 1.5 \times IQR
$$

**Step 6: Identify Outliers**

$$
X < Lower
$$

or

$$
X > Upper
$$


```python
import pandas as pd
import numpy as np

# Create dataset
df = pd.DataFrame({
    "purchase_amount": [
        100, 120, 130, 140, 150,
        160, 170, 180, 200, 5000
    ]
})

# Calculate Q1 and Q3
Q1 = df["purchase_amount"].quantile(0.25)
Q3 = df["purchase_amount"].quantile(0.75)

# Calculate IQR
IQR = Q3 - Q1

# Calculate boundaries
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

# Identify outliers
df["Outlier"] = np.where(
    (df["purchase_amount"] < lower) |
    (df["purchase_amount"] > upper),
    "Yes",
    "No"
)

# Display results
print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Boundary:", lower)
print("Upper Boundary:", upper)

print("\nOutlier Analysis:")
print(df)

print("\nDetected Outliers:")
print(df[df["Outlier"] == "Yes"])
```

Why Is the IQR Method Used?

- Identifies unusually high and low values
- Does not depend heavily on the mean
- Works well for skewed distributions compared with some standard-deviation-based approaches
- Helps identify potential data quality issues
- Supports decisions during EDA and preprocessing
  
An IQR-detected outlier is not automatically an error. Investigate whether it is a valid extreme observation before removing, capping, or transforming it.

### Z-Score Method


The Z-Score Method is a statistical technique used to identify potential outliers by measuring how far a data point is from the mean in terms of standard deviations

In simple words, it tells us how many standard deviations an observation is above or below the average

- A positive Z-score indicates a value above the mean
- A negative Z-score indicates a value below the mean
- A Z-score close to zero indicates a value near the mean


##### Mathematical Formula

The Z-score is calculated using:

$$
z = \frac{x-\mu}{\sigma}
$$

Where:

- \(z\) = Z-score.
- \(x\) = Individual observation.
- \(\mu\) = Mean of the dataset.
- \(\sigma\) = Standard deviation of the dataset.


##### Outlier Identification Condition

A common rule is to consider an observation a potential outlier when:

$$
|z| > 3
$$

This means:

$$
z < -3
$$

or

$$
z > 3
$$

Where:

- \(z < -3\) indicates a potential lower outlier.
- \(z > 3\) indicates a potential upper outlier.

**Note:** An observation with a Z-score below -3 or above +3 is flagged as a potential outlier. It should be investigated before deciding whether to remove or retain it.

```python
import pandas as pd
import numpy as np

# Create dataset
df = pd.DataFrame({
    "customer_id": [
        "C101", "C102", "C103",
        "C104", "C105", "C106"
    ],
    "purchase_amount": [
        800, 900, 1000, 1100, 1200, 5000
    ]
})

# Calculate mean
mean = df["purchase_amount"].mean()

# Calculate standard deviation
std = df["purchase_amount"].std(ddof=0)

# Calculate Z-score
df["z_score"] = (
    df["purchase_amount"] - mean
) / std

# Identify potential outliers
df["is_outlier"] = np.where(
    df["z_score"].abs() > 3,
    "Yes",
    "No"
)

# Display results
print("Mean:", mean)
print("Standard Deviation:", std)

print("\nZ-Score Analysis:")
print(df)

print("\nPotential Outliers:")
print(df[df["is_outlier"] == "Yes"])
```

```python
How to Interpret Z-Scores

Z-Score	      Interpretation

0	          Equal to the mean
+1	          One standard deviation above mean
-1	          One standard deviation below mean
+2	          Two standard deviations above mean
-2	          Two standard deviations below mean
+3.5	      Potential upper outlier
-3.5	      Potential lower outlier
```
