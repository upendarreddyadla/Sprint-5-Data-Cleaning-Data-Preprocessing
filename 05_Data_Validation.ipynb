# Notebook 5 – Data Validation

### What is Data Validation?

Data Validation is the process of checking whether data follows predefined rules and conditions before accepting or storing it in a dataset

It ensures that the data is correct, meaningful, consistent, and within acceptable limits

In simple words, Data Validation helps us identify invalid or incorrect values in a dataset

Why is Data Validation Important?

- Prevents incorrect data from entering a dataset.
- Maintains data quality and consistency.
- Identifies invalid values.
- Reduces errors during data analysis.
- Improves machine learning model performance.
- Ensures reliable business decisions.

```python
Real-World Examples:
    
Column	                Validation Rule	                       Invalid Example
    
Age	             Age must be greater than or equal to 0	      -5
Salary	         Salary cannot be negative	                  -10000
Percentage	     Must be between 0 and 100	                  120
Joining Date	Cannot be before Date of Birth	              Invalid date order
Gender	        Must belong to accepted categories	          Unknown
```


Data validation can be represented using logical conditions.

Age validation:

$$
Age \geq 0
$$

Salary validation:

$$
Salary \geq 0
$$

Percentage validation:

$$
0 \leq Percentage \leq 100
$$

Date validation:

$$
JoiningDate \geq DateOfBirth
$$

Gender validation:

$$
Gender \in \{Male,Female,Other\}
$$

A record is valid only when all required validation conditions are satisfied.


```python
import pandas as pd

# Step 1: Create Sample Dataset

data = {
    'Name': ['Rahul', 'Priya', 'Arjun', 'Sneha', 'Kiran'],
    'Age': [25, -5, 30, 22, 150],
    'Salary': [30000, 40000, -5000, 25000, 50000],
    'Percentage': [85, 120, 75, -10, 95],
    'DOB': ['2000-05-10', '1998-03-15', '1995-07-20',
            '2002-01-12', '1999-11-25'],
    'Joining_Date': ['2022-01-10', '2020-05-15',
                     '1990-01-01', '2023-06-10', '2021-08-20'],
    'Gender': ['Male', 'Female', 'Unknown', 'Female', 'Other']
}

df = pd.DataFrame(data)

# Step 2: Convert Date Columns

df['DOB'] = pd.to_datetime(df['DOB'])
df['Joining_Date'] = pd.to_datetime(df['Joining_Date'])

# Step 3: Apply Data Validation Rules

# Age Validation
df['Valid_Age'] = df['Age'] >= 0

# Salary Validation
df['Valid_Salary'] = df['Salary'] >= 0

# Percentage Validation
df['Valid_Percentage'] = (
    (df['Percentage'] >= 0) &
    (df['Percentage'] <= 100)
)

# Date Validation
df['Valid_Date'] = df['Joining_Date'] >= df['DOB']

# Gender Validation
accepted_gender = ['Male', 'Female', 'Other']

df['Valid_Gender'] = df['Gender'].isin(accepted_gender)

# Step 4: Overall Validation

validation_columns = [
    'Valid_Age',
    'Valid_Salary',
    'Valid_Percentage',
    'Valid_Date',
    'Valid_Gender'
]

df['Is_Valid'] = df[validation_columns].all(axis=1)

# Step 5: Display Complete Validation Results

print("DATA VALIDATION RESULTS")

print(df.to_string(index=False))

# Step 6: Identify Invalid Records

invalid_records = df[df['Is_Valid'] == False]

print("\nINVALID RECORDS:")
print(invalid_records[['Name', 'Age', 'Salary',
                       'Percentage', 'Gender']])

# Step 7: Validation Summary

total_records = len(df)
valid_records = df['Is_Valid'].sum()
invalid_count = (~df['Is_Valid']).sum()

print("\nVALIDATION SUMMARY")

print("Total Records:", total_records)
print("Valid Records:", valid_records)
print("Invalid Records:", invalid_count)

print("Validation Percentage:",
      round((valid_records / total_records) * 100, 2), "%")
```

### Range Validation

Range Validation is the process of checking whether numerical values fall within a predefined minimum and maximum range

It ensures that numerical data stays within acceptable limits

In simple words, Range Validation checks whether a value is neither below the minimum limit nor above the maximum limit

Why is Range Validation Important?
   
- Prevents unrealistic numerical values.
- Identifies values outside acceptable limits.
- Improves data quality.
- Reduces errors during data analysis.
- Helps maintain consistency in datasets.
- Improves machine learning model reliability.


## Mathematical Concept

Range Validation checks whether a value falls between a minimum and maximum limit.

The general formula is:

$$
Minimum \leq X \leq Maximum
$$

Where:

- $X$ = Input value.
- Minimum = Lowest acceptable value.
- Maximum = Highest acceptable value.


```python
import pandas as pd

# Step 1: Create Sample Dataset

data = {
    'Name': ['Rahul', 'Priya', 'Arjun', 'Sneha', 'Kiran'],
    'Age': [25, 15, 30, 105, 40],
    'Salary': [30000, 40000, -5000, 25000, 1200000],
    'Percentage': [85, 120, 75, -10, 95],
    'Rating': [4, 7, 3, 5, 0]
}

df = pd.DataFrame(data)

# Step 2: Apply Range Validation Rules

df['Valid_Age'] = df['Age'].between(18, 100)

df['Valid_Salary'] = df['Salary'].between(0, 1000000)

df['Valid_Percentage'] = df['Percentage'].between(0, 100)

df['Valid_Rating'] = df['Rating'].between(1, 5)

# Step 3: Overall Validation

validation_columns = [
    'Valid_Age',
    'Valid_Salary',
    'Valid_Percentage',
    'Valid_Rating'
]

df['Is_Valid'] = df[validation_columns].all(axis=1)

# Step 4: Display Results

print("RANGE VALIDATION RESULTS")

print(df.to_string(index=False))

# Step 5: Identify Invalid Records

invalid_records = df[df['Is_Valid'] == False]

print("\nINVALID RECORDS:")
print(invalid_records[['Name', 'Age', 'Salary',
                       'Percentage', 'Rating']])

# Step 6: Validation Summary

print("\nVALIDATION SUMMARY")

print("Total Records:", len(df))
print("Valid Records:", df['Is_Valid'].sum())
print("Invalid Records:", (~df['Is_Valid']).sum())

print("Validation Percentage:",
      round(df['Is_Valid'].mean() * 100, 2), "%")
```

### Type Validation

Type Validation is the process of checking whether a data value belongs to its expected data type

It ensures that each column contains the correct type of information before performing data analysis or machine learning

In simple words, Type Validation checks whether a value has the correct data type, such as integer, float, string, boolean, or date

2. Why is Type Validation Important?
   
- Prevents incorrect data types in a dataset.
- Identifies unexpected values.
- Avoids errors during mathematical calculations.
- Maintains data consistency.
- Improves data quality.
- Helps machine learning algorithms process data correctly.


##### Mathematical Concept

Type Validation checks whether a given value belongs to its expected data type.

The general condition is:

$$
Type(X)=ExpectedType
$$

Where:

- $X$ = Input value.
- $Type(X)$ = Actual data type of the value.
- $ExpectedType$ = Required data type.


```python
import pandas as pd
from datetime import datetime

# Step 1: Create Sample Dataset

data = {
    'Name': ['Rahul', 123, 'Arjun', 'Sneha', 'Kiran'],
    'Age': [25, 'Twenty', 30, 22, 40],
    'Salary': [30000.50, 40000.75, 'Unknown', 25000.00, 50000.25],
    'Is_Active': [True, False, 'Maybe', True, False],
    'Joining_Date': [
        pd.Timestamp('2022-01-10'),
        'Invalid Date',
        pd.Timestamp('2020-05-15'),
        pd.Timestamp('2023-06-10'),
        pd.Timestamp('2021-08-20')
    ]
}

df = pd.DataFrame(data, dtype=object)

# Step 2: Define Type Validation Rules

def validate_name(value):
    return isinstance(value, str)

def validate_age(value):
    return isinstance(value, int) and not isinstance(value, bool)

def validate_salary(value):
    return isinstance(value, (int, float)) and not isinstance(value, bool)

def validate_active(value):
    return isinstance(value, bool)

def validate_date(value):
    return isinstance(value, (pd.Timestamp, datetime))

# Step 3: Apply Type Validation

df['Valid_Name'] = df['Name'].apply(validate_name)

df['Valid_Age'] = df['Age'].apply(validate_age)

df['Valid_Salary'] = df['Salary'].apply(validate_salary)

df['Valid_Active'] = df['Is_Active'].apply(validate_active)

df['Valid_Date'] = df['Joining_Date'].apply(validate_date)

# Step 4: Overall Validation

validation_columns = [
    'Valid_Name',
    'Valid_Age',
    'Valid_Salary',
    'Valid_Active',
    'Valid_Date'
]

df['Is_Valid'] = df[validation_columns].all(axis=1)

# Step 5: Display Results

print("TYPE VALIDATION RESULTS")

print(df.to_string(index=False))

# Step 6: Identify Invalid Records

invalid_records = df[df['Is_Valid'] == False]

print("\nINVALID RECORDS:")

print(invalid_records[['Name', 'Age', 'Salary']])

# Step 7: Validation Summary

print("\nVALIDATION SUMMARY")

print("Total Records:", len(df))
print("Valid Records:", df['Is_Valid'].sum())
print("Invalid Records:", (~df['Is_Valid']).sum())

print("Validation Percentage:",
      round(df['Is_Valid'].mean() * 100, 2), "%")
```

### Format Validation

Format Validation is the process of checking whether data follows a predefined format or pattern

It ensures that values are entered in the correct structure before using them for data analysis or machine learning

In simple words, Format Validation checks whether data follows the expected pattern, such as email addresses, phone numbers, dates, PIN codes, and employee IDs

2. Why is Format Validation Important?
   
- Identifies incorrectly formatted values.
- Maintains data consistency.
- Prevents invalid information from entering a dataset.
- Improves data quality.
- Reduces errors during data processing.
- Helps maintain reliable business records.


##### Mathematical Concept

Format Validation checks whether a given value matches a predefined pattern.

The general condition is:

$$
Format(X)=ExpectedPattern
$$

Where:

- $X$ = Input value.
- $Format(X)$ = Actual format of the value.
- $ExpectedPattern$ = Required format or pattern.


```python
import pandas as pd
import re

# Step 1: Create Sample Dataset

data = {
    'Name': ['Rahul', 'Priya', 'Arjun', 'Sneha', 'Kiran'],
    'Email': [
        'rahul@gmail.com',
        'priya@gmail',
        'arjun@yahoo.com',
        'sneha123.com',
        'kiran@gmail.com'
    ],
    'Phone': [
        '9876543210',
        '98765',
        '9123456789',
        '12345678901',
        '8765432109'
    ],
    'PIN': ['500072', '5000AB', '500081', '12345', '500090'],
    'Employee_ID': [
        'EMP1001',
        'ABC1002',
        'EMP1003',
        'EMP12',
        'EMP1005'
    ]
}

df = pd.DataFrame(data)

# Step 2: Define Format Validation Functions

def validate_email(value):
    pattern = r'^[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}$'
    return bool(re.fullmatch(pattern, value))

def validate_phone(value):
    return bool(re.fullmatch(r'[6-9]\d{9}', value))

def validate_pin(value):
    return bool(re.fullmatch(r'[1-9]\d{5}', value))

def validate_employee_id(value):
    return bool(re.fullmatch(r'EMP\d{4}', value))

# Step 3: Apply Format Validation

df['Valid_Email'] = df['Email'].apply(validate_email)

df['Valid_Phone'] = df['Phone'].apply(validate_phone)

df['Valid_PIN'] = df['PIN'].apply(validate_pin)

df['Valid_Employee_ID'] = df['Employee_ID'].apply(
    validate_employee_id
)

# Step 4: Overall Validation

validation_columns = [
    'Valid_Email',
    'Valid_Phone',
    'Valid_PIN',
    'Valid_Employee_ID'
]

df['Is_Valid'] = df[validation_columns].all(axis=1)

# Step 5: Display Results

print("FORMAT VALIDATION RESULTS")

print(df.to_string(index=False))

# Step 6: Identify Invalid Records

invalid_records = df[df['Is_Valid'] == False]

print("\nINVALID RECORDS:")

print(invalid_records[['Name', 'Email', 'Phone',
                       'PIN', 'Employee_ID']])

# Step 7: Validation Summary

total_records = len(df)
valid_records = df['Is_Valid'].sum()
invalid_count = (~df['Is_Valid']).sum()

print("\nVALIDATION SUMMARY")

print("Total Records:", total_records)
print("Valid Records:", valid_records)
print("Invalid Records:", invalid_count)

print("Validation Percentage:",
      round((valid_records / total_records) * 100, 2), "%")
```

### Category Validation


Category Validation is a data validation technique used to check whether categorical values belong to a predefined set of allowed categories

It ensures that a dataset contains only permitted category values

Purpose of Category Validation

- Identifies unexpected category values.
- Prevents spelling mistakes in categorical data.
- Maintains consistency across records.
- Prevents invalid categories from entering the dataset.
- Improves data quality before Machine Learning.


##### Mathematical Concept

Suppose a feature contains a predefined set of allowed categories.

The validation condition is:

$$
Category(X) \in AllowedCategories
$$

Where:

- $X$ = Input value.
- $Category(X)$ = Category assigned to the input value.
- $AllowedCategories$ = Set of permitted categories.


```python
import pandas as pd

# Sample dataset
df = pd.DataFrame({
    "Customer": ["Rahul", "Priya", "Arjun", "Sneha", "Kiran"],
    "Membership": ["Gold", "Silver", "Platinum", "Bronze", "Premium"]
})

# Allowed categories
allowed_categories = ["Gold", "Silver", "Bronze"]

# Category validation
df["Is_Valid"] = df["Membership"].isin(allowed_categories)

# Display results
print(df)

# Validation percentage
valid_records = df["Is_Valid"].sum()
total_records = len(df)

validation_percentage = (valid_records / total_records) * 100

print("\nValid Records:", valid_records)
print("Total Records:", total_records)
print("Validation Percentage:", validation_percentage, "%")

# Display invalid records
print("\nInvalid Records:")
print(df[df["Is_Valid"] == False])
```

### Null Validation

Null Validation is a data validation technique used to check whether a dataset contains missing, null, or empty values in important columns.

It ensures that required fields contain valid values instead of missing data

Purpose of Null Validation

- Identifies missing values in a dataset.
- Ensures important columns are not left empty.
- Prevents incomplete records from entering the system.
- Improves data quality and reliability.
- Helps prepare clean data for Machine Learning.


##### Mathematical Concept

Suppose a dataset contains $N$ records.

Let:

- $N$ = Total number of records.
- $V$ = Number of non-null records.
- $M$ = Number of null records.

The relationship is:

$$
N = V + M
$$

The Null Validation condition is:

$$
X \neq Null
$$

Where:

- $X$ = Input value.
- $Null$ = Missing or null value.


```python
import pandas as pd
import numpy as np

# Sample dataset
df = pd.DataFrame({
    "Customer_ID": [101, 102, 103, 104, 105],
    "Name": ["Rahul", "Priya", "Arjun", "Sneha", "Kiran"],
    "Age": [25, np.nan, 30, 28, np.nan],
    "City": ["Hyderabad", "Chennai", np.nan, "Bangalore", "Mumbai"]
})

# Null validation
df["Is_Valid"] = df[["Age", "City"]].notna().all(axis=1)

# Display dataset
print("Validation Results:")
print(df)

# Calculate validation percentage
valid_records = df["Is_Valid"].sum()
total_records = len(df)

validation_percentage = (valid_records / total_records) * 100

print("\nValid Records:", valid_records)
print("Total Records:", total_records)
print("Null Records:", total_records - valid_records)
print("Validation Percentage:", validation_percentage, "%")

# Display invalid records
print("\nRecords Containing Null Values:")
print(df[df["Is_Valid"] == False])
```

### Business Rule Validation

Business Rule Validation is a data validation technique used to check whether data follows predefined business rules, conditions, and logical requirements.

It ensures that data is not only correctly formatted but also makes sense according to the organization's business requirements.
Purpose of Business Rule Validation
- Ensures data follows business requirements.
- Identifies logically incorrect records.
- Prevents invalid business transactions.
- Maintains data accuracy and consistency.
- Improves data quality before Machine Learning.


##### Mathematical Concept

Suppose a dataset contains a set of business rules.

Let:

- $X$ = Input data.
- $R(X)$ = Business rule applied to the input.
- True = Business rule is satisfied.
- False = Business rule is violated.

The validation condition is:

$$
R(X) = True
$$


```python
import pandas as pd

# Sample dataset
df = pd.DataFrame({
    "Order_ID": [101, 102, 103, 104, 105],
    "Amount": [5000, -1000, 3000, 2000, 1500],
    "Quantity": [2, 1, 0, 2, 1],
    "Discount": [500, 100, 200, 2500, 100]
})

# Business rule validation
df["Is_Valid"] = (
    (df["Amount"] > 0) &
    (df["Quantity"] >= 1) &
    (df["Discount"] <= df["Amount"])
)

# Display validation results
print("Business Rule Validation:")
print(df)

# Calculate validation percentage
valid_records = df["Is_Valid"].sum()
total_records = len(df)

validation_percentage = (valid_records / total_records) * 100

print("\nValid Records:", valid_records)
print("Total Records:", total_records)
print("Invalid Records:", total_records - valid_records)
print("Validation Percentage:", validation_percentage, "%")

# Display invalid records
print("\nInvalid Records:")
print(df[df["Is_Valid"] == False])
```

### Referential Validation

Referential Validation is a data validation technique used to check whether values in one dataset or table correctly refer to existing records in another dataset or table

It ensures that relationships between tables remain valid and consistent

Purpose of Referential Validation

- Ensures foreign key values exist in the referenced table.
- Identifies invalid references.
- Prevents orphan records.
- Maintains relationships between database tables.
- Improves data integrity and reliability.

Mathematical Concept:

Suppose two database tables are available:

- Parent Table = Referenced table.
- Child Table = Table containing foreign key values.
  
Let:

- \(P\) = Set of primary key values in the parent table.
- \(F\) = Set of foreign key values in the child table.
The referential validation condition is:

$$
F \subseteq P
$$

This means every foreign key value must exist in the parent table.

```python
import pandas as pd

# Parent table
customers = pd.DataFrame({
    "Customer_ID": [101, 102, 103, 104],
    "Customer_Name": ["Rahul", "Priya", "Arjun", "Sneha"]
})

# Child table
orders = pd.DataFrame({
    "Order_ID": [1001, 1002, 1003, 1004, 1005],
    "Customer_ID": [101, 102, 105, 103, 106],
    "Amount": [5000, 3000, 2000, 1500, 4000]
})

# Referential validation
orders["Is_Valid"] = orders["Customer_ID"].isin(
    customers["Customer_ID"]
)

# Display validation results
print("Referential Validation:")
print(orders)

# Calculate validation percentage
valid_records = orders["Is_Valid"].sum()
total_records = len(orders)

validation_percentage = (valid_records / total_records) * 100

print("\nValid Records:", valid_records)
print("Total Records:", total_records)
print("Invalid Records:", total_records - valid_records)
print("Validation Percentage:", validation_percentage, "%")

# Display invalid references
print("\nInvalid References:")
print(orders[orders["Is_Valid"] == False])
```

### Date Validation


Date Validation is a data validation technique used to check whether date values are valid, correctly formatted, and within an expected date range

It ensures that dates follow the required format and satisfy business rules

Purpose of Date Validation

- Identifies invalid date values.
- Checks whether dates follow the required format.
- Detects future or past dates when they are not allowed.
- Ensures date consistency across records.
- Prevents incorrect date values from entering a dataset.


##### Mathematical Concept

Suppose a dataset contains date values.

Let:

- $D$ = Input date.
- $D_{\min}$ = Minimum allowed date.
- $D_{\max}$ = Maximum allowed date.

The date validation condition is:

$$
D_{\min} \leq D \leq D_{\max}
$$

A date is valid when it falls within the permitted date range and follows the expected date format.


```python
import pandas as pd

# Sample dataset
df = pd.DataFrame({
    "Customer": ["Rahul", "Priya", "Arjun", "Sneha", "Kiran"],
    "Registration_Date": [
        "2025-01-15",
        "2025-05-20",
        "2025-15-10",
        "2026-12-01",
        "2025-08-25"
    ]
})

# Convert dates
df["Registration_Date"] = pd.to_datetime(
    df["Registration_Date"],
    format="%Y-%m-%d",
    errors="coerce"
)

# Define allowed date range
min_date = pd.Timestamp("2025-01-01")
max_date = pd.Timestamp("2025-12-31")

# Date validation
df["Is_Valid"] = (
    df["Registration_Date"].notna() &
    df["Registration_Date"].between(min_date, max_date)
)

# Display results
print("Date Validation:")
print(df)

# Calculate validation percentage
valid_records = df["Is_Valid"].sum()
total_records = len(df)

validation_percentage = (valid_records / total_records) * 100

print("\nValid Records:", valid_records)
print("Total Records:", total_records)
print("Invalid Records:", total_records - valid_records)
print("Validation Percentage:", validation_percentage, "%")

# Display invalid records
print("\nInvalid Records:")
print(df[df["Is_Valid"] == False])
```

### Constraint Validation

Constraint Validation is a data validation technique used to check whether data satisfies predefined restrictions, conditions, and constraints

It ensures that values follow specific rules before they are stored, processed, or used for analysis

Purpose of Constraint Validation

- Ensures data follows predefined restrictions.
- Prevents invalid values from entering a dataset.
- Maintains data accuracy and consistency.
- Identifies records that violate constraints.
- Improves database integrity and data quality


##### Mathematical Concept

Suppose a dataset contains values that must satisfy predefined constraints.

Let:

- $X$ = Input value.
- $C(X)$ = Constraint applied to the input.
- True = Constraint is satisfied.
- False = Constraint is violated.

The validation condition is:

$$
C(X) = True
$$


```python
import pandas as pd

# Sample dataset
df = pd.DataFrame({
    "Customer_ID": [101, 102, 103, 104, 101],
    "Age": [25, 17, 30, 28, 35],
    "Email": [
        "rahul@gmail.com",
        "priya@gmail.com",
        "arjun@gmail.com",
        "sneha@gmail.com",
        "kiran@gmail.com"
    ],
    "Balance": [5000, 3000, -500, 2000, 1000]
})

# Constraint validation
df["Is_Valid"] = (
    (df["Age"] >= 18) &
    (df["Customer_ID"].notna()) &
    (df["Email"].notna()) &
    (df["Email"].str.strip() != "") &
    (df["Balance"] >= 0) &
    (~df["Customer_ID"].duplicated(keep=False))
)

# Display validation results
print("Constraint Validation:")
print(df)

# Calculate validation percentage
valid_records = df["Is_Valid"].sum()
total_records = len(df)

validation_percentage = (valid_records / total_records) * 100

print("\nValid Records:", valid_records)
print("Total Records:", total_records)
print("Invalid Records:", total_records - valid_records)
print("Validation Percentage:", validation_percentage, "%")

# Display invalid records
print("\nInvalid Records:")
print(df[df["Is_Valid"] == False])
```
