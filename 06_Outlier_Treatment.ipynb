### What is an Outlier?

An Outlier is a data point that is significantly different from most other observations in a dataset

In simple words, an outlier is a value that is unusually high or unusually low compared with the remaining values

Outliers may occur because of:

- Data entry mistakes.
- Measurement errors.
- Incorrect data collection.
- Unusual customer behavior.
- Rare real-world events.
- Natural variations in the data.


##### Mathematical Concept

Suppose a dataset contains \(N\) observations:

$$
X = \{x_1,x_2,x_3,\ldots,x_N\}
$$

An observation is considered an **outlier** when it is unusually far from the overall distribution of the data.

One common method to detect outliers is the **IQR (Interquartile Range) Method**.

##### Interquartile Range (IQR)

$$
IQR = Q_3 - Q_1
$$

Where:

- \(Q_1\) = First quartile (25th percentile).
- \(Q_3\) = Third quartile (75th percentile).
- \(IQR\) = Interquartile Range.

##### Calculate the Lower and Upper Boundaries

$$
Lower = Q_1 - 1.5(IQR)
$$

$$
Upper = Q_3 + 1.5(IQR)
$$

##### Outlier Identification Condition

An observation is flagged as a potential outlier when:

$$
X < Lower
$$

or

$$
X > Upper
$$

**Note:** These boundaries identify potential outliers. They do not automatically determine whether a value is incorrect.


```python
import pandas as pd

# Sample dataset
df = pd.DataFrame({
    "customer_id": [
        "C101", "C102", "C103", "C104",
        "C105", "C106", "C107", "C108"
    ],
    "purchase_amount": [
        500, 600, 700, 800,
        650, 10000, 550, 750
    ]
})

# Calculate quartiles
Q1 = df["purchase_amount"].quantile(0.25)
Q3 = df["purchase_amount"].quantile(0.75)

# Calculate IQR
IQR = Q3 - Q1

# Calculate boundaries
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

# Identify potential outliers
df["is_outlier"] = (
    (df["purchase_amount"] < lower) |
    (df["purchase_amount"] > upper)
)

# Display results
print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)

print("\nPotential Outliers:")
print(df[df["is_outlier"]])
```

### Outlier vs Error

An Outlier is an observation that is unusually high or low compared with most other observations in a dataset

An Error is an incorrect value caused by problems such as data entry mistakes, measurement errors, or system failures

Important: Every error may create an outlier, but not every outlier is an error

```python
Real-World Business Example:
    
Suppose an e-commerce company stores customer purchase amounts

Customer	Purchase     Amount	Observation
    C101	₹500	      Normal
    C102	₹600	      Normal
    C103	₹700	      Normal
    C104	₹800	      Normal
    C105	₹10,000	      Valid Outlier
    C106	₹1,00,000	  Possible Error

Case 1: Valid Outlier

Customer C105 purchased expensive products worth ₹10,000.
- The transaction is genuine.
- The amount is unusually high.
- It is a valid extreme observation.
    
Decision: Retain the value.

Case 2: Data Entry Error

Customer C106 actually purchased products worth ₹10,000, but an employee accidentally entered ₹1,00,000.
    
- The amount is incorrect.
- It resulted from a data entry mistake.
- It creates an unusual observation.
    
Decision: Correct the value using the verified transaction record.
```

```python
How to Identify Outlier vs Error?

Unusual Observation Detected
Investigate the Original Record


Genuine
Valid Outlier

Retain the observation if it is relevant to the analysis.


Incorrect
Data Error

Correct it or mark it as missing if it cannot be verified.
```

```python
Difference Between Outlier and Error

Outlier	                                   Error

An unusually high or low value.	        An incorrect or invalid value.
May be genuine.	                        Represents a data quality problem.
Can occur naturally.	                Usually occurs due to mistakes or failures.
May contain valuable information.	    May mislead analysis.
Should be investigated.	                Should be corrected or handled appropriately.
```

```python
import pandas as pd

# Create sample dataset
df = pd.DataFrame({
    "customer_id": ["C101", "C102", "C103", "C104",
                    "C105", "C106", "C107", "C108"],
    "purchase_amount": [500, 600, 700, 800,
                        650, 10000, 550, 100000]
})

# Calculate Q1 and Q3
Q1 = df["purchase_amount"].quantile(0.25)
Q3 = df["purchase_amount"].quantile(0.75)

# Calculate IQR
IQR = Q3 - Q1

# Calculate boundaries
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

# Detect potential outliers
df["is_outlier"] = (
    (df["purchase_amount"] < lower) |
    (df["purchase_amount"] > upper)
)

# Classify observations after investigation
df["observation_type"] = "Normal"

# C106 is confirmed as a data entry error
df.loc[
    df["customer_id"] == "C106",
    "observation_type"
] = "Error"

# C105 is confirmed as a genuine extreme purchase
df.loc[
    df["customer_id"] == "C105",
    "observation_type"
] = "Valid Outlier"

# Display results
print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Boundary:", lower)
print("Upper Boundary:", upper)

print("\nOutlier vs Error Analysis:")
print(df)
```

### IQR Method (Interquartile Range Method)

The IQR Method is a statistical technique used to identify potential outliers in a dataset.

It identifies unusually low or high values by measuring the spread of the middle 50% of the observations.



##### Mathematical Formula

**Step 1: Calculate First Quartile (Q1)**

$$
Q_1 = 25^{th} \text{ Percentile}
$$

**Step 2: Calculate Third Quartile (Q3)**

$$
Q_3 = 75^{th} \text{ Percentile}
$$

**Step 3: Calculate Interquartile Range**

$$
IQR = Q_3 - Q_1
$$

**Step 4: Calculate Lower Boundary**

$$
Lower = Q_1 - 1.5 \times IQR
$$

**Step 5: Calculate Upper Boundary**

$$
Upper = Q_3 + 1.5 \times IQR
$$

**Step 6: Identify Outliers**

$$
X < Lower
$$

or

$$
X > Upper
$$


```python
import pandas as pd
import numpy as np

# Create dataset
df = pd.DataFrame({
    "purchase_amount": [
        100, 120, 130, 140, 150,
        160, 170, 180, 200, 5000
    ]
})

# Calculate Q1 and Q3
Q1 = df["purchase_amount"].quantile(0.25)
Q3 = df["purchase_amount"].quantile(0.75)

# Calculate IQR
IQR = Q3 - Q1

# Calculate boundaries
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

# Identify outliers
df["Outlier"] = np.where(
    (df["purchase_amount"] < lower) |
    (df["purchase_amount"] > upper),
    "Yes",
    "No"
)

# Display results
print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Boundary:", lower)
print("Upper Boundary:", upper)

print("\nOutlier Analysis:")
print(df)

print("\nDetected Outliers:")
print(df[df["Outlier"] == "Yes"])
```

Why Is the IQR Method Used?

- Identifies unusually high and low values
- Does not depend heavily on the mean
- Works well for skewed distributions compared with some standard-deviation-based approaches
- Helps identify potential data quality issues
- Supports decisions during EDA and preprocessing
  
An IQR-detected outlier is not automatically an error. Investigate whether it is a valid extreme observation before removing, capping, or transforming it.

### Z-Score Method


The Z-Score Method is a statistical technique used to identify potential outliers by measuring how far a data point is from the mean in terms of standard deviations

In simple words, it tells us how many standard deviations an observation is above or below the average

- A positive Z-score indicates a value above the mean
- A negative Z-score indicates a value below the mean
- A Z-score close to zero indicates a value near the mean


##### Mathematical Formula

The Z-score is calculated using:

$$
z = \frac{x-\mu}{\sigma}
$$

Where:

- \(z\) = Z-score.
- \(x\) = Individual observation.
- \(\mu\) = Mean of the dataset.
- \(\sigma\) = Standard deviation of the dataset.


##### Outlier Identification Condition

A common rule is to consider an observation a potential outlier when:

$$
|z| > 3
$$

This means:

$$
z < -3
$$

or

$$
z > 3
$$

Where:

- \(z < -3\) indicates a potential lower outlier.
- \(z > 3\) indicates a potential upper outlier.

**Note:** An observation with a Z-score below -3 or above +3 is flagged as a potential outlier. It should be investigated before deciding whether to remove or retain it.

```python
import pandas as pd
import numpy as np

# Create dataset
df = pd.DataFrame({
    "customer_id": [
        "C101", "C102", "C103",
        "C104", "C105", "C106"
    ],
    "purchase_amount": [
        800, 900, 1000, 1100, 1200, 5000
    ]
})

# Calculate mean
mean = df["purchase_amount"].mean()

# Calculate standard deviation
std = df["purchase_amount"].std(ddof=0)

# Calculate Z-score
df["z_score"] = (
    df["purchase_amount"] - mean
) / std

# Identify potential outliers
df["is_outlier"] = np.where(
    df["z_score"].abs() > 3,
    "Yes",
    "No"
)

# Display results
print("Mean:", mean)
print("Standard Deviation:", std)

print("\nZ-Score Analysis:")
print(df)

print("\nPotential Outliers:")
print(df[df["is_outlier"] == "Yes"])
```

```python
How to Interpret Z-Scores

Z-Score	      Interpretation

0	          Equal to the mean
+1	          One standard deviation above mean
-1	          One standard deviation below mean
+2	          Two standard deviations above mean
-2	          Two standard deviations below mean
+3.5	      Potential upper outlier
-3.5	      Potential lower outlier
```

### Percentile Method


The Percentile Method is an outlier detection technique that identifies unusually low or high values based on specified percentiles of a dataset

For example, we can use the 5th percentile and 95th percentile as lower and upper boundaries

##### Mathematical Formula

Suppose a dataset contains \(N\) observations.

The lower and upper boundaries are calculated using percentiles.

Step 1: Calculate the Lower Boundary

$$
Lower = P_5
$$

Step 2: Calculate the Upper Boundary

$$
Upper = P_{95}
$$

Where:

- \(P_5\) = 5th percentile of the dataset.
- \(P_{95}\) = 95th percentile of the dataset.

Step 3: Outlier Identification Condition

An observation is flagged as a potential outlier when:

$$
X < P_5
$$

or

$$
X > P_{95}
$$

Where:

- \(X < P_5\) indicates a potential lower outlier.
- \(X > P_{95}\) indicates a potential upper outlier.

**Note:** The 5th and 95th percentiles are example thresholds. The appropriate percentiles depend on the dataset and business problem.

Real-World Business Example:

Suppose an e-commerce company analyzes customer purchase amounts.
- 5th percentile = ₹100
- 95th percentile = ₹5,000
  
Therefore:

- Purchases below ₹100 are flagged as potential lower outliers.
- Purchases above ₹5,000 are flagged as potential upper outliers.

```python
import pandas as pd

# Create sample dataset
df = pd.DataFrame({
    "customer_id": ["C101", "C102", "C103", "C104", "C105",
                    "C106", "C107", "C108", "C109", "C110"],
    "purchase_amount": [50, 100, 200, 300, 400,
                        500, 600, 700, 1000, 10000]
})

# Calculate percentile boundaries
lower = df["purchase_amount"].quantile(0.05)
upper = df["purchase_amount"].quantile(0.95)

# Identify potential outliers
df["is_outlier"] = (
    (df["purchase_amount"] < lower) |
    (df["purchase_amount"] > upper)
)

print("Lower Boundary:", lower)
print("Upper Boundary:", upper)

print("\nOutlier Analysis:")
print(df)

print("\nPotential Outliers:")
print(df[df["is_outlier"]])
```

Why is the Percentile Method useful?

- Easy to understand and implement.
- Does not require normally distributed data.
- Useful for skewed datasets.
- Helps identify unusually low and high observations.

### Winsorization

Winsorization is an outlier-handling technique that reduces the effect of extreme values by replacing them with predefined percentile boundary values instead of removing them

For example, if the upper boundary is ₹5,000 and a customer's purchase amount is ₹10,000, Winsorization replaces ₹10,000 with ₹5,000

##### Mathematical Formula

Suppose a dataset contains \(N\) observations.

Let:

- \(X\) = Original observation.
- \(P_5\) = 5th percentile.
- \(P_{95}\) = 95th percentile.
- \(X_w\) = Winsorized observation.

The Winsorization formula is:

$$
X_w =
\begin{cases}
P_5, & X < P_5 \\
X, & P_5 \leq X \leq P_{95} \\
P_{95}, & X > P_{95}
\end{cases}
$$

Explanation

- Values below the 5th percentile are replaced with the 5th percentile.
- Values within the boundaries remain unchanged.
- Values above the 95th percentile are replaced with the 95th percentile.

```python
Real-World Business Example:
    
Suppose an e-commerce company analyzes customer purchase amounts

- Lower boundary (5th percentile) = ₹100
- Upper boundary (95th percentile) = ₹5,000

Original amount	           Winsorized amount

₹50  	                       ₹100
₹500	                       ₹500
₹2,000	                       ₹2,000
₹5,000	                       ₹5,000
₹10,000	                       ₹5,000

The extreme values are capped rather than deleted
```

```python
import pandas as pd

# Create sample dataset
df = pd.DataFrame({
    "customer_id": ["C101", "C102", "C103", "C104", "C105",
                    "C106", "C107", "C108", "C109", "C110"],
    "purchase_amount": [50, 100, 200, 300, 400,
                        500, 600, 700, 1000, 10000]
})

# Calculate percentile boundaries
lower = df["purchase_amount"].quantile(0.05)
upper = df["purchase_amount"].quantile(0.95)

# Apply Winsorization
df["winsorized_amount"] = df["purchase_amount"].clip(
    lower=lower,
    upper=upper
)

print("Lower Boundary:", lower)
print("Upper Boundary:", upper)

print("\nWinsorization Results:")
print(df)
```

Why is Winsorization useful?

- Reduces the influence of extreme values
- Retains all records in the dataset
- Can make statistical summaries such as the mean less sensitive to extremes
- Useful when extreme values are valid but disproportionately influential

### Capping


Capping is an outlier treatment technique in which extreme values are replaced with predefined lower and upper boundary values instead of deleting the records

For example, if a customer's purchase amount is ₹10,000 and the upper boundary is ₹5,000, capping replaces ₹10,000 with ₹5,000.

##### Mathematical Formula

Suppose a dataset contains \(N\) observations.

Let:

- \(X\) = Original observation.
- \(L\) = Lower boundary.
- \(U\) = Upper boundary.
- \(X_c\) = Capped observation.

The capping formula is:

$$
X_c =
\begin{cases}
L, & X < L \\
X, & L \leq X \leq U \\
U, & X > U
\end{cases}
$$

Explanation

- Values below the lower boundary are replaced with the lower boundary.
- Values within the boundaries remain unchanged.
- Values above the upper boundary are replaced with the upper boundary.

```python
Real-World Business Example:
    
Suppose an e-commerce company analyzes customer purchase amounts

- Lower boundary = ₹100
- Upper boundary = ₹5,000

Original amount	                Capped amount
₹50	                               ₹100
₹500	                           ₹500
₹2,000	                           ₹2,000
₹5,000	                           ₹5,000
₹10,000                            ₹5,000

The company retains all customer records but limits extreme purchase values
```

```python
import pandas as pd

# Create sample dataset
df = pd.DataFrame({
    "customer_id": ["C101", "C102", "C103", "C104", "C105",
                    "C106", "C107", "C108", "C109", "C110"],
    "purchase_amount": [50, 100, 200, 300, 400,
                        500, 600, 700, 1000, 10000]
})

# Define lower and upper boundaries
lower = 100
upper = 5000

# Apply capping
df["capped_amount"] = df["purchase_amount"].clip(
    lower=lower,
    upper=upper
)

print("Original and Capped Values:")
print(df)

# Display capped records
print("\nRecords affected by Capping:")
print(df[df["purchase_amount"] != df["capped_amount"]])
```

Why is Capping useful?

- Reduces the influence of extreme values.
- Retains all records in the dataset.
- Prevents extremely large or small values from dominating statistical calculations.
- Can improve model stability when extreme values are influential

### Flooring

Flooring is an outlier treatment technique in which unusually low values are replaced with a predefined lower boundary instead of deleting the records.

For example, if a customer's purchase amount is ₹20 and the lower boundary is ₹100, flooring replaces ₹20 with ₹100.

Flooring mainly handles lower-end outliers

##### Mathematical Formula

Suppose a dataset contains \(N\) observations.

Let:

- \(X\) = Original observation.
- \(L\) = Lower boundary.
- \(X_f\) = Floored observation.

The flooring formula is:

$$
X_f =
\begin{cases}
L, & X < L \\
X, & X \geq L
\end{cases}
$$

##### Outlier Identification Condition

A value is floored when:

$$
X < L
$$

Explanation

- Values below the lower boundary are replaced with the lower boundary.
- Values equal to or above the lower boundary remain unchanged.

```python
Real-World Business Example:
    
Suppose an e-commerce company analyzes customer purchase amounts.

- Lower boundary = ₹100

Original amount	        Floored amount
    ₹20	                     ₹100
    ₹50	                     ₹100
    ₹80	                     ₹100
    ₹100	                 ₹100
    ₹500	                 ₹500
    ₹2,000	                 ₹2,000

The company replaces unusually low purchase amounts with ₹100 while retaining all customer records
```

```python
import pandas as pd

# Create sample dataset
df = pd.DataFrame({
    "customer_id": ["C101", "C102", "C103", "C104", "C105",
                    "C106", "C107", "C108"],
    "purchase_amount": [20, 50, 80, 100, 200, 500, 1000, 2000]
})

# Define lower boundary
lower = 100

# Apply flooring
df["floored_amount"] = df["purchase_amount"].clip(
    lower=lower
)

print("Original and Floored Values:")
print(df)

# Display affected records
print("\nRecords affected by Flooring:")
print(df[df["purchase_amount"] != df["floored_amount"]])
```

Why is Flooring useful?

- Handles unusually low values.
- Reduces the influence of lower-end extreme observations.
- Retains all records.
- Can improve model stability when very low values distort statistical analysis.

### Transformation

Transformation is a data preprocessing technique that changes the scale or distribution of numerical values using mathematical functions

It helps reduce the influence of extreme values and can make highly skewed data more suitable for statistical analysis and machine learning

Common transformation techniques include:

1. Log Transformation
2. Square Root Transformation
3. Cube Root Transformation
4. Box-Cox Transformation
5. Yeo-Johnson Transformation

##### Mathematical Formula

1. Log Transformation

The logarithm of an observation is calculated as:

$$
X' = \log(1+X)
$$

Where:

- \(X\) = Original observation.
- \(X'\) = Transformed observation.
- \(\log\) = Natural logarithm in this example.

2. Square Root Transformation

$$
X' = \sqrt{X}
$$

3. Cube Root Transformation

$$
X' = \sqrt[3]{X}
$$

Explanation

- Log transformation compresses large values more strongly.
- Square root transformation reduces the scale of positive values.
- Cube root transformation can also handle negative values.

```python
Real-World Business Example:
    
Suppose an e-commerce company analyzes customer purchase amounts

Customer	Originalpurchase amount
    C101	₹100
    C102	₹200
    C103	₹300
    C104	₹500
    C105	₹1,000
    C106	₹10,000

The value ₹10,000 is much larger than the other purchase amounts

Applying log transformation reduces the difference in scale between small and large values
```

```python
import pandas as pd
import numpy as np

# Create sample dataset
df = pd.DataFrame({
    "customer_id": ["C101", "C102", "C103", "C104", "C105", "C106"],
    "purchase_amount": [100, 200, 300, 500, 1000, 10000]
})

# Apply transformations
df["log_transformed"] = np.log1p(df["purchase_amount"])

df["sqrt_transformed"] = np.sqrt(df["purchase_amount"])

df["cbrt_transformed"] = np.cbrt(df["purchase_amount"])

print("Original and Transformed Values:")
print(df.round(2))
```

Why is Transformation useful

- Reduces positive skewness in some datasets.
- Compresses the scale of large numerical values.
- Can reduce the influence of extreme observations.
- May improve the performance of models that are sensitive to data distributions.
- Helps make relationships easier to analyze

### Removing Outliers

Removing Outliers is a data preprocessing technique in which unusual or incorrect observations are deleted from a dataset when there is sufficient justification.

For example, suppose an e-commerce company records a customer purchase amount of ₹10,00,000, but investigation confirms that the actual amount was ₹10,000 due to a data-entry error.

The incorrect record may be removed or corrected, depending on the available source information.

##### Mathematical Formula

Suppose a dataset contains \(N\) observations.

Let:

- \(N\) = Total number of observations.
- \(R\) = Number of outlier records removed.
- \(U\) = Number of remaining observations.

The remaining number of observations is:

$$
U = N - R
$$

Example

Suppose an e-commerce company has 1,000 customer records.

- Total records = 1,000
- Outlier records removed = 20

$$
U = 1000 - 20
$$

$$
U = 980
$$

Therefore, 980 records remain after removing 20 records.

```python
Real-World Business Example:
    
Suppose an e-commerce company analyzes customer purchase amounts

Customer	Purchase    amount	Decision
C101	     ₹500	    Retain
C102	    ₹700	    Retain
C103	   ₹1,000	    Retain
C104	   ₹1,200	    Retain
C105	  ₹10,00,000	Investigate

Suppose the company investigates C105 and confirms that ₹10,00,000 is an incorrect data-entry value.

The company may remove that record if the correct value cannot be recovered
```

```python
import pandas as pd

# Create sample dataset
df = pd.DataFrame({
    "customer_id": ["C101", "C102", "C103", "C104", "C105",
                    "C106", "C107", "C108"],
    "purchase_amount": [500, 700, 1000, 1200, 1000000,
                        1500, 1800, 2000]
})

# Calculate IQR boundaries
Q1 = df["purchase_amount"].quantile(0.25)
Q3 = df["purchase_amount"].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

# Identify potential outliers
outliers = df[
    (df["purchase_amount"] < lower) |
    (df["purchase_amount"] > upper)
]

# Remove potential outliers
cleaned_df = df[
    (df["purchase_amount"] >= lower) &
    (df["purchase_amount"] <= upper)
].copy()

print("Original Dataset:")
print(df)

print("\nPotential Outliers:")
print(outliers)

print("\nDataset After Removing Outliers:")
print(cleaned_df)

print("\nTotal Records:", len(df))
print("Removed Records:", len(outliers))
print("Remaining Records:", len(cleaned_df))
```

`Why is Removing Outliers useful?

- Removes confirmed incorrect observations.
- Reduces the influence of extreme errors.
- Can improve statistical analysis.
- May improve model performance when extreme observations are invalid and harmful.

### Retaining Outliers

Retaining Outliers is a data preprocessing decision in which unusual or extreme observations are kept in the dataset because they represent genuine and meaningful information.

Not every outlier is an error. Some extreme values contain important business information.

For example, an e-commerce company records a customer purchase of ₹1,00,000. After investigation, the company confirms that the transaction is genuine.

Instead of removing or modifying the value, the company retains it.

##### Mathematical Concept

Suppose a dataset contains \(N\) observations.

Let:

- \(N\) = Total number of observations.
- \(O\) = Number of identified outliers.
- \(R\) = Number of outliers retained.
- \(U\) = Total observations after retaining outliers.

When all identified outliers are genuine and retained:

$$
R = O
$$

The total number of observations remains unchanged:

$$
U = N
$$

Example

Suppose a company has 1,000 customer records.

- Total records = 1,000
- Identified outliers = 20
- Genuine outliers retained = 20

Therefore:

$$
U = 1000
$$

All 1,000 records remain in the dataset.

```python
Real-World Business Example:
    
Suppose an e-commerce company analyzes customer purchase amounts.
    
Customer	Purchase      amountDecision

C101	     ₹500	     Retain
C102	    ₹700	     Retain
C103	    ₹1,000	     Retain
C104	    ₹1,200	     Retain
C105	   ₹1,00,000	 Retain

The company investigates C105 and confirms that the purchase is genuine.

Therefore, the company keeps the ₹1,00,000 transaction without modifying it.
```

```python
import pandas as pd

# Create sample dataset
df = pd.DataFrame({
    "customer_id": ["C101", "C102", "C103", "C104", "C105",
                    "C106", "C107", "C108"],
    "purchase_amount": [500, 700, 1000, 1200, 100000,
                        1500, 1800, 2000]
})

# Calculate IQR boundaries
Q1 = df["purchase_amount"].quantile(0.25)
Q3 = df["purchase_amount"].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

# Identify potential outliers
df["is_outlier"] = (
    (df["purchase_amount"] < lower) |
    (df["purchase_amount"] > upper)
)

# Retain all records
retained_df = df.copy()

print("Potential Outlier Records:")
print(retained_df[retained_df["is_outlier"]])

print("\nDataset After Retaining Outliers:")
print(retained_df)

print("\nOriginal Records:", len(df))
print("Remaining Records:", len(retained_df))
```

Why is Retaining Outliers useful?

- Preserves genuine business transactions
- Avoids losing valuable information
- Prevents unnecessary data deletion
- Helps understand unusual customer behavior
- Important for fraud detection and financial risk analysis
