### What is Categorical Encoding?

Categorical Encoding is a data preprocessing technique used to convert categorical (text-based) data into numerical values so that Machine Learning algorithms can understand and process it

Most Machine Learning algorithms work with numerical data. Therefore, categorical values such as Male, Female, Hyderabad, and Chennai must be converted into numerical representations

##### Mathematical Concept

Suppose a categorical feature contains different categories:

$$
C = \{C_1,C_2,C_3,\ldots,C_n\}
$$

Categorical Encoding converts these categories into numerical representations:

$$
f(C_i)=X_i
$$

Where:

- $C_i$ = Original categorical value.
- $f$ = Encoding function.
- $X_i$ = Encoded numerical value.
- $n$ = Number of unique categories.

 Example

Suppose a Gender feature contains two categories:

$$
C = \{\text{Male},\text{Female}\}
$$

Using Label Encoding:

$$
f(\text{Male})=0
$$

$$
f(\text{Female})=1
$$

Therefore, the encoded values are:

$$
X = \{0,1\}
$$

The categorical text values are converted into numerical representations
that can be used by suitable Machine Learning algorithms.

Why Is Categorical Encoding Important?

1. Converts text data into numerical data
2. Makes categorical features usable by ML algorithms
3. Helps prepare datasets for model training
4. Supports mathematical calculations on encoded features
5. Helps ML models process customer, product, and business information

```python
Types of Categorical Encoding

1. One-Hot Encoding

Creates separate binary columns for each category.

Example: Male → [1, 0], Female → [0, 1]

2. Label Encoding

Assigns a unique integer to each category.

Example: Red → 0, Blue → 1, Green → 2

3. Ordinal Encoding

Assigns numerical values based on category order.

Example: Low → 1, Medium → 2, High → 3

4. Frequency Encoding

Replaces each category with its frequency in the dataset.

Example: Hyderabad → 50, Chennai → 30

5. Target Encoding

Replaces each category with a statistic calculated from the target variable, commonly its mean.
```

```python
Real-World Business Example:
    
Suppose an online shopping company wants to predict whether a customer will purchase a product.

Its dataset contains:

Age	Gender	Membership	Purchased
25	Male	Gold	      1
32	Female	Silver	      0
28	Male	Bronze	      1
40	Female	Gold	      0

The ML algorithm cannot directly process the text values Male, Female, Gold, Silver, and Bronze in most standard numerical workflows.

The company applies categorical encoding to convert these features into numerical representations before training the model.
```

```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder

# Create dataset
df = pd.DataFrame({
    "Gender": ["Male", "Female", "Male", "Female"],
    "City": ["Hyderabad", "Chennai", "Bangalore", "Hyderabad"]
})

print("Original Dataset:")
print(df)

# Apply Label Encoding
encoder = LabelEncoder()

df["Gender_Encoded"] = encoder.fit_transform(df["Gender"])
df["City_Encoded"] = encoder.fit_transform(df["City"])

print("\nEncoded Dataset:")
print(df)
```

### Nominal Variables

Nominal Variables are categorical variables that contain different categories or groups without any natural order or ranking

In simple words, nominal variables represent names or labels rather than numerical quantities or ordered values

Examples

- Gender: Male, Female
- City: Hyderabad, Chennai, Bangalore
- Color: Red, Blue, Green
- Payment Method: Cash, UPI, Credit Card
- Product Category: Electronics, Clothing, Furniture
  
There is no natural ranking between these categories.

##### Mathematical Concept

Suppose a categorical feature contains different nominal categories.

$$
C = \{C_1,C_2,C_3,\ldots,C_n\}
$$

Where:

- $C$ = Categorical feature.
- $C_i$ = Individual category.
- $n$ = Total number of unique categories.

Example

Suppose a Color feature contains three nominal categories:

$$
C = \{\text{Red},\text{Blue},\text{Green}\}
$$

These categories have no natural order.

Therefore:

$$
\text{Red} \neq \text{Blue} \neq \text{Green}
$$

However, this does not mean one category is greater or smaller than another.

Conclusion

Nominal variables represent categories without any
natural ranking or ordering.

```python
Real-World Business Example:
    
Suppose an e-commerce company collects customer payment information

Customer	Payment Method
C101	     UPI
C102	     Cash
C103	     Credit Card
C104	     UPI
C105	     Debit Card

Here, Payment Method is a nominal variable because payment methods do not have a natural ranking.
```

```python
import pandas as pd

# Create sample dataset
df = pd.DataFrame({
    "Customer_ID": [101, 102, 103, 104, 105],
    "Payment_Method": [
        "UPI",
        "Cash",
        "Credit Card",
        "UPI",
        "Debit Card"
    ]
})

# Display dataset
print(df)

# Identify unique nominal categories
print("\nUnique Payment Methods:")
print(df["Payment_Method"].unique())

# Count each category
print("\nCategory Counts:")
print(df["Payment_Method"].value_counts())

# Count unique categories
print("\nTotal Unique Categories:")
print(df["Payment_Method"].nunique())
```
