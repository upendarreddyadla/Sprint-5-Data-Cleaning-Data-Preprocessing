### What is Data Preprocessing?

Data Preprocessing is the process of converting raw, incomplete, inconsistent, or unstructured data into a clean, organized, and suitable format for data analysis and Machine Learning.

Real-world datasets often contain missing values, duplicate records, incorrect data types, inconsistent categories, and outliers.

Data preprocessing helps us handle these issues before building a Machine Learning model.

Why is Data Preprocessing Important?

Data preprocessing is important because:

1. It improves data quality
2. It handles missing values
3. It removes unnecessary duplicate records
4. It corrects inconsistent and invalid values
5. It converts categorical data into numerical format
6. It helps features have appropriate scales
7. It can improve Machine Learning model performance
8. It reduces errors during model training

Important: Poor-quality data can lead to misleading insights and unreliable predictions.


Main Steps in Data Preprocessing

1. Raw Dataset

Original data collected from different sources


2. Data Cleaning

Missing values, duplicates, invalid values


3. Data Transformation

Encoding, scaling, data type conversion


4. Feature Engineering

Create useful features from existing data


5. Preprocessed Dataset

Ready for Machine Learning model training

````python
Types of Data Preprocessing:
    
Technique	              Purpose

Missing Value Handling	 Fill or remove missing values
Duplicate Removal	     Remove repeated records
Outlier Treatment	     Investigate and handle unusual values
Encoding	             Convert categorical data into numerical values
Feature Scaling	         Bring numerical features to appropriate scales
Data Type Conversion	 Convert columns to correct data types
Feature Engineering	     Create new meaningful features
Data Splitting	         Divide data into training and testing sets

Example:

Suppose we have a customer dataset:

CustomerID	Age	Salary	Gender	Purchased
101	        25	30000	Male	Yes
102      	NaN	40000	Female	No
103	        35	-5000	male	Yes
103	        35	-5000	male	Yes

Problems in this dataset:
    
- Missing value in Age
- Invalid negative Salary
- Inconsistent Gender categories (Male and male)
- Duplicate record for CustomerID 103
    
After preprocessing:
    
CustomerID	Age	Salary	Gender	Purchased
  101	    25	30000	Male	Yes
  102	    30	40000	Female	No
  103	    35	35000	Male	Yes

The cleaned dataset is now more suitable for analysis and model development

Note: The values used to fix the missing age and invalid salary are illustrative
````

````python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "Age": [25, np.nan, 35, 35],
    "Salary": [30000, 40000, -5000, -5000],
    "Gender": ["Male", "Female", "male", "male"]
})

df
````

````python
df["Age"] = df["Age"].fillna(df["Age"].median())         #Handle missing values

df = df.drop_duplicates()                               #Remove duplicate records

df["Gender"] = df["Gender"].str.title()                 #Correct inconsistent categories

df["Salary"] = df["Salary"].mask(
    df["Salary"] < 0,
    np.nan
)

df
````

### Why is Data Preprocessing required?

Data preprocessing is required because real-world datasets are often incomplete, inconsistent, noisy, and unstructured

Before performing data analysis or training a Machine Learning model, we need to clean and transform the data into a suitable format

The main objective is to improve data quality and make the data reliable for analysis and prediction

````python
Reasons for Data Preprocessing:
    
Reason	                              Explanation

1. Handle Missing Values	   Missing data can affect statistical calculations and model training.
2. Remove Duplicate Records	   Duplicate records can create biased results.
3. Handle Invalid Values	   Incorrect values, such as negative age or salary, can produce misleading results.
4. Handle Inconsistent        Data	Different formats or categories must be standardized.
5. Handle Outliers	          Extreme values may affect statistical analysis and some ML models.
6. Convert Categorical Data	  Many ML algorithms require numerical input rather than text categories.
7. Scale Numerical Features	  Features with different scales can affect distance-based and gradient-based algorithms.
8. Improve Model Performance  Better-quality data can help models learn meaningful patterns.
9. Prevent Data Leakage	      Proper preprocessing ensures that test data does not influence training transformations.


Example:

Consider a customer dataset used to predict whether a customer will purchase a product

Age	Salary	Gender	Purchased
25	30000	Male	Yes
NaN	40000	Female	No
35	-5000	male	Yes

Problems

- Age contains a missing value.
- Salary contains an invalid negative value.
- Gender has inconsistent capitalization.
- The data may not be directly suitable for ML algorithms.
    
After preprocessing

Age	Salary	Gender	Purchased
25	30000	Male	Yes
30	40000	Female	No
35	35000	Male	Yes

The dataset is now cleaner and more consistent

The replacement values are illustrative. Actual preprocessing decisions should be based on the dataset and domain knowledge

What Happens If We Do Not Preprocess Data?

Without preprocessing	  Possible impact
Missing values	          Errors or loss of useful records
Duplicate records	      Biased statistical results
Invalid values	          Misleading analysis
Unencoded categories	  Model input errors
Unscaled features	      Some algorithms may give excessive importance to large-scale features
Data leakage	          Unrealistically high test performance
````

````python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "Age": [25, np.nan, 35],
    "Salary": [30000, 40000, -5000]
})

# Handle missing age
df["Age"] = df["Age"].fillna(
    df["Age"].median()
)

# Handle invalid salary
df["Salary"] = df["Salary"].mask(
    df["Salary"] < 0,
    np.nan
)

df["Salary"] = df["Salary"].fillna(
    df["Salary"].median()
)

print(df)
````

### Data Quality

Data Quality refers to how accurate, complete, consistent, valid, and reliable data is for its intended purpose

High-quality data helps us perform correct data analysis, generate meaningful business insights, and build reliable Machine Learning models

Low-quality data may contain missing values, duplicate records, incorrect values, inconsistent formats, or outdated information

````python
Example:

Suppose a company maintains customer information.
    
CustomerID	Age	Salary	City
101	        25	30000	Hyderabad
102      	NaN	40000	Hyderabad
103	        -5	50000	hyderabad
103	        -5	50000	hyderabad

Problems:

- Age is missing for customer 102
- Age is invalid for customer 103
- City has inconsistent capitalization
- Customer 103 is duplicated
This dataset has poor data quality

Six Main Dimensions of Data Quality

1. Accuracy

Data correctly represents real-world information.

Example: A customer's actual age is 25, but the dataset records 52.

2. Completeness

Required data is available without unwanted missing values.

Example: Customer age or email is missing.

3. Consistency

Data does not contradict itself across records or sources.

Example: The same customer's city is recorded as Hyderabad in one table and Mumbai in another.

4. Validity

Data follows predefined rules, formats, and constraints.

Example: Age must be a positive integer within a reasonable range.

5. Uniqueness

Records that should be unique are not unnecessarily duplicated.

Example: CustomerID 101 appears multiple times when each customer should have one record.

6. Timeliness

Data is sufficiently up to date for its intended use.

Example: A bank uses an outdated account balance for a current transaction.
````

````python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "CustomerID": [101, 102, 103, 103],
    "Age": [25, np.nan, -5, -5],
    "Salary": [30000, 40000, 50000, 50000],
    "City": [
        "Hyderabad",
        "Hyderabad",
        "hyderabad",
        "hyderabad"
    ]
})

df
````

````python
# Missing values count              #Check completeness
print(df.isnull().sum())

# Missing values percentage
missing_percentage = (
    df.isnull().mean() * 100
)

print(missing_percentage)
````

````python
# Count duplicate rows               #Check uniqueness
print(df.duplicated().sum())

# Check duplicate CustomerIDs
print(df["CustomerID"].duplicated().sum())
````

````python
# Check invalid ages                   #Check validity
invalid_age = df[
    (df["Age"] < 0) |
    (df["Age"] > 120)
]

print(invalid_age)
````

````python
# Unique city values                    #Check consistency
print(df["City"].unique())

# Standardize city names
df["City"] = df["City"].str.title()

print(df["City"].unique())
````

````python
quality_summary = pd.DataFrame({                         #Data Quality Summary
    "Missing_Count": df.isnull().sum(),
    "Missing_Percentage": df.isnull().mean() * 100,
    "Unique_Count": df.nunique()
})

print(quality_summary)
````

### Data Consistency

Data Consistency refers to the condition in which data is uniform, compatible, and does not contradict itself across records, columns, or different data sources

Consistent data follows the same formats, naming conventions, units, and business rules throughout a dataset

Data consistency is an important dimension of Data Quality

````python
Example:

Consider a customer dataset:

CustomerID	City	  Gender
101	      Hyderabad	  Male
102	      hyderabad	  male
103	      HYDERABAD	  MALE
104	      Hyderabad	  Female

Here, the same city and gender categories are represented using different capitalization.

Although the values refer to the same categories, they are stored inconsistently

After standardization:
    
CustomerID	City	Gender
  101	 Hyderabad	Male
  102	Hyderabad	Male
  103	Hyderabad	Male
  104	Hyderabad	Female

Now the data is consistent

Types of Data Consistency Problems:

Problem	                     Example

Inconsistent         capitalization	Male, male, MALE
Inconsistent         spelling	Hyderabad, Hydrabad
Inconsistent         date formats	2026-11-04, 04/11/2026
Inconsistent         units	Height in cm and inches
Inconsistent        category labels	Yes, Y, 1
Conflictingrecords	Same customer has different birth dates in two tables
````

````python
import pandas as pd

df = pd.DataFrame({
    "CustomerID": [101, 102, 103, 104],
    "City": [
        "Hyderabad",
        "hyderabad",
        "HYDERABAD",
        "Hydrabad"
    ],
    "Gender": [
        "Male",
        "male",
        "MALE",
        "Female"
    ]
})

df
````

````python
print(df["City"].unique())       #Identify inconsistent values

print(df["Gender"].unique())
````

````python
# Remove extra spaces                        #Standardize the data
df["City"] = df["City"].str.strip()

# Standardize capitalization
df["City"] = df["City"].str.title()

# Standardize gender
df["Gender"] = df["Gender"].str.title()

# Correct known spelling variation
df["City"] = df["City"].replace({
    "Hydrabad": "Hyderabad"
})

df
````

````python
print(df["City"].unique())          #Verify consistency

print(df["Gender"].unique())
````

#### How to Measure Data Consistency?

A simple way to measure category consistency is to compare the number of distinct raw values with the number of standardized values.

For example:
- Raw city categories: 4
- Standardized city categories: 1
  
This indicates that multiple raw representations refer to the same city.

A simple consistency percentage can be defined for a known standardization task as:

Consistency Percentage

The consistency percentage measures the percentage of records that follow the standard data rules.

$$
\text{Consistency Percentage}
=
\frac{\text{Records Following Standard Rules}}
{\text{Total Records}}
\times 100
$$

**Example:**

If 95 out of 100 records follow the required format:

$$
\frac{95}{100}\times100=95\%
$$

**Result:** Consistency Percentage = 95%

#### Why is Data Consistency Important?

Data consistency helps:

1. Prevent incorrect category counts.
2. Improve data analysis accuracy.
3. Avoid incorrect grouping and aggregation.
4. Reduce errors during data integration.
5. Improve the reliability of Machine Learning features.
For example, without standardization, Pandas may treat Male, male, and MALE as three separate categories. This can create incorrect frequency distributions and unnecessary categories during encoding.

### Data Completeness

Data Completeness refers to the extent to which all required data values are present in a dataset.

A dataset is considered complete when the necessary information is available without unwanted missing values.

Missing values may occur because of:

- Users not providing required information.
- Data collection errors.
- Database or system failures.
- Incorrect data integration.
- Sensor or measurement failures.

#### How to Calculate Data Completeness?


Completeness can be measured using the following formula:

$$
\text{Completeness Percentage}
=
\frac{\text{Number of Non-Missing Values}}
{\text{Total Expected Values}}
\times 100
$$

**Example:**

Suppose a dataset contains 100 records, and 90 records have the required information.

$$
\text{Completeness}
=
\frac{90}{100}\times100
$$

$$
=90\%
$$

**Result:** The dataset has 90% completeness.

````python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "CustomerID": [101, 102, 103, 104, 105],
    "Name": ["Ravi", "Kumar", "Raju", "Sai", "Arun"],
    "Age": [25, np.nan, 30, 28, 35],
    "Salary": [30000, 40000, np.nan, 35000, 45000],
    "City": [
        "Hyderabad",
        "Hyderabad",
        "Mumbai",
        np.nan,
        "Chennai"
    ]
})

df
````

````python
df.isnull().sum() #Identify missing values
````

````python
total_values = df.size            #Calculate completeness percentage

non_missing_values = df.notnull().sum().sum()

completeness = (
    non_missing_values / total_values
) * 100

print("Completeness Percentage:", completeness)
````

````python
column_completeness = (            #column-wise completeness
    df.notnull().mean() * 100
)

print(column_completeness)
````

````python
missing_percentage = (        #missing value percentage
    df.isnull().mean() * 100
)

print(missing_percentage)
````

### Data Accuracy

Data Accuracy refers to the degree to which data correctly represents real-world information

In simple words, data accuracy means checking whether the values stored in a dataset are actually correct

Accurate data should match the original source or verified real-world information

#### How to Calculate Data Accuracy?

A simple record-level accuracy formula is:

$$
\text{Accuracy Percentage}
=
\frac{\text{Number of Accurate Records}}
{\text{Total Records}}
\times 100
$$

**Example:**

Suppose a dataset contains 100 records, and 95 records have correct information.

$$
\text{Accuracy}
=
\frac{95}{100}\times100
$$

$$
=95\%
$$

**Result:** 95% Accuracy

This is a simple metric for checking records against verified reference data.

````python
Simple Example:
    
Consider a customer dataset:
    
CustomerID	ActualAge	RecordedAge
   101	      25	       25
   102	      30	       35
   103	      40	       40
   104	      28	       18

Here:

- Customer 101: Accurate
- Customer 102: Inaccurate
- Customer 103: Accurate
- Customer 104: Inaccurate

Only 2 out of 4 records contain the correct age

Common Data Accuracy Problems:
    
Problem                       	Example

Incorrect values	     Actual age is 25, but recorded age is 52
Typographical errors	Hyderbad instead of Hyderabad
Incorrect measurements	Weight recorded as 700 kg instead of 70 kg
Incorrect dates	        Birth date recorded incorrectly
Incorrect calculations	Total salary calculated incorrectly
Outdated information	Old customer address stored in the database
````

````python
import pandas as pd

df = pd.DataFrame({
    "CustomerID": [101, 102, 103, 104, 105],
    "Actual_Age": [25, 30, 40, 28, 35],
    "Recorded_Age": [25, 35, 40, 18, 35]
})

df
````

````python
df["Is_Accurate"] = (                            #Compare actual and recorded values
    df["Actual_Age"] == df["Recorded_Age"]
)

df
````

````python
accurate_records = df["Is_Accurate"].sum()            #Calculate accuracy percentage

total_records = len(df)

accuracy = (
    accurate_records / total_records
) * 100

print("Accuracy Percentage:", accuracy, "%")
````

````python
inaccurate_records = df[                  #Identify inaccurate records
    df["Is_Accurate"] == False
]

print(inaccurate_records)
````

### Data Validity

Data Validity refers to the extent to which data follows predefined rules, formats, constraints, and acceptable values

In simple words, data validity checks whether the data is in the correct format and whether its values satisfy the required conditions

Data can be valid without necessarily being accurate. For example, an age of 35 may be valid according to the rules, but inaccurate if the person's actual age is 25

#### How to Calculate Data Validity?

A simple validity percentage formula is:

$$
\text{Validity Percentage}
=
\frac{\text{Number of Valid Records}}
{\text{Total Records}}
\times 100
$$

**Example:**

Suppose a dataset contains 100 records, and 90 records follow the predefined rules.

$$
\text{Validity}
=
\frac{90}{100}\times100
$$

$$
=90\%
$$

**Result:** 90% Validity.

````python
Example:

Consider a customer dataset:

CustomerID	Age	Gender	Email
101      	25	Male	ravi@gmail.com
102     	-5	Female	kumar@gmail.com
103	       150	Other	invalid-email
104	        30	Unknown	sai@gmail.com

Validity problems:
    
- Age -5 is invalid because age cannot be negative
- Age 150 may violate the defined age range
- invalid-email does not follow the email format
- Unknown may not be an accepted Gender category

Types of Data Validity

Type	                Explanation	                                          Example
Range Validity	       Values must fall within an allowed range	        Age between 0 and 120
Format Validity	       Values must follow a required format	            Email address
Type Validity	       Values must have the correct data type	        Age must be numeric
Category Validity	   Values must belong to allowed categories	        Gender: Male, Female, Other
Date Validity	        Dates must satisfy rules	                    Joining date cannot be in the future
Constraint Validity	   Values must satisfy business rules	            Salary must be greater than 0
````

````python
import pandas as pd

df = pd.DataFrame({
    "CustomerID": [101, 102, 103, 104, 105],
    "Age": [25, -5, 150, 30, 45],
    "Gender": [
        "Male", "Female", "Other", "Unknown", "Male"
    ]
})

df
````

````python
valid_age = df["Age"].between(0, 120)      #Define validity rules

valid_gender = df["Gender"].isin([
    "Male", "Female", "Other"
])

df["Is_Valid"] = valid_age & valid_gender

df
````

````python
valid_records = df["Is_Valid"].sum()     #Calculate validity percentage

total_records = len(df)

validity_percentage = (
    valid_records / total_records
) * 100

print("Validity Percentage:", validity_percentage, "%")
````

````python
invalid_records = df[              #Identify invalid records
    df["Is_Valid"] == False
]

print(invalid_records)
````

### Data Integrity

Data Integrity refers to the accuracy, consistency, completeness, and reliability of data throughout its lifecycle

It ensures that data remains correct, trustworthy, and properly maintained when it is collected, stored, updated, transferred, or processed

In simple words, data integrity means maintaining data in a correct and reliable condition

````python
Types of Data Integrity

Type	                          Explanation	                                Example
Entity Integrity	    Each record must have a unique identifier	    CustomerID cannot be duplicated
Referential Integrity	Relationships between tables must remain valid	An order must reference an existing customer
Domain Integrity	    Values must follow allowed rules	            Age must be within a defined range
User-Defined Integrity	Data must follow specific business rules	   Bank account balance cannot go below an allowed limit

A. Entity Integrity

Every record should have a unique primary key.

Example:
CustomerID	Name
101	       Ravi
102	       Kumar
103	       Raju

CustomerID must not contain duplicate or missing values.
    
B. Referential Integrity
    
Relationships between tables must remain valid.

Example:

Customers Table
    
CustomerID	Name
101	        Ravi
102	        Kumar

Orders Table
    
OrderID	CustomerID
501 	101
502	    102
503	    105

Order 503 references CustomerID 105, which does not exist in the Customers table.

This creates a referential integrity problem.
    
C. Domain Integrity
    
Values must follow predefined rules.

Example:
    
- Age must be between 0 and 120.
- Salary must be non-negative.
- Gender must belong to allowed categories.
````

````python
import pandas as pd

df = pd.DataFrame({
    "CustomerID": [101, 102, 103, 103],
    "Age": [25, 30, -5, 40],
    "Salary": [30000, 40000, 50000, -1000]
})

df
````

````python
# Check duplicate CustomerIDs                #Check entity integrity
print(df["CustomerID"].duplicated().sum())

# Check missing CustomerIDs
print(df["CustomerID"].isnull().sum())
````

````python
# Check invalid ages            #Check domain integrity
invalid_age = df[
    ~df["Age"].between(0, 120)
]

print(invalid_age)
````

````python
invalid_salary = df[       #Check salary constraints
    df["Salary"] < 0
]

print(invalid_salary)
````

### Data Leakage

Data Leakage is a problem in Machine Learning where information that would not legitimately be available during prediction is used to train a model.

This causes the model to learn from information it should not have access to.

As a result, the model may show very high accuracy during training or testing but perform poorly on real-world, unseen data.

In simple words:
Data Leakage happens when a Machine Learning model gets access to answers or future information that it should not know.

````python
Real-World Example:
    
Suppose we want to predict whether a customer will leave a bank
    
Customer	Age	  Salary	AccountClosed	Churn
   A	    25	  30000	       Yes	         Yes
   B	    35	  50000	       No	          No
   C	    40	  60000	       Yes	         Yes

Here, Account Closed reveals the answer to the target variable Churn.

If we use this feature to predict churn, the model may achieve extremely high accuracy. However, account closure information may only become available after the customer has already left.

This is an example of Target Leakage.

The model should instead use information available before the prediction, such as account activity, transaction frequency, or customer complaints. 
    The key question is whether a feature would actually be available at prediction time

Types of Data Leakage:
    
Type	             Explanation

Target Leakage	     A feature directly or indirectly reveals the target
Train-Test Leakage	Test data influences model training or preprocessing
Temporal Leakage	Future information is used to predict the past
Group Leakage	    Related records appear in both training and test sets

Example of Temporal Leakage

Suppose we predict tomorrow's stock price.

Using tomorrow's actual closing price as an input feature would reveal future information.

The model cannot legitimately use that value when making today's prediction.

Data Leakage During Preprocessing

One common mistake is applying preprocessing to the entire dataset before splitting it.

For example, calculating the mean of the complete dataset to fill missing values allows the test dataset to influence the training process.

The correct approach is:
    
1. Split the dataset into training and testing sets.
2. Learn preprocessing parameters using training data only.
3. Apply the same transformations to the test data.
4. Evaluate the model on the test data.
````

````python
import pandas as pd

df = pd.DataFrame({
    "Study_Hours": [2, 3, 4, 5, 6, 7, 8, 1, 4, 6],
    "Attendance": [50, 60, 65, 70, 80, 85, 90, 40, 75, 88],
    "Final_Marks": [25, 30, 35, 45, 55, 65, 80, 20, 50, 70],
    "Result": [
        "Fail", "Fail", "Fail", "Pass", "Pass",
        "Pass", "Pass", "Fail", "Pass", "Pass"
    ]
})

df
````

````python
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score

X = df[["Final_Marks"]]

y = df["Result"]

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.3,
    random_state=42,
    stratify=y
)

model = DecisionTreeClassifier(random_state=42)

model.fit(X_train, y_train)

y_pred = model.predict(X_test)

accuracy = accuracy_score(y_test, y_pred)

print("Actual Results:", y_test.tolist())
print("Predicted Results:", y_pred.tolist())
print("Accuracy:", accuracy * 100, "%")
````

How to Avoid Data Leakage?

- Split data before preprocessing
- Never use test data to fit a scaler or imputer
- Remove features that reveal future outcomes
- Use time-based splitting for time-series data
- Keep related groups together when splitting
- Use Scikit-learn pipelines to combine preprocessing and model training

$$
\text{Split Data}
\rightarrow
\text{Preprocess Training Data}
\rightarrow
\text{Train Model}
\rightarrow
\text{Evaluate on Test Data}
$$

### Preprocessing Pipeline

A Preprocessing Pipeline is a sequence of steps used to prepare raw data before providing it to a Machine Learning model

Raw datasets may contain:

- Missing values
- Duplicate records
- Incorrect data types
- Categorical values
- Different numerical scales
- Outliers
- Unnecessary columns
  
A preprocessing pipeline applies the required transformations in a systematic and consistent order

Simple definition: A preprocessing pipeline is an organized sequence of data-cleaning and transformation steps that converts raw data into a format suitable for machine learning

Why do we need a preprocessing pipeline?

1. Handle missing values
2. Convert categorical data into numerical data
3. Scale numerical features
4. Maintain consistent transformations
5. Reduce data-quality problems
6. Prevent data leakage
7. Make model training and prediction easier

##### Preprocessing Pipeline

```mermaid
flowchart TD
    A["Raw Dataset"] --> B["Inspect Dataset"]
    B --> C["Separate Features and Target"]
    C --> D["Split Training and Testing Data"]

    D --> E["Preprocessing Pipeline"]

    E --> F["Handle Missing Values"]
    F --> G["Encode Categorical Features"]
    G --> H["Scale Numerical Features"]

    H --> I["Prepared Training Data"]
    I --> J["Train ML Model"]

    J --> K["Evaluate Model"]
    K --> L["Final Predictions"]

````python

import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report

# 1. Create dataset
data = {
    "Age": [25, 30, 35, np.nan, 40, 28, 45, 32, 38, 50],
    "Experience": [1, 5, 8, 10, np.nan, 3, 15, 6, 12, 20],
    "Education": [
        "Degree", "Masters", "Degree", "PhD", "Masters",
        "Degree", "PhD", "Masters", "Degree", "PhD"
    ],
    "Department": [
        "IT", "HR", "IT", "Finance", "HR",
        "Finance", "IT", "HR", "Finance", "IT"
    ],
    "Salary": [
        25000, 45000, 55000, 65000, 70000,
        30000, 80000, 48000, 60000, 90000
    ]
}

df = pd.DataFrame(data)

# 2. Create target variable
df["High_Salary"] = (df["Salary"] > 50000).astype(int)

# 3. Separate features and target
X = df.drop(["Salary", "High_Salary"], axis=1)
y = df["High_Salary"]

# 4. Identify numerical and categorical features
numerical_features = ["Age", "Experience"]
categorical_features = ["Education", "Department"]

# 5. Create numerical preprocessing pipeline
numerical_pipeline = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

# 6. Create categorical preprocessing pipeline
categorical_pipeline = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore"))
])

# 7. Combine preprocessing pipelines
preprocessor = ColumnTransformer(transformers=[
    ("num", numerical_pipeline, numerical_features),
    ("cat", categorical_pipeline, categorical_features)
])

# 8. Create complete machine learning pipeline
model_pipeline = Pipeline(steps=[
    ("preprocessor", preprocessor),
    ("classifier", RandomForestClassifier(
        n_estimators=100,
        random_state=42
    ))
])

# 9. Split dataset
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.3,
    random_state=42,
    stratify=y
)

# 10. Train model
model_pipeline.fit(X_train, y_train)

# 11. Make predictions
y_pred = model_pipeline.predict(X_test)

# 12. Evaluate model
print("Actual Values:", y_test.tolist())
print("Predicted Values:", y_pred.tolist())

print("\nModel Accuracy:", accuracy_score(y_test, y_pred))

print("\nClassification Report:")
print(classification_report(y_test, y_pred, zero_division=0))

# 13. Predict for a new employee
new_employee = pd.DataFrame({
    "Age": [29],
    "Experience": [4],
    "Education": ["Masters"],
    "Department": ["IT"]
})

prediction = model_pipeline.predict(new_employee)

if prediction[0] == 1:
    print("\nPrediction: High Salary (Above ₹50,000)")
else:
    print("\nPrediction: Low Salary (₹50,000 or below)")
````
