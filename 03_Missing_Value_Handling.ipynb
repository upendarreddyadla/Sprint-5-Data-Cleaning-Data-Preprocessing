# Notebook 3 – Missing Value Handling

### What are Missing Values?

Missing Values are data points or entries that are not available in a dataset for one or more columns

In simple words, when a dataset contains empty cells, NaN, None, or NULL instead of actual values, those are called Missing Values

Missing values occur when information is not collected, recorded, or stored properly

````python
Example of Missing Values

Consider a student dataset:
    
Student_ID	Name	Age	 Marks
    101	    Ravi	22	 85
    102	    Kumar	NaN	 90
    103	    Ramesh	21	 NaN
    104	    Suresh	NaN	 75
    105	    Mahesh	23	 88

Here:

- Age is missing for Kumar and Suresh
- Marks are missing for Ramesh
- NaN represents a missing value

2. Common Representations of Missing Values

Representation	    Meaning

NaN	               Not a Number
None	           No value in Python
NULL	           Missing value in SQL databases
NaT	               Not a Time (missing datetime value)
Empty cell	       No data entered
""	               Empty string
````

##### Why Do Missing Values Occur?

Common reasons include:

1. Data Entry Errors: Users forget to enter information.
2. Data Collection Problems: Information is not collected properly.
3. Sensor Failures: Devices fail to record measurements.
4. Data Integration: Some information is unavailable when combining datasets.
5. User Privacy: Users choose not to provide personal information.

##### Why Are Missing Values Important in Data Analysis?
   
 Missing values can cause several problems:

- Incorrect statistical calculations
- Reduced dataset quality
- Biased analysis results
- Problems during machine learning model training
- Reduced model prediction accuracy

````python
import pandas as pd
import numpy as np

# Create a sample dataset
data = {
    'Name': ['Ravi', 'Kumar', 'Ramesh', 'Suresh', 'Mahesh'],
    'Age': [22, np.nan, 21, np.nan, 23],
    'Marks': [85, 90, np.nan, 75, 88]
}

df = pd.DataFrame(data)

# Display dataset
print("Dataset:")
print(df)

# Identify missing values
print("\nMissing Values:")
print(df.isnull())

# Count missing values in each column
print("\nMissing Values Count:")
print(df.isnull().sum())

# Total missing values
print("\nTotal Missing Values:")
print(df.isnull().sum().sum())
````

### Why Missing Values Occur

Missing values occur when certain information is not collected, recorded, stored, or transferred properly in a dataset

In simple words, missing values occur when expected data is unavailable for a particular record or feature

There are several reasons why missing values occur

````python
1.Data Entry Errors

Data entry errors happen when users forget to enter information or accidentally leave fields empty.

Example:

Student_ID	Name	Age

  101   	Ravi	22
  102	   Kumar	NaN
  103	   Ramesh	21


2.Data Collection Problems

Sometimes, data collection systems fail to collect all required information.

Example:

An online registration form asks customers to enter their name, age, email, and phone number. Some customers skip the age field.

As a result, Age contains missing values.
    
3. Sensor or Device Failures
    
Sensors and electronic devices sometimes fail to record measurements because of technical problems.

Example:

A temperature sensor records temperatures every hour.
    
Time	Temperature
9 AM	32°C
10 AM	33°C
11 AM	NaN
12 PM	34°C

The temperature at 11 AM is missing because the sensor failed.
    
4. User Privacy

Some users do not want to share their personal information.

Example:

Customer	Name	Age	Salary
    1	    Ravi	25	30000
    2	    Kumar	NaN	40000
         3	Ramesh	30	NaN

Kumar does not provide his age, and Ramesh does not provide his salary

5. Data Integration Problems

Data integration means combining information from multiple datasets or sources.

Sometimes, one dataset contains information that another dataset does not have.

Example:

Dataset 1:
    
Customer_ID 	Name
101	            Ravi
102         	Kumar

Dataset 2:

Customer_ID 	Salary
101         	30000
103	            40000

After combining both datasets:
    
Customer_ID	Name	Salary
    101	    Ravi	30000
    102	    Kumar	NaN
    103   	NaN	    40000

Missing values appear because customer information is unavailable in one of the datasets

6. Data Transmission Failures

Sometimes, data gets lost while transferring information between systems, applications, or databases.

Example:

An online payment system transfers transaction information to a database. Due to a network failure, some transaction details are not received.

As a result, certain columns contain missing values

7. Conditional Data Collection

Some fields are required only under specific conditions. If those conditions are not applicable, the values may remain empty.

Example:

Customer	Purchased	Delivery_Date
  Ravi	      Yes	    2026-11-01
  Kumar    	   No	       NaN
 Ramesh	       Yes	     2026-11-03

Kumar did not purchase anything, so a delivery date is not applicable.
````

### MCAR

MCAR stands for Missing Completely At Random

MCAR means missing values occur completely randomly in a dataset. The probability of a value being missing is not related to any other feature or to the missing value itself

In simple words, there is no specific reason or pattern behind why the data is missing

````python
Example of MCAR

Suppose a hospital maintains patient information.
    
Patient_ID	Name	Age 	Weight
  101	   Ravi	     25	    65
  102	  Kumar	     30	    NaN
  103	  Ramesh	 28	    70
  104	  Suresh	 NaN	    68
  105	  Mahesh	 32	    72

Here:

- Kumar's weight is missing because of a random recording interruption.
- Suresh's age is missing because of an accidental system error.
- These missing values are unrelated to the patients' ages, weights, or other characteristics.

Therefore, this is an example of MCAR.
    
2. Real-World Examples
    
Domain    	         MCAR Example
    
Healthcare	    A random technical error causes some patient measurements to be lost
Banking	        A random database interruption causes some transaction fields to be missing
Education	    A random software error removes some students' marks
E-commerce	    A temporary system failure loses some customer records
````

##### Mathematical Representation of MCAR

Let M represent whether a value is missing.

$$
P(M = 1 \mid X, Y) = P(M = 1)
$$

**Where:**

- M = 1: Value is missing.
- X = Other observed features.
- Y = Value that could be missing.
- P = Probability.



The probability of a value being missing does not depend on observed or unobserved data.

In simple words, missing values occur completely randomly without depending on any other feature or the missing value itself.

````python
import pandas as pd
import numpy as np

# Create dataset
data = {
    'Student': ['Ravi', 'Kumar', 'Ramesh', 'Suresh', 'Mahesh'],
    'Age': [22, 23, 21, 24, 25],
    'Marks': [85, np.nan, 90, 75, 88],
    'Attendance': [90, 85, np.nan, 95, 80]
}

df = pd.DataFrame(data)

print("Original Dataset:")
print(df)

# Identify missing values
print("\nMissing Values:")
print(df.isnull())

# Count missing values
print("\nMissing Values Count:")
print(df.isnull().sum())

# Calculate missing percentage
print("\nMissing Percentage:")
print(df.isnull().mean() * 100)
````

How to Handle MCAR?

Since missing values are unrelated to other variables, common approaches include:
1. Deletion: Remove records containing missing values when the amount of missing data is small and deletion is appropriate.
2. Mean Imputation: Replace missing numerical values with the mean.
3. Median Imputation: Replace missing numerical values with the median.
4. Mode Imputation: Replace missing categorical values with the mode.

### MAR

MAR stands for Missing At Random

MAR means missing values occur because of other observed variables (features) in the dataset, but they do not depend on the missing value itself after considering the observed variables

In simple words, the reason a value is missing is related to another available column in the dataset

````python
Example of MAR:
    
Suppose a company maintains employee information

Employee	Age	Salary	Experience

    Ravi	25	30000	2
   Kumar	30	NaN	    5
   Ramesh	35	NaN	    8
   Suresh	24	25000	1
   Mahesh	40	NaN	    12

Here:

- Salary values are missing more frequently among employees with higher experience.
- Experience is available in the dataset.
- Missing salary information is related to an observed feature (Experience).
Therefore, this could be an example of MAR if the missingness is explained by observed experience rather than the actual salary itself.
    
2. Real-World Examples
    
Domain	     MAR Example
    
Healthcare	 Older patients are more likely to have certain test results missing.
Education	Students from a particular school are more likely to have missing attendance records.
Banking  	Customers from a particular branch have more missing income information due to its data collection process.
E-commerce	Mobile users are more likely than desktop users to leave certain optional fields empty.
````

##### Mathematical Representation of MAR

Let \(M\) represent whether a value is missing.

For MAR:

$$
P(M = 1 \mid X, Y) = P(M = 1 \mid X)
$$

**Where:**

- \(M = 1\): Value is missing.
- \(X\): Other observed features.
- \(Y\): Value that could be missing.
- \(P\): Probability.

**Meaning:**

The probability of a value being missing depends on observed features (X), but does not depend on the missing value itself (Y), once X is considered.

````python
import pandas as pd
import numpy as np

# Create employee dataset
data = {
    'Employee': ['Ravi', 'Kumar', 'Ramesh', 'Suresh', 'Mahesh'],
    'Experience': [1, 5, 8, 2, 12],
    'Salary': [25000, np.nan, np.nan, 28000, np.nan]
}

df = pd.DataFrame(data)

print("Employee Dataset:")
print(df)

# Identify missing values
print("\nMissing Values:")
print(df.isnull())

# Group employees based on experience
df['Experience_Group'] = pd.cut(
    df['Experience'],
    bins=[0, 3, 15],
    labels=['Low Experience', 'High Experience']
)

# Analyze missing salary values
print("\nMissing Salary by Experience Group:")
print(df.groupby('Experience_Group', observed=False)['Salary']
      .apply(lambda x: x.isnull().mean() * 100))
````

How to Handle MAR?

Common approaches include:

1. Mean Imputation
2. Median Imputation
3. Mode Imputation
4. KNN Imputation
5. Multiple Imputation (MICE)
6. Regression Imputation
   
The appropriate technique depends on the dataset and missingness pattern.

### MNAR

MNAR stands for Missing Not At Random

MNAR means missing values occur because of the value itself that is missing. The probability of a value being missing depends on its own unobserved value

In simple words, the reason data is missing is directly related to the actual value that is not available

````python
Example of MNAR:
    
Suppose a company collects employee salary information.
    
    Employee	Age	 Salary
     Ravi	    25	 30000
     Kumar	    30	  NaN
     Ramesh	    35	  NaN
     Suresh	    28	25000
     Mahesh	    40	60000

Here:

- Employees with very high salaries may intentionally avoid sharing their salary information.
- Employees with lower salaries may be more willing to provide their information.
- The probability of Salary being missing depends on the actual salary.
    
Therefore, this is an example of MNAR.
    
2. Real-World Examples
    
Domain	      MNAR Example
    
Healthcare	 Patients with severe symptoms may be more likely to avoid reporting their actual pain levels
Banking	     Customers with very low income may avoid reporting their income
Education	 Students with very low marks may intentionally avoid reporting their marks
Surveys	     People with very high or very low incomes may avoid answering income-related questions
````

##### Mathematical Representation of MNAR

Let \(M\) represent whether a value is missing.

For MNAR:

$$
P(M = 1 \mid X, Y) \neq P(M = 1 \mid X)
$$

**Where:**

- \(M = 1\): Value is missing.
- \(X\): Other observed features.
- \(Y\): Value that could be missing.
- \(P\): Probability.

**Meaning:**

The probability of a value being missing depends on the actual value that is missing, even after considering other observed features.

````python
import pandas as pd
import numpy as np

# Create employee dataset
data = {
    'Employee': ['Ravi', 'Kumar', 'Ramesh', 'Suresh', 'Mahesh'],
    'Age': [25, 30, 35, 28, 40],
    'Salary': [30000, np.nan, np.nan, 25000, 60000]
}

df = pd.DataFrame(data)

# Display dataset
print("Employee Dataset:")
print(df)

# Identify missing values
print("\nMissing Values:")
print(df.isnull())

# Count missing values
print("\nMissing Values Count:")
print(df.isnull().sum())

# Calculate missing percentage
print("\nMissing Percentage:")
print(df.isnull().mean() * 100)
````

How to Handle MNAR?

Common approaches include:

1. Domain Knowledge: Understand why the values are missing
2. Missing Indicator: Create an additional column indicating whether a value is missing
3. Advanced Imputation: Use suitable statistical or model-based imputation techniques
4. Sensitivity Analysis: Compare results under different assumptions about missing values
5. Collect Additional Data: Try to obtain the missing information directly

````python
Difference Between MCAR, MAR, and MNAR

Type	Meaning	                                                Example

MCAR	Missing completely randomly	                       Random system failure
MAR	    Missingness depends on observed features	       Missing salary related to observed experience
MNAR	Missingness depends on the missing value itself	   High-income customers avoid reporting income
````

### Missing Value Percentage

Missing Value Percentage is the percentage of missing values present in a particular column compared to the total number of records in a dataset.

In simple words, it tells us how much data is missing in each column, expressed as a percentage.

##### Mathematical Formula

$$
\text{Missing Value Percentage} =
\frac{\text{Number of Missing Values}}
{\text{Total Number of Records}} \times 100
$$

**Where:**

- Number of Missing Values = Total empty or unavailable values in a column.
- Total Number of Records = Total rows in the dataset.
- 100 = Converts the result into a percentage.

````python
import pandas as pd
import numpy as np

# Create sample dataset
data = {
    'Student': ['Ravi', 'Kumar', 'Ramesh', 'Suresh', 'Mahesh'],
    'Age': [22, np.nan, 21, np.nan, 23],
    'Marks': [85, 90, np.nan, 75, 88],
    'Attendance': [90, np.nan, 85, 95, np.nan]
}

df = pd.DataFrame(data)

# Display dataset
print("Student Dataset:")
print(df)

# Calculate missing value count
missing_count = df.isnull().sum()

# Calculate missing value percentage
missing_percentage = (df.isnull().sum() / len(df)) * 100

# Combine results
missing_summary = pd.DataFrame({
    'Missing Count': missing_count,
    'Missing Percentage': missing_percentage
})

print("\nMissing Value Summary:")
print(missing_summary)
````

````python
Real-World Business Example:
    
Suppose an e-commerce company maintains 1,000 customer records.
    
Feature	        Missing Count	   Missing Percentage
 Age	           100	            10%
 Salary	           250	            25%
 Location	       50	             5%
 PurchaseAmount	   20	             2%
````

Business Insights:

- Age: 10% of customer ages are missing.
- Salary: 25% of salary information is missing.
- Location: Only 5% of location information is missing.
- Purchase Amount: Only 2% of purchase amounts are missing.
  
The company should investigate the missing salary values before performing customer segmentation

Important Points

- 0% → No missing values.
- 10% → 10% of records have missing values.
- 25% → One-fourth of the records have missing values.
- 50% → Half of the records have missing values.
- 100% → All records have missing values in that column.
  
Important: A higher missing-value percentage does not automatically mean that the column should be deleted. Consider business importance, missingness patterns, and the reason data is missing before deciding.

### Dropping Rows

Dropping Rows is a missing-value handling technique in which entire rows containing missing values are removed from a dataset

In simple words, if a row contains one or more missing values, we can delete that entire row from the dataset

In Pandas, we use the dropna() function to remove rows containing missing values

````python
Example of Dropping Rows:
    
Consider the following student dataset:

Before Dropping Rows:

 Student	Age	Marks
  Ravi	    22	85
  Kumar	   NaN	90
  Ramesh	21	NaN
  Suresh	24	75
  Mahesh	23	88

Here:

- Kumar's Age is missing.
- Ramesh's Marks are missing.
- These two rows contain missing values.
    
After Dropping Rows:

Student	Age	Marks
Ravi	22	85
Suresh	24	75
Mahesh	23	88

The rows containing missing values are removed
````

````python
import pandas as pd
import numpy as np

# Create sample dataset
data = {
    'Student': ['Ravi', 'Kumar', 'Ramesh', 'Suresh', 'Mahesh'],
    'Age': [22, np.nan, 21, 24, 23],
    'Marks': [85, 90, np.nan, 75, 88]
}

df = pd.DataFrame(data)

# Display original dataset
print("Original Dataset:")
print(df)

# Drop rows containing missing values
df_cleaned = df.dropna()

# Display cleaned dataset
print("\nDataset After Dropping Rows:")
print(df_cleaned)

# Display remaining rows
print("\nRemaining Rows:", len(df_cleaned))
````

### Mean Imputation

Mean Imputation is a missing-value handling technique in data preprocessing. It replaces missing values (NaN) in a numerical column with the mean (average) of the available values in that column

It is commonly used when numerical data contains missing values and the mean is an appropriate replacement

1. Why Do We Use Mean Imputation?
1. To handle missing numerical values.
2. To avoid losing important records.
3. To maintain the dataset size.
4. To prepare clean data for Machine Learning models

##### Mathematical Formula

The mean is calculated by dividing the sum of available values by the number of available values.

$$
\text{Mean} = \frac{\text{Sum of Available Values}}{\text{Number of Available Values}}
$$

##### Example

Suppose an employee dataset contains the following salaries:

`20000, 25000, NaN, 30000, 35000`

Calculate the mean by ignoring the missing value.

$$
\text{Mean} = \frac{20000+25000+30000+35000}{4}
$$

$$
= \frac{110000}{4}
$$

$$
= 27500
$$

Replace the missing value with 27500.

##### Before Mean Imputation

The original salary values contain one missing value.

$$
[20000,\ 25000,\ NaN,\ 30000,\ 35000]
$$

##### Calculated Mean

$$
\boxed{\text{Mean} = 27500}
$$

##### After Mean Imputation

Replace the missing value (NaN) with the calculated mean (27500).

$$
[20000,\ 25000,\ 27500,\ 30000,\ 35000]
$$

````python
import pandas as pd
import numpy as np

# Create sample dataset
data = {
    "Employee": ["Ravi", "Kumar", "Raju", "Arun", "Vijay"],
    "Salary": [20000, 25000, np.nan, 30000, 35000]
}

df = pd.DataFrame(data)

print("Original Dataset:")
print(df)

# Calculate mean salary
mean_salary = df["Salary"].mean()

print("\nMean Salary:", mean_salary)

# Replace missing values with mean
df["Salary"] = df["Salary"].fillna(mean_salary)

print("\nDataset After Mean Imputation:")
print(df)

# Check missing values
print("\nMissing Values:")
print(df.isnull().sum())
````

##### Real-World Business Example:

Suppose an e-commerce company has 1,000 customer records, but some customers have not provided their annual income

- Total customers = 1,000
- Available income records = 900
- Missing income records = 100
- Average annual income = ₹4,00,000
  
The company can replace the 100 missing income values with ₹4,00,000 if mean imputation is appropriate

Missing Values Replaced = 100
Final Records = 1000

The company retains all 1,000 customer records

6. Advantages of Mean Imputation
   
- Simple and easy to implement
- Maintains the original number of records
- Useful for numerical columns
- Helps prepare data for Machine Learning
  
7. Disadvantages of Mean Imputation
   
- Sensitive to outliers.
- Reduces the natural variation (variance) in data.
- Can distort the original data distribution.
- May introduce bias when values are not missing completely at random.

### Median Imputation

Median Imputation is a missing-value handling technique in data preprocessing. It replaces missing values (NaN) in a numerical column with the median (middle value) of the available values

It is especially useful when numerical data contains outliers or has a skewed distribution

1. Why Do We Use Median Imputation?
1. To handle missing numerical values
2. To avoid losing important records
3. To maintain the dataset size
4. To reduce the influence of outliers compared with mean imputation

##### Mathematical Formula

The median is the middle value when data is arranged in ascending order.

**For odd number of observations:**

$$
\text{Median} = \text{Middle Value}
$$

**For even number of observations:**

$$
\text{Median} =
\frac{\text{Two Middle Values Sum}}{2}
$$

##### Example

Suppose an employee dataset contains the following salaries:

`20000, 25000, NaN, 30000, 100000`

First, remove the missing value and arrange the available values in ascending order.

`20000, 25000, 30000, 100000`

Here, the number of available values is 4 (even).

Calculate the median:

$$
\text{Median} = \frac{25000+30000}{2}
$$

$$
= \frac{55000}{2}
$$

$$
= 27500
$$

Replace the missing value with 27500.

##### Before Median Imputation

The original salary values contain one missing value.

$$
[20000,\ 25000,\ NaN,\ 30000,\ 100000]
$$

##### Calculated Median

$$
\boxed{\text{Median} = 27500}
$$

##### After Median Imputation

Replace the missing value (NaN) with the calculated median (27500).

$$
[20000,\ 25000,\ 27500,\ 30000,\ 100000]
$$

````python
import pandas as pd
import numpy as np

# Create sample dataset
data = {
    "Employee": ["Ravi", "Kumar", "Raju", "Arun", "Vijay"],
    "Salary": [20000, 25000, np.nan, 30000, 100000]
}

df = pd.DataFrame(data)

print("Original Dataset:")
print(df)

# Calculate median salary
median_salary = df["Salary"].median()

print("\nMedian Salary:", median_salary)

# Replace missing values with median
df["Salary"] = df["Salary"].fillna(median_salary)

print("\nDataset After Median Imputation:")
print(df)

# Check missing values
print("\nMissing Values:")
print(df.isnull().sum())
````

##### Real-World Business Example:

Suppose an e-commerce company maintains 1,000 customer records containing annual income

- Total customers = 1,000
- Available income records = 900
- Missing income records = 100
- Median annual income = ₹4,50,000
  
The company replaces the 100 missing income values with ₹4,50,000


Final Records = 900 + 100 = 1000


The company retains all 1,000 customer records

 Advantages of Median Imputation
 
- Less sensitive to outliers.
- Useful for skewed numerical data.
- Simple to implement.
- Maintains the number of records

Disadvantages of Median Imputation
   
- Reduces data variability.
- Can distort the original distribution.
- May introduce bias depending on the missing-data mechanism.

### Mode Imputation

Mode Imputation is a missing-value handling technique in data preprocessing. It replaces missing values (NaN) with the mode, which is the most frequently occurring value in a column

It is mainly used for categorical data, but it can also be used for numerical data when appropriate

1. Why Do We Use Mode Imputation?
1. To handle missing categorical values.
2. To avoid losing important records.
3. To maintain the dataset size.
4. To fill missing values using the most frequently occurring value.

##### Mathematical Formula

Mode is the value that occurs most frequently in a dataset.

$$
\text{Mode} = \text{Most Frequently Occurring Value}
$$

##### Frequency Formula:

$$
\text{Mode} = X_i \text{ where Frequency}(X_i)
\text{ is Maximum}
$$

##### Example

Suppose an employee dataset contains the following department values:

`HR, IT, Sales, IT, NaN, IT, HR, Sales`

First, count the frequency of each available department.

| Department | Frequency |
|:-----------|----------:|
| HR         | 2         |
| IT         | 3         |
| Sales      | 2         |

Here, IT occurs 3 times, which is the highest frequency.

$$
\text{Mode} = IT
$$

Replace the missing value with IT.

##### Before Mode Imputation

The original department values contain one missing value.

`HR, IT, Sales, IT, NaN, IT, HR, Sales`

##### Calculated Mode

$$
\boxed{\text{Mode} = IT}
$$

##### After Mode Imputation

Replace the missing value (NaN) with the calculated mode (IT).

`HR, IT, Sales, IT, IT, IT, HR, Sales`

````python
import pandas as pd
import numpy as np

# Create sample dataset
data = {
    "Employee": ["Ravi", "Kumar", "Raju", "Arun",
                 "Vijay", "Suresh", "Ajay", "Rahul"],
    "Department": ["HR", "IT", "Sales", "IT",
                   np.nan, "IT", "HR", "Sales"]
}

df = pd.DataFrame(data)

print("Original Dataset:")
print(df)

# Calculate mode
mode_department = df["Department"].mode()[0]

print("\nMode Department:", mode_department)

# Replace missing values with mode
df["Department"] = df["Department"].fillna(mode_department)

print("\nDataset After Mode Imputation:")
print(df)

# Check missing values
print("\nMissing Values:")
print(df.isnull().sum())
````

1. Why Do We Use Constant Value Imputation?

##### Mathematical Formula

### Forward Fill

Forward Fill is a missing-value handling technique in data preprocessing. It replaces missing values (NaN) with the previous available value in the same column

It is commonly used in time-series datasets, sales records, stock prices, weather data, and sensor readings

1. Why Do We Use Forward Fill?
1. To handle missing values.
2. To maintain the original dataset size.
3. To carry forward the previous available value.
4. To handle missing observations in time-series data.

##### Mathematical Formula

Forward Fill replaces a missing value with the previous available value.

$$
X_t = X_{t-1}
$$

Where:

- $X_t$ = Current missing value.
- $X_{t-1}$ = Previous available value.

##### Example

Suppose a company records daily sales:

`1000, 1500, NaN, NaN, 2000, NaN`

Apply Forward Fill.

**Before Forward Fill:**

`1000, 1500, NaN, NaN, 2000, NaN`

Replace each missing value with the most recent available value.

**After Forward Fill:**

`1000, 1500, 1500, 1500, 2000, 2000`

##### Forward Fill Process

Previous Value:

$$
X_{t-1} = 1500
$$

Missing Value 1:

$$
X_t = 1500
$$

Missing Value 2:

$$
X_t = 1500
$$

**Final Result:**

$$
[1000,\ 1500,\ 1500,\ 1500,\ 2000,\ 2000]
$$

````python
import pandas as pd
import numpy as np

# Create sample dataset
data = {
    "Date": ["01-01-2026", "02-01-2026", "03-01-2026",
             "04-01-2026", "05-01-2026", "06-01-2026"],
    "Sales": [1000, 1500, np.nan, np.nan, 2000, np.nan]
}

df = pd.DataFrame(data)

print("Original Dataset:")
print(df)

# Apply Forward Fill
df["Sales"] = df["Sales"].ffill()

print("\nDataset After Forward Fill:")
print(df)

# Check missing values
print("\nMissing Values:")
print(df.isnull().sum())
````

- Simple to implement.
- Maintains dataset size.
- Useful for time-series data.
- Preserves the previous available observation
  
Disadvantages of Forward Fill

- Can create repeated values.
- May introduce bias if values change rapidly.
- Not suitable for every dataset.
- Cannot fill leading missing values when no previous value exists.

### Backward Fill

Backward Fill is a missing-value handling technique in data preprocessing. It replaces missing values (NaN) with the next available value in the same column

It is commonly used in time-series datasets, sales records, stock prices, weather data, and sensor readings

1. Why Do We Use Backward Fill?
1. To handle missing values.
2. To maintain the original dataset size.
3. To fill missing values using the next available observation.
4. To handle missing data in sequential datasets.

##### Mathematical Formula

Backward Fill replaces a missing value with the next available value.

$$
X_t = X_{t+1}
$$

Where:

- $X_t$ = Current missing value.
- $X_{t+1}$ = Next available value.

##### Example

Suppose a company records daily sales:

`1000, NaN, NaN, 2000, 2500, NaN`

Apply Backward Fill.

**Before Backward Fill:**

`1000, NaN, NaN, 2000, 2500, NaN`

Replace each missing value with the next available value.

**After Backward Fill:**

`1000, 2000, 2000, 2000, 2500, NaN`

#### Backward Fill Process

**Next Available Value:**

$$
X_{t+1} = 2000
$$

**Missing Value 1:**

$$
X_t = 2000
$$

**Missing Value 2:**

$$
X_t = 2000
$$

**Final Result:**

$$
[1000,\ 2000,\ 2000,\ 2000,\ 2500,\ NaN]
$$

````python
import pandas as pd
import numpy as np

# Create sample dataset
data = {
    "Date": ["01-01-2026", "02-01-2026", "03-01-2026",
             "04-01-2026", "05-01-2026", "06-01-2026"],
    "Sales": [1000, np.nan, np.nan, 2000, 2500, np.nan]
}

df = pd.DataFrame(data)

print("Original Dataset:")
print(df)

# Apply Backward Fill
df["Sales"] = df["Sales"].bfill()

print("\nDataset After Backward Fill:")
print(df)

# Check missing values
print("\nMissing Values:")
print(df.isnull().sum())
````

### Interpolation


Interpolation is a missing-value handling technique in data preprocessing. It estimates missing values using the available data points before and after the missing value

It is commonly used in time-series datasets, sales records, stock prices, temperature measurements, and sensor data

1. Why Do We Use Interpolation?
1. To estimate missing numerical values.
2. To maintain the original dataset size.
3. To identify approximate values between known observations.
4. To preserve the trend and continuity of data.

##### Mathematical Formula

Linear Interpolation calculates a missing value using two known data points.

$$
X = X_1 + \frac{(X_2-X_1)}{(t_2-t_1)}(t-t_1)
$$

Where:

- $X$ = Estimated missing value.
- $X_1$ = First known value.
- $X_2$ = Second known value.
- $t_1$ = First known position.
- $t_2$ = Second known position.
- $t$ = Missing value position.

##### Example

Suppose a company records daily sales:

`1000, 1500, NaN, 2500, 3000`

We need to estimate the missing sales value between 1500 and 2500.

**Given:**

- $X_1 = 1500$
- $X_2 = 2500$
- $t_1 = 2$
- $t_2 = 4$
- $t = 3$

Calculate the missing value:

$$
X = 1500 + \frac{(2500-1500)}{(4-2)}(3-2)
$$

$$
X = 1500 + \frac{1000}{2}\times 1
$$

$$
X = 1500 + 500
$$

$$
\boxed{X = 2000}
$$

Replace the missing value with 2000.

##### Before Interpolation

The original sales values contain one missing value.

$$
[1000,\ 1500,\ NaN,\ 2500,\ 3000]
$$

##### Estimated Missing Value

$$
\boxed{X = 2000}
$$

##### Estimated Missing Value

$$
\boxed{X = 2000}
$$

````python
import pandas as pd
import numpy as np

# Create sample dataset
data = {
    "Day": [1, 2, 3, 4, 5],
    "Sales": [1000, 1500, np.nan, 2500, 3000]
}

df = pd.DataFrame(data)

print("Original Dataset:")
print(df)

# Apply Linear Interpolation
df["Sales"] = df["Sales"].interpolate(method="linear")

print("\nDataset After Interpolation:")
print(df)

# Check missing values
print("\nMissing Values:")
print(df.isnull().sum())
````

### Group-Based Imputation


Group-Based Imputation is a missing-value handling technique in which missing values are replaced using statistical values calculated separately for different groups in a dataset

Instead of filling all missing values with one overall mean, median, or mode, we calculate the values within each group

For example, in a customer dataset, we can fill missing salaries based on each customer's department rather than using the overall average salary

Why Do We Use Group-Based Imputation?

- Different groups may have different average values.
- It preserves group-level patterns in the dataset.
- It can provide more accurate imputed values than global mean or median.
- It is useful when numerical values depend on categories.

  Types of Group-Based Imputation:
  
1. Group-Based Mean Imputation – Replaces missing numerical values with the mean of the corresponding group.
2. Group-Based Median Imputation – Replaces missing numerical values with the median of the corresponding group.
3. Group-Based Mode Imputation – Replaces missing categorical values with the most frequent value in the corresponding group.

##### Mathematical Formula

For numerical data, group-based mean imputation is:

$$
X_{\text{missing}} = \mu_g
$$

Where:

- $X_{\text{missing}}$ = Missing value.
- $\mu_g$ = Mean of the corresponding group.

The group mean is calculated as:

$$
\mu_g = \frac{\sum_{i=1}^{n_g} X_i}{n_g}
$$

Where:

- $X_i$ = Observed values in a group.
- $n_g$ = Number of non-missing observations in that group.

##### Simple Example

Consider the following employee dataset:

| Employee | Department | Salary |
|---|---|---:|
| A | IT | 40,000 |
| B | IT | 50,000 |
| C | IT | NaN |
| D | HR | 30,000 |
| E | HR | 35,000 |
| F | HR | NaN |

Calculate the average salary separately for each department.

##### IT Department

$$
\text{Mean Salary} = \frac{40000 + 50000}{2}
$$

$$
= \frac{90000}{2}
$$

$$
= 45000
$$

The missing salary for employee C is replaced with **45,000**.

##### HR Department

$$
\text{Mean Salary} = \frac{30000 + 35000}{2}
$$

$$
= \frac{65000}{2}
$$

$$
= 32500
$$

The missing salary for employee F is replaced with **32,500**.

##### After Group-Based Imputation

| Employee | Department | Salary |
|---|---|---:|
| A | IT | 40,000 |
| B | IT | 50,000 |
| C | IT | 45,000 |
| D | HR | 30,000 |
| E | HR | 35,000 |
| F | HR | 32,500 |

**Result:** Missing salary values are replaced using the average salary of their respective departments.

````python
# Step 1: Import libraries
import pandas as pd
import numpy as np

# Step 2: Create employee dataset
data = {
    "Employee_ID": [101, 102, 103, 104, 105, 106, 107, 108],
    "Employee_Name": [
        "Ravi", "Kiran", "Arjun", "Priya",
        "Sneha", "Anil", "Divya", "Rahul"
    ],
    "Department": [
        "IT", "IT", "IT", "HR",
        "HR", "HR", "Finance", "Finance"
    ],
    "Salary": [
        40000, 50000, np.nan, 30000,
        35000, np.nan, 60000, np.nan
    ]
}

df = pd.DataFrame(data)

# Step 3: Display original dataset
print("Original Dataset:")
print(df)

# Step 4: Check missing values
print("\nMissing Values:")
print(df.isnull().sum())

# Step 5: Calculate department-wise mean salary
print("\nDepartment-wise Mean Salary:")
print(df.groupby("Department")["Salary"].mean())

# Step 6: Perform group-based mean imputation
df["Salary"] = df.groupby("Department")["Salary"].transform(
    lambda group: group.fillna(group.mean())
)

# Step 7: Display updated dataset
print("\nDataset After Group-Based Imputation:")
print(df)

# Step 8: Check missing values after imputation
print("\nMissing Values After Imputation:")
print(df.isnull().sum())

# Step 9: Display final dataset information
print("\nFinal Dataset Information:")
print(df.info())
````

### KNN Imputation

KNN Imputation is a missing-value handling technique that replaces missing values by finding the most similar records (nearest neighbors) in a dataset

KNN stands for K-Nearest Neighbors

Instead of replacing missing values with a simple mean or median, KNN identifies similar rows based on their available feature values and uses their values to estimate the missing data

Example:

Suppose a customer dataset contains missing age values. KNN finds customers with similar income, spending score, and other available features, then uses their ages to estimate the missing age

## Mathematical Formula


Euclidean distance measures the distance between two data points.

$$
d(A,B)=\sqrt{\sum_{i=1}^{n}(A_i-B_i)^2}
$$

Where:

- $d(A,B)$ = Distance between two data points.
- $A_i$ = Feature value of the first data point.
- $B_i$ = Feature value of the second data point.
- $n$ = Number of features used to calculate distance.

##### KNN Imputation Formula

For numerical data, the missing value can be estimated using the mean of the $K$ nearest neighbors:

$$
X_{\text{missing}}=\frac{1}{K}\sum_{i=1}^{K}X_i
$$

Where:

- $X_{\text{missing}}$ = Estimated missing value.
- $K$ = Number of nearest neighbors.
- $X_i$ = Corresponding observed values of the nearest neighbors.

##### Simple Example

Consider the following employee dataset:

| Employee | Age | Experience | Performance Score |
|---|---:|---:|---:|
| A | 25 | 2 | 80 |
| B | 30 | 5 | 75 |
| C | 35 | 7 | 70 |
| D | 40 | 10 | 65 |
| E | NaN | 6 | 72 |

Suppose we want to fill the missing age of employee E.

KNN identifies employees with similar experience and performance scores.

If the two nearest neighbors are B and C:

##### Estimated Age Calculation

$$
\text{Estimated Age}=\frac{30+35}{2}
$$

$$
=\frac{65}{2}
$$

$$
=32.5
$$

The missing age is estimated as **32.5 years**.

#####  Python Implementation Using Scikit-Learn

We use the `KNNImputer` class from Scikit-Learn.

**Syntax:**

```python
from sklearn.impute import KNNImputer

imputer = KNNImputer(n_neighbors=2)

df_imputed = imputer.fit_transform(df)


**Note:** Use `Esc → M → Enter` to convert a cell into Markdown and `Shift + Enter` to render it.

````python
# Step 1: Import libraries
import pandas as pd
import numpy as np
from sklearn.impute import KNNImputer

# Step 2: Create employee dataset
data = {
    "Age": [25, 30, 35, 40, np.nan],
    "Experience": [2, 5, 7, 10, 6],
    "Performance_Score": [80, 75, 70, 65, 72]
}

df = pd.DataFrame(data)

# Step 3: Display original dataset
print("Original Dataset:")
print(df)

# Step 4: Check missing values
print("\nMissing Values:")
print(df.isnull().sum())

# Step 5: Create KNN Imputer
imputer = KNNImputer(n_neighbors=2)

# Step 6: Perform KNN Imputation
imputed_data = imputer.fit_transform(df)

# Step 7: Convert result into DataFrame
df_imputed = pd.DataFrame(
    imputed_data,
    columns=df.columns
)

# Step 8: Display updated dataset
print("\nDataset After KNN Imputation:")
print(df_imputed)

# Step 9: Check missing values after imputation
print("\nMissing Values After Imputation:")
print(df_imputed.isnull().sum())
````

### Iterative Imputation

Iterative Imputation is a missing-value handling technique that estimates missing values by repeatedly predicting each incomplete feature using the other available features in the dataset

Instead of filling missing values with a simple mean or median, it uses machine learning models to predict missing values based on relationships between features

In Python, Scikit-Learn provides IterativeImputer for this technique.

Example:

Suppose an employee dataset contains:
- Age
- Experience
- Salary
- Performance Score
  
If salary is missing, Iterative Imputation can use age, experience, and performance score to predict the missing salary.

##### Mathematical Concept

Suppose a dataset contains three features:

$$
X_1,\ X_2,\ X_3
$$

If $X_1$ contains missing values, the algorithm predicts it using the other features:

$$
X_1=f(X_2,X_3)
$$

Similarly:

$$
X_2=f(X_1,X_3)
$$

$$
X_3=f(X_1,X_2)
$$

Here, $f$ represents a predictive model, such as linear regression.

The predictions are updated repeatedly until the imputed values stabilize or the iteration limit is reached.

#### Simple Example

Consider the following employee dataset:

| Employee | Age | Experience | Salary |
|---|---:|---:|---:|
| A | 25 | 2 | 30,000 |
| B | 30 | 5 | 40,000 |
| C | 35 | 7 | NaN |
| D | 40 | 10 | 60,000 |
| E | NaN | 6 | 45,000 |

We have missing values in:

- Age
- Salary

Iterative Imputation uses the relationships between age, experience, and salary to estimate the missing values.

##### How Iterative Imputation Works

For example:

1. Predict missing salary using age and experience.
2. Predict missing age using experience and salary.
3. Repeat the predictions across multiple iterations.

The final estimated values depend on the selected predictive model and its parameters.

````python
# Step 1: Import libraries
import pandas as pd
import numpy as np

from sklearn.experimental import enable_iterative_imputer
from sklearn.impute import IterativeImputer

# Step 2: Create employee dataset
data = {
    "Age": [25, 30, 35, 40, np.nan],
    "Experience": [2, 5, 7, 10, 6],
    "Salary": [30000, 40000, np.nan, 60000, 45000]
}

df = pd.DataFrame(data)

# Step 3: Display original dataset
print("Original Dataset:")
print(df)

# Step 4: Check missing values
print("\nMissing Values:")
print(df.isnull().sum())

# Step 5: Create Iterative Imputer
imputer = IterativeImputer(
    max_iter=10,
    random_state=42
)

# Step 6: Perform Iterative Imputation
imputed_data = imputer.fit_transform(df)

# Step 7: Convert result into DataFrame
df_imputed = pd.DataFrame(
    imputed_data,
    columns=df.columns
)

# Step 8: Display updated dataset
print("\nDataset After Iterative Imputation:")
print(df_imputed)

# Step 9: Check missing values
print("\nMissing Values After Imputation:")
print(df_imputed.isnull().sum())
````

Advantages of Iterative Imputation:

1. Uses relationships between multiple features.
2. Can provide better estimates than simple mean or median imputation when feature relationships are informative.
3. Supports datasets with multiple incomplete numerical columns.
4. Can capture more complex patterns depending on the prediction model.
5. Useful in statistical analysis and machine learning.
   
Disadvantages of Iterative Imputation:

1. Computationally expensive compared with simple imputation.
2. Requires more processing time.
3. Results depend on the predictive model.
4. Incorrect feature relationships can produce inaccurate estimates.
5. Sensitive to poor-quality data and extreme values.
6. May overfit when the dataset is small.
